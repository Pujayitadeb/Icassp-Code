```python
# ============================================================
# RECOVERY CELL 4 — LOAD OFFICIAL REVE + ARCHIVED REVE WEIGHTS
# ============================================================

import os
import gc
import torch
from transformers import AutoModel

print("=" * 90)
print("RECOVERY CELL 4 — OFFICIAL REVE RECOVERY")
print("=" * 90)

# ------------------------------------------------------------
# 0. Paths
# ------------------------------------------------------------

REVE_NAME = "brain-bzh/reve-base"

REVE_ARCHIVE = (
    "/content/drive/MyDrive/Seinna_REVE_Novelty3_NEW/"
    "EXPERIMENT_ARCHIVE/REVE_MODEL/"
    "reve_base_state_dict.pt"
)

REVE_CONFIG_PATH = (
    "/content/drive/MyDrive/Seinna_REVE_Novelty3_NEW/"
    "EXPERIMENT_ARCHIVE/REVE_MODEL/"
    "reve_base_config.json"
)

assert os.path.exists(REVE_ARCHIVE), (
    f"Archived REVE weights not found:\n{REVE_ARCHIVE}"
)

print("\nArchived REVE weights:")
print(REVE_ARCHIVE)

print(
    f"Size: {os.path.getsize(REVE_ARCHIVE) / (1024**2):.2f} MB"
)

# ------------------------------------------------------------
# 1. Clear unused memory
# ------------------------------------------------------------

gc.collect()

if torch.cuda.is_available():
    torch.cuda.empty_cache()

# ------------------------------------------------------------
# 2. Load official REVE architecture
# ------------------------------------------------------------

print("\n" + "-" * 90)
print("LOADING OFFICIAL REVE ARCHITECTURE")
print("-" * 90)

print("Model:", REVE_NAME)

reve_model = AutoModel.from_pretrained(
    REVE_NAME,
    trust_remote_code=True,
    dtype=torch.float32
)

print("\nOfficial REVE loaded.")
print("REVE class:", type(reve_model).__name__)

# ------------------------------------------------------------
# 3. Verify official architecture
# ------------------------------------------------------------

print("\n" + "-" * 90)
print("VERIFYING REVE ARCHITECTURE")
print("-" * 90)

assert hasattr(
    reve_model,
    "transformer"
), "REVE has no transformer."

assert hasattr(
    reve_model.transformer,
    "layers"
), "REVE transformer has no layers."

N_BLOCKS = len(
    reve_model.transformer.layers
)

print("Transformer blocks:", N_BLOCKS)

assert N_BLOCKS == 22, (
    f"Expected 22 Transformer blocks, found {N_BLOCKS}"
)

for i, block in enumerate(
    reve_model.transformer.layers
):

    assert len(block) == 2, (
        f"Block {i} does not contain "
        "Attention + FeedForward."
    )

print("22 Transformer blocks: PASS")
print("Attention + FeedForward structure: PASS")

# ------------------------------------------------------------
# 4. Load archived REVE state dict
# ------------------------------------------------------------

print("\n" + "-" * 90)
print("LOADING ARCHIVED REVE WEIGHTS")
print("-" * 90)

archived_state = torch.load(
    REVE_ARCHIVE,
    map_location="cpu",
    weights_only=False
)

print(
    "Archived object type:",
    type(archived_state)
)

# ------------------------------------------------------------
# 5. Handle possible checkpoint formats
# ------------------------------------------------------------

if isinstance(archived_state, dict):

    # Case A:
    # The file is already a direct state dict.
    direct_tensor_count = sum(
        torch.is_tensor(v)
        for v in archived_state.values()
    )

    if (
        len(archived_state) > 0
        and direct_tensor_count == len(archived_state)
    ):

        reve_state = archived_state

        print(
            "\nDetected format:"
            " DIRECT REVE STATE DICT"
        )

    # Case B:
    # The file is a complete checkpoint.
    elif "model_state_dict" in archived_state:

        checkpoint_state = (
            archived_state["model_state_dict"]
        )

        reve_state = {}

        for key, value in checkpoint_state.items():

            if key.startswith("reve."):

                clean_key = key[
                    len("reve.") :
                ]

                reve_state[clean_key] = value

        print(
            "\nDetected format:"
            " COMPLETE CHECKPOINT"
        )

    else:

        # Try common nested state-dict names.
        possible_keys = [
            "state_dict",
            "model",
            "reve_state_dict"
        ]

        found_key = None

        for key in possible_keys:

            if (
                key in archived_state
                and isinstance(
                    archived_state[key],
                    dict
                )
            ):

                found_key = key
                break

        assert found_key is not None, (
            "Could not identify a REVE state dict "
            "inside the archived checkpoint."
        )

        reve_state = archived_state[
            found_key
        ]

        print(
            "\nDetected nested state dict:",
            found_key
        )

else:

    raise TypeError(
        "Archived REVE checkpoint is not a "
        "dictionary/state dict."
    )

# ------------------------------------------------------------
# 6. Clean possible 'reve.' prefixes
# ------------------------------------------------------------

clean_reve_state = {}

for key, value in reve_state.items():

    if key.startswith("reve."):

        key = key[len("reve."):]

    clean_reve_state[key] = value

reve_state = clean_reve_state

print(
    "\nREVE parameters found:",
    len(reve_state)
)

assert len(reve_state) > 0, (
    "REVE state dict is empty."
)

# ------------------------------------------------------------
# 7. Load weights into official REVE
# ------------------------------------------------------------

print("\n" + "-" * 90)
print("LOADING WEIGHTS INTO OFFICIAL REVE")
print("-" * 90)

load_result = (
    reve_model.load_state_dict(
        reve_state,
        strict=False
    )
)

print(
    "Missing keys:",
    len(load_result.missing_keys)
)

print(
    "Unexpected keys:",
    len(load_result.unexpected_keys)
)

if load_result.missing_keys:

    print("\nMissing keys:")
    for key in load_result.missing_keys:
        print("  ", key)

if load_result.unexpected_keys:

    print("\nUnexpected keys:")
    for key in load_result.unexpected_keys:
        print("  ", key)

assert len(
    load_result.missing_keys
) == 0, (
    "Some official REVE parameters were not loaded."
)

# ------------------------------------------------------------
# 8. Freeze all REVE parameters
# ------------------------------------------------------------

print("\n" + "-" * 90)
print("FREEZING REVE")
print("-" * 90)

for parameter in reve_model.parameters():

    parameter.requires_grad = False

reve_model = reve_model.to(
    DEVICE
)

reve_model.eval()

# ------------------------------------------------------------
# 9. Parameter verification
# ------------------------------------------------------------

total_params = sum(
    parameter.numel()
    for parameter in reve_model.parameters()
)

trainable_params = sum(
    parameter.numel()
    for parameter in reve_model.parameters()
    if parameter.requires_grad
)

print("\nTotal REVE parameters:")
print(total_params)

print("\nTrainable REVE parameters:")
print(trainable_params)

assert trainable_params == 0

# ------------------------------------------------------------
# 10. Final verification
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("RECOVERY CELL 4 COMPLETE")
print("=" * 90)

print("\nPASS: Official REVE architecture loaded")
print("PASS: 22 Transformer blocks verified")
print("PASS: Archived REVE weights loaded")
print("PASS: No missing REVE keys")
print("PASS: All REVE parameters frozen")
print("PASS: No dataset forward pass performed")

print("\nREVE ready for gated-model recovery.")
print("=" * 90)
```

```python
# ============================================================
# RECOVERY CELL 5 — RECREATE VERIFIED 22-GATE REVE
# ============================================================

import copy
import torch
import torch.nn as nn

print("=" * 90)
print("RECOVERY CELL 5 — RECREATE VERIFIED GATED REVE")
print("=" * 90)

# ------------------------------------------------------------
# 1. Confirm official REVE exists
# ------------------------------------------------------------

assert "reve_model" in globals(), (
    "reve_model not found. Run Recovery Cell 4 first."
)

assert len(reve_model.transformer.layers) == 22

# ------------------------------------------------------------
# 2. Gated Transformer
# ------------------------------------------------------------

class GatedTransformer67(nn.Module):

    def __init__(self, original_transformer):

        super().__init__()

        original_layers = list(
            original_transformer.layers
        )

        assert len(original_layers) == 22

        # Preserve original Attention + FF pairs
        self.layers = nn.ModuleList(
            original_layers
        )

        # Exactly 22 learnable scalar gates
        self.gates = nn.Parameter(
            torch.ones(
                22,
                dtype=torch.float32
            )
        )

        # Freeze original Transformer parameters
        for layer in self.layers:
            for parameter in layer.parameters():
                parameter.requires_grad = False

    def forward(
        self,
        x,
        return_output=False
    ):

        for i, layer in enumerate(self.layers):

            attn, ff = layer

            block_input = x

            # Attention residual
            x = attn(x) + x

            # Feed-forward residual
            x = ff(x) + x

            # Complete block gate
            x = (
                block_input
                + self.gates[i]
                * (x - block_input)
            )

        return x


# ------------------------------------------------------------
# 3. Gated REVE wrapper
# ------------------------------------------------------------

class GatedREVE67(nn.Module):

    def __init__(self, original_reve):

        super().__init__()

        # Copy the complete official REVE
        self.reve = copy.deepcopy(
            original_reve
        )

        # Replace official transformer
        # with gated transformer
        self.reve.transformer = (
            GatedTransformer67(
                self.reve.transformer
            )
        )

        # Freeze everything first
        for parameter in self.reve.parameters():
            parameter.requires_grad = False

        # Only gates are trainable
        self.reve.transformer.gates.requires_grad = True

    def forward(
        self,
        eeg,
        positions
    ):

        return self.reve(
            eeg,
            positions
        )


# ------------------------------------------------------------
# 4. Create gated model
# ------------------------------------------------------------

gated_reve = GatedREVE67(
    reve_model
)

gated_reve = gated_reve.to(
    DEVICE
)

gated_reve.eval()

# ------------------------------------------------------------
# 5. Locate gate parameter
# ------------------------------------------------------------

gate_parameters = []

for name, parameter in gated_reve.named_parameters():

    if "gates" in name.lower():

        gate_parameters.append(
            (name, parameter)
        )

assert len(gate_parameters) == 1, (
    f"Expected exactly one gate tensor, "
    f"found {len(gate_parameters)}"
)

GATE_NAME, GATE_PARAMETER = (
    gate_parameters[0]
)

# ------------------------------------------------------------
# 6. Verification
# ------------------------------------------------------------

total_trainable = sum(
    parameter.numel()
    for parameter in gated_reve.parameters()
    if parameter.requires_grad
)

print("\nGated REVE:")
print(
    "Class:",
    type(gated_reve).__name__
)

print(
    "Gate parameter:",
    GATE_NAME
)

print(
    "Gate shape:",
    tuple(GATE_PARAMETER.shape)
)

print(
    "Gate mean:",
    GATE_PARAMETER.mean().item()
)

print(
    "Gate min:",
    GATE_PARAMETER.min().item()
)

print(
    "Gate max:",
    GATE_PARAMETER.max().item()
)

print(
    "Trainable parameters:",
    total_trainable
)

assert GATE_PARAMETER.shape == (22,)
assert total_trainable == 22

# Verify all non-gate parameters are frozen
for name, parameter in gated_reve.named_parameters():

    if "gates" not in name.lower():

        assert not parameter.requires_grad, (
            f"Non-gate parameter is trainable: {name}"
        )

print("\nPASS: 22 gates created")
print("PASS: Gate initialization = 1.0")
print("PASS: Exactly 22 trainable parameters")
print("PASS: All original REVE parameters frozen")
print("PASS: Complete Transformer blocks preserved")
print("PASS: Gate = 0 corresponds to identity block")
print("PASS: Gated REVE ready")

print("\n" + "=" * 90)
print("RECOVERY CELL 5 COMPLETE")
print("=" * 90)
```

```python
# ============================================================
# RECOVERY CELL 6 — LOAD REVE POSITION MODEL
# ============================================================

import os
import torch
from transformers import AutoModel

print("=" * 90)
print("RECOVERY CELL 6 — REVE POSITION MODEL")
print("=" * 90)

POSITION_MODEL_NAME = "brain-bzh/reve-positions"

POSITION_ARCHIVE = (
    "/content/drive/MyDrive/"
    "Seinna_REVE_Novelty3_NEW/"
    "EXPERIMENT_ARCHIVE/"
    "REVE_POSITION_MODEL/"
    "reve_positions_state_dict.pt"
)

# ------------------------------------------------------------
# 1. Verify archived position weights
# ------------------------------------------------------------

assert os.path.exists(
    POSITION_ARCHIVE
), (
    "Position checkpoint not found:\n"
    + POSITION_ARCHIVE
)

print("\nArchived position weights:")
print(POSITION_ARCHIVE)

# ------------------------------------------------------------
# 2. Load official position architecture
# ------------------------------------------------------------

position_model = AutoModel.from_pretrained(
    POSITION_MODEL_NAME,
    trust_remote_code=True,
    dtype=torch.float32
)

print(
    "\nPosition model loaded:",
    type(position_model).__name__
)

# ------------------------------------------------------------
# 3. Load archived position weights
# ------------------------------------------------------------

position_state = torch.load(
    POSITION_ARCHIVE,
    map_location="cpu",
    weights_only=False
)

print(
    "Position archive type:",
    type(position_state)
)

assert isinstance(
    position_state,
    dict
)

# Direct state dict
position_tensor_count = sum(
    torch.is_tensor(v)
    for v in position_state.values()
)

if (
    len(position_state) > 0
    and position_tensor_count == len(position_state)
):

    clean_position_state = position_state

else:

    if "state_dict" in position_state:

        clean_position_state = (
            position_state["state_dict"]
        )

    elif "model_state_dict" in position_state:

        clean_position_state = (
            position_state["model_state_dict"]
        )

    else:

        raise ValueError(
            "Could not identify position-model state dict."
        )

# Remove possible prefix
final_position_state = {}

for key, value in clean_position_state.items():

    if key.startswith("position_model."):

        key = key[
            len("position_model.") :
        ]

    final_position_state[key] = value

# ------------------------------------------------------------
# 4. Load position weights
# ------------------------------------------------------------

position_load = (
    position_model.load_state_dict(
        final_position_state,
        strict=False
    )
)

print(
    "\nMissing keys:",
    len(position_load.missing_keys)
)

print(
    "Unexpected keys:",
    len(position_load.unexpected_keys)
)

if position_load.missing_keys:

    print("\nMissing:")
    for key in position_load.missing_keys:
        print(" ", key)

if position_load.unexpected_keys:

    print("\nUnexpected:")
    for key in position_load.unexpected_keys:
        print(" ", key)

assert len(
    position_load.missing_keys
) == 0

# ------------------------------------------------------------
# 5. Freeze position model
# ------------------------------------------------------------

for parameter in position_model.parameters():

    parameter.requires_grad = False

position_model = position_model.to(
    DEVICE
)

position_model.eval()

# ------------------------------------------------------------
# 6. Define Siena montage
# ------------------------------------------------------------

EEG_CHANNELS_22 = [
    "Fp1",
    "F3",
    "C3",
    "P3",
    "O1",
    "F7",
    "T5",
    "Fc1",
    "Fc5",
    "Cp1",
    "Cp5",
    "F9",
    "Fz",
    "Pz",
    "F4",
    "C4",
    "P4",
    "O2",
    "F8",
    "T4",
    "T6",
    "Fc2"
]

assert len(
    EEG_CHANNELS_22
) == 22

# ------------------------------------------------------------
# 7. Generate positions
# ------------------------------------------------------------

with torch.no_grad():

    pos = position_model(
        EEG_CHANNELS_22
    )

# Convert if necessary
if not torch.is_tensor(pos):

    if hasattr(pos, "last_hidden_state"):

        pos = pos.last_hidden_state

    elif hasattr(pos, "pooler_output"):

        pos = pos.pooler_output

    else:

        raise TypeError(
            "Unknown position-model output type."
        )

if pos.ndim == 2:

    pos = pos.unsqueeze(0)

assert pos.ndim == 3

assert pos.shape[1] == 22

pos = pos.to(
    DEVICE,
    dtype=torch.float32
)

# ------------------------------------------------------------
# Final verification
# ------------------------------------------------------------

print("\nPosition shape:")
print(tuple(pos.shape))

print(
    "Position dtype:",
    pos.dtype
)

print(
    "Position device:",
    pos.device
)

print(
    "Position model trainable params:",
    sum(
        p.numel()
        for p in position_model.parameters()
        if p.requires_grad
    )
)

assert pos.shape[1] == 22

print("\nPASS: Official position model loaded")
print("PASS: Position weights loaded")
print("PASS: 22-channel montage verified")
print("PASS: Position tensor generated")
print("PASS: Position model frozen")

print("\n" + "=" * 90)
print("RECOVERY CELL 6 COMPLETE")
print("=" * 90)
```

```python
# ============================================================
# RECOVERY CELL 7 — FINAL GATED REVE FORWARD VERIFICATION
# ============================================================

import os
import numpy as np
import torch

print("=" * 90)
print("RECOVERY CELL 7 — FINAL GATED REVE FORWARD VERIFICATION")
print("=" * 90)

# ------------------------------------------------------------
# 1. Confirm required objects
# ------------------------------------------------------------

assert "gated_reve" in globals(), (
    "gated_reve not found. Run Recovery Cell 5."
)

assert "GATE_PARAMETER" in globals(), (
    "GATE_PARAMETER not found. Run Recovery Cell 5."
)

assert "pos" in globals(), (
    "pos not found. Run Recovery Cell 6."
)

assert "RAW_DIR" in globals(), (
    "RAW_DIR not found. Run Recovery Cell 1."
)

# ------------------------------------------------------------
# 2. Load one real Siena epoch
# ------------------------------------------------------------

TEST_SUBJECT = "PN00"

TEST_FILE = os.path.join(
    RAW_DIR,
    f"{TEST_SUBJECT}_epochs.npz"
)

assert os.path.exists(TEST_FILE), (
    f"Test subject file not found:\n{TEST_FILE}"
)

data = np.load(TEST_FILE)

X = data["X"]

print("\nSubject:", TEST_SUBJECT)
print("Original saved shape:", X.shape)

# Saved format:
# (N, 1000, 22)
assert X.ndim == 3
assert X.shape[1] == 1000
assert X.shape[2] == 22

# Take ONE epoch
eeg = X[0]

# Convert:
# (1000,22) -> (22,1000)
eeg = np.transpose(
    eeg,
    (1, 0)
).astype(np.float32)

eeg = torch.from_numpy(
    eeg
).unsqueeze(0).to(DEVICE)

print("Model input shape:", tuple(eeg.shape))

assert eeg.shape == (
    1,
    22,
    1000
)

# ------------------------------------------------------------
# 3. Prepare positions
# ------------------------------------------------------------

pos_device = pos.to(
    DEVICE,
    dtype=torch.float32
)

if pos_device.shape[0] == 1:

    pos_batch = pos_device

else:

    pos_batch = pos_device.unsqueeze(0)

print(
    "Position shape:",
    tuple(pos_batch.shape)
)

assert pos_batch.shape == (
    1,
    22,
    3
)

# ------------------------------------------------------------
# 4. Verify gates before forward
# ------------------------------------------------------------

gates_before = (
    GATE_PARAMETER
    .detach()
    .cpu()
    .clone()
)

print("\nGate shape:", tuple(gates_before.shape))
print(
    "Gate mean:",
    f"{gates_before.mean().item():.6f}"
)
print(
    "Gate min:",
    f"{gates_before.min().item():.6f}"
)
print(
    "Gate max:",
    f"{gates_before.max().item():.6f}"
)

assert gates_before.shape == (22,)
assert torch.isfinite(gates_before).all()

# ------------------------------------------------------------
# 5. Real forward pass
# ------------------------------------------------------------

print("\nRunning one real gated REVE forward pass...")

gated_reve.eval()

with torch.no_grad():

    output = gated_reve(
        eeg,
        pos_batch
    )

# ------------------------------------------------------------
# 6. Verify output
# ------------------------------------------------------------

print("\nOutput shape:", tuple(output.shape))
print("Output dtype:", output.dtype)
print("Output device:", output.device)

assert output.ndim == 4

assert output.shape[0] == 1
assert output.shape[1] == 22
assert output.shape[2] == 5
assert output.shape[3] == 512

assert torch.isfinite(
    output
).all()

# ------------------------------------------------------------
# 7. Verify gates were not modified
# ------------------------------------------------------------

gates_after = (
    GATE_PARAMETER
    .detach()
    .cpu()
)

gate_forward_change = (
    gates_after - gates_before
).abs().max().item()

print(
    "\nMaximum gate change during "
    f"forward verification: {gate_forward_change:.10f}"
)

assert gate_forward_change == 0.0

# ------------------------------------------------------------
# 8. Final checks
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("RECOVERY CELL 7 RESULTS")
print("=" * 90)

print("PASS: Real Siena PN00 epoch loaded")
print("PASS: Input shape = (1, 22, 1000)")
print("PASS: Position shape = (1, 22, 3)")
print("PASS: Gated REVE forward successful")
print("PASS: Output shape = (1, 22, 5, 512)")
print("PASS: Output contains no NaN/Inf")
print("PASS: 22 gates preserved")
print("PASS: Forward pass did not modify gates")

print("\n" + "=" * 90)
print("RECOVERY CELL 7 COMPLETE")
print("=" * 90)
```

```python
# ============================================================
# RECOVERY CELL 8A — AUDIT ALL GATE-TRAINING CHECKPOINTS
# ============================================================
#
# PURPOSE:
#   Inspect every checkpoint_latest.pt in the 60 run folders.
#
#   This cell:
#     - DOES NOT load REVE
#     - DOES NOT load Siena EEG
#     - DOES NOT run GPU computation
#     - DOES NOT modify any checkpoint
#     - DOES NOT modify any training run
#
#   It only reads checkpoint metadata and reports:
#     - checkpoint existence
#     - checkpoint keys
#     - checkpoint epoch
#     - available gate information
#     - available classifier information
#     - checkpoint format
#
# ============================================================

import os
import json
import numpy as np
import pandas as pd
import torch


print("=" * 100)
print("RECOVERY CELL 8A — AUDIT ALL GATE-TRAINING CHECKPOINTS")
print("=" * 100)


# ============================================================
# 1. PATHS
# ============================================================

BASE_DIR = "/content/drive/MyDrive/Seinna_REVE_LOSO"

GATE_DIR = os.path.join(
    BASE_DIR,
    "novelty3_pruning",
    "gate_training"
)

RUN_DIR = os.path.join(
    GATE_DIR,
    "runs"
)

MANIFEST_PATH = os.path.join(
    GATE_DIR,
    "gate_training_60_runs.json"
)


assert os.path.exists(
    RUN_DIR
), f"Run directory not found:\n{RUN_DIR}"

assert os.path.exists(
    MANIFEST_PATH
), f"Manifest not found:\n{MANIFEST_PATH}"


# ============================================================
# 2. LOAD 60-RUN MANIFEST
# ============================================================

with open(
    MANIFEST_PATH,
    "r"
) as f:

    gate_manifest = json.load(f)


if isinstance(
    gate_manifest,
    dict
) and "jobs" in gate_manifest:

    jobs = gate_manifest["jobs"]

elif isinstance(
    gate_manifest,
    list
):

    jobs = gate_manifest

else:

    raise ValueError(
        "Unexpected gate-training manifest format."
    )


assert len(jobs) == 60

print(
    "\nManifest jobs:",
    len(jobs)
)


# ============================================================
# 3. HELPER — SAFE OBJECT DESCRIPTION
# ============================================================

def describe_object(
    name,
    obj
):

    information = {
        "name": name,
        "python_type": type(obj).__name__,
        "is_tensor": False,
        "shape": "",
        "dtype": "",
        "numel": "",
    }

    if torch.is_tensor(obj):

        information["is_tensor"] = True

        information["shape"] = str(
            tuple(
                obj.shape
            )
        )

        information["dtype"] = str(
            obj.dtype
        )

        information["numel"] = int(
            obj.numel()
        )

    elif isinstance(
        obj,
        np.ndarray
    ):

        information["shape"] = str(
            obj.shape
        )

        information["dtype"] = str(
            obj.dtype
        )

        information["numel"] = int(
            obj.size
        )

    return information


# ============================================================
# 4. AUDIT STORAGE
# ============================================================

audit_rows = []

checkpoint_details = {}


# ============================================================
# 5. INSPECT EVERY RUN
# ============================================================

for index, job in enumerate(
    jobs,
    start=1
):

    run_name = job[
        "run_name"
    ]

    run_dir = job[
        "run_dir"
    ]


    if not os.path.isabs(
        run_dir
    ):

        run_dir = os.path.join(
            BASE_DIR,
            run_dir
        )


    checkpoint_path = os.path.join(
        run_dir,
        "checkpoint_latest.pt"
    )

    completed_path = os.path.join(
        run_dir,
        "completed.json"
    )

    best_checkpoint_path = os.path.join(
        run_dir,
        "best_checkpoint.pt"
    )

    final_gates_path = os.path.join(
        run_dir,
        "final_gates.npy"
    )


    row = {

        "index":
            index,

        "run_name":
            run_name,

        "fold":
            job.get(
                "fold",
                ""
            ),

        "seed":
            job.get(
                "seed",
                ""
            ),

        "run_dir":
            run_dir,

        "checkpoint_exists":
            os.path.exists(
                checkpoint_path
            ),

        "completed_exists":
            os.path.exists(
                completed_path
            ),

        "best_checkpoint_exists":
            os.path.exists(
                best_checkpoint_path
            ),

        "final_gates_exists":
            os.path.exists(
                final_gates_path
            ),

        "checkpoint_format":
            "",

        "checkpoint_epoch":
            "",

        "checkpoint_keys":
            "",

        "has_gate_state_dict":
            False,

        "has_gates":
            False,

        "has_classifier_state_dict":
            False,

        "has_classifier":
            False,

        "has_optimizer":
            False,

        "has_scaler":
            False,

        "has_best_gate_state":
            False,

        "has_best_classifier_state":
            False,

        "has_history":
            False,

        "gate_shape":
            "",

        "gate_dtype":
            "",

        "gate_mean":
            "",

        "gate_min":
            "",

        "gate_max":
            "",

        "error":
            "",
    }


    # --------------------------------------------------------
    # No checkpoint
    # --------------------------------------------------------

    if not os.path.exists(
        checkpoint_path
    ):

        row[
            "checkpoint_format"
        ] = "MISSING"

        audit_rows.append(
            row
        )

        continue


    # --------------------------------------------------------
    # Load checkpoint CPU only
    # --------------------------------------------------------

    try:

        checkpoint = torch.load(

            checkpoint_path,

            map_location="cpu",

            weights_only=False
        )


        checkpoint_details[
            run_name
        ] = checkpoint


        # ----------------------------------------------------
        # Identify top-level type
        # ----------------------------------------------------

        if isinstance(
            checkpoint,
            dict
        ):

            keys = list(
                checkpoint.keys()
            )

            row[
                "checkpoint_keys"
            ] = " | ".join(
                str(k)
                for k in keys
            )

        else:

            row[
                "checkpoint_format"
            ] = (
                "NON_DICT_"
                +
                type(
                    checkpoint
                ).__name__
            )

            audit_rows.append(
                row
            )

            continue


        # ----------------------------------------------------
        # Epoch
        # ----------------------------------------------------

        if "epoch" in checkpoint:

            try:

                row[
                    "checkpoint_epoch"
                ] = int(
                    checkpoint[
                        "epoch"
                    ]
                )

            except Exception:

                row[
                    "checkpoint_epoch"
                ] = str(
                    checkpoint[
                        "epoch"
                    ]
                )


        # ----------------------------------------------------
        # Detect gate formats
        # ----------------------------------------------------

        if (
            "gate_state_dict"
            in checkpoint
        ):

            row[
                "has_gate_state_dict"
            ] = True

            row[
                "checkpoint_format"
            ] = (
                "CURRENT_FORMAT"
            )


            gate_state = checkpoint[
                "gate_state_dict"
            ]


            if isinstance(
                gate_state,
                dict
            ):

                if "gates" in gate_state:

                    row[
                        "has_gates"
                    ] = True

                    gate_obj = (
                        gate_state[
                            "gates"
                        ]
                    )

                    info = describe_object(
                        "gates",
                        gate_obj
                    )

                    row[
                        "gate_shape"
                    ] = info[
                        "shape"
                    ]

                    row[
                        "gate_dtype"
                    ] = info[
                        "dtype"
                    ]

                    if (
                        torch.is_tensor(
                            gate_obj
                        )
                        or
                        isinstance(
                            gate_obj,
                            np.ndarray
                        )
                    ):

                        gate_array = (
                            gate_obj
                            .detach()
                            .cpu()
                            .numpy()
                            if torch.is_tensor(
                                gate_obj
                            )
                            else
                            np.asarray(
                                gate_obj
                            )
                        )

                        if (
                            gate_array.size
                            > 0
                        ):

                            row[
                                "gate_mean"
                            ] = float(
                                gate_array.mean()
                            )

                            row[
                                "gate_min"
                            ] = float(
                                gate_array.min()
                            )

                            row[
                                "gate_max"
                            ] = float(
                                gate_array.max()
                            )


        # ----------------------------------------------------
        # Older / alternate gate keys
        # ----------------------------------------------------

        possible_gate_keys = [

            "gates",

            "gate",

            "gate_values",

            "final_gates",

            "GATE_PARAMETER",

            "gate_parameter",

        ]


        for key in possible_gate_keys:

            if key not in checkpoint:

                continue


            row[
                "has_gates"
            ] = True


            if (
                row[
                    "checkpoint_format"
                ]
                == ""
            ):

                row[
                    "checkpoint_format"
                ] = (
                    "ALTERNATE_GATE_FORMAT"
                )


            gate_obj = checkpoint[
                key
            ]


            info = describe_object(
                key,
                gate_obj
            )


            if info[
                "shape"
            ]:

                row[
                    "gate_shape"
                ] = info[
                    "shape"
                ]

            if info[
                "dtype"
            ]:

                row[
                    "gate_dtype"
                ] = info[
                    "dtype"
                ]


            if (
                torch.is_tensor(
                    gate_obj
                )
                or
                isinstance(
                    gate_obj,
                    np.ndarray
                )
            ):

                gate_array = (
                    gate_obj
                    .detach()
                    .cpu()
                    .numpy()
                    if torch.is_tensor(
                        gate_obj
                    )
                    else
                    np.asarray(
                        gate_obj
                    )
                )


                if (
                    gate_array.size
                    > 0
                ):

                    row[
                        "gate_mean"
                    ] = float(
                        gate_array.mean()
                    )

                    row[
                        "gate_min"
                    ] = float(
                        gate_array.min()
                    )

                    row[
                        "gate_max"
                    ] = float(
                        gate_array.max()
                    )


        # ----------------------------------------------------
        # Classifier formats
        # ----------------------------------------------------

        if (
            "classifier_state_dict"
            in checkpoint
        ):

            row[
                "has_classifier_state_dict"
            ] = True


        if (
            "classifier"
            in checkpoint
        ):

            row[
                "has_classifier"
            ] = True


        # ----------------------------------------------------
        # Optimizer / scaler
        # ----------------------------------------------------

        if (
            "optimizer_state_dict"
            in checkpoint
        ):

            row[
                "has_optimizer"
            ] = True


        if (
            "optimizer"
            in checkpoint
        ):

            row[
                "has_optimizer"
            ] = True


        if (
            "scaler_state_dict"
            in checkpoint
        ):

            row[
                "has_scaler"
            ] = True


        if (
            "scaler"
            in checkpoint
        ):

            row[
                "has_scaler"
            ] = True


        # ----------------------------------------------------
        # Best states
        # ----------------------------------------------------

        if (
            "best_gate_state"
            in checkpoint
        ):

            row[
                "has_best_gate_state"
            ] = True


        if (
            "best_classifier_state"
            in checkpoint
        ):

            row[
                "has_best_classifier_state"
            ] = True


        # ----------------------------------------------------
        # History
        # ----------------------------------------------------

        if (
            "history"
            in checkpoint
        ):

            row[
                "has_history"
            ] = True


        # ----------------------------------------------------
        # Unknown format
        # ----------------------------------------------------

        if (
            row[
                "checkpoint_format"
            ]
            == ""
        ):

            row[
                "checkpoint_format"
            ] = (
                "UNKNOWN_DICT_FORMAT"
            )


    except Exception as e:

        row[
            "checkpoint_format"
        ] = "LOAD_ERROR"

        row[
            "error"
        ] = (
            type(e).__name__
            +
            ": "
            +
            str(e)
        )


    audit_rows.append(
        row
    )


# ============================================================
# 6. CREATE AUDIT TABLE
# ============================================================

audit_df = pd.DataFrame(
    audit_rows
)


AUDIT_PATH = os.path.join(
    GATE_DIR,
    "checkpoint_audit_60_runs.csv"
)


audit_df.to_csv(
    AUDIT_PATH,
    index=False
)


# ============================================================
# 7. PRINT COMPLETE SUMMARY
# ============================================================

print("\n")
print("=" * 100)
print("CHECKPOINT AUDIT SUMMARY")
print("=" * 100)


print(
    "\nTotal runs:",
    len(audit_df)
)


print(
    "Checkpoint files found:",
    int(
        audit_df[
            "checkpoint_exists"
        ].sum()
    )
)


print(
    "Completed markers:",
    int(
        audit_df[
            "completed_exists"
        ].sum()
    )
)


print(
    "Final gate files:",
    int(
        audit_df[
            "final_gates_exists"
        ].sum()
    )
)


# ============================================================
# 8. FORMAT COUNTS
# ============================================================

print("\nCheckpoint formats:")

format_counts = (
    audit_df[
        "checkpoint_format"
    ]
    .value_counts(
        dropna=False
    )
)


for fmt, count in (
    format_counts.items()
):

    print(
        f"  {fmt}: {count}"
    )


# ============================================================
# 9. CHECK GATE SHAPES
# ============================================================

print("\nGate shapes:")

gate_shape_counts = (
    audit_df[
        audit_df[
            "has_gates"
        ]
        ==
        True
    ][
        "gate_shape"
    ]
    .value_counts(
        dropna=False
    )
)


for shape, count in (
    gate_shape_counts.items()
):

    print(
        f"  {shape}: {count}"
    )


# ============================================================
# 10. CHECK PROBLEMATIC RUNS
# ============================================================

print("\n")
print("=" * 100)
print("RUNS REQUIRING ATTENTION")
print("=" * 100)


problem_mask = (

    (
        audit_df[
            "checkpoint_exists"
        ]
        ==
        False
    )

    |

    (
        audit_df[
            "checkpoint_format"
        ].isin(
            [
                "LOAD_ERROR",
                "UNKNOWN_DICT_FORMAT",
                "MISSING",
                "NON_DICT_OrderedDict",
                "NON_DICT_dict"
            ]
        )
    )

    |

    (
        (
            audit_df[
                "has_gates"
            ]
            ==
            True
        )
        &
        (
            audit_df[
                "gate_shape"
            ]
            !=
            "(22,)"
        )
    )

)


problem_df = audit_df[
    problem_mask
]


if len(
    problem_df
) == 0:

    print(
        "NONE"
    )

else:

    print(
        problem_df[
            [
                "index",
                "run_name",
                "fold",
                "seed",
                "checkpoint_exists",
                "checkpoint_format",
                "checkpoint_epoch",
                "gate_shape",
                "has_classifier_state_dict",
                "has_optimizer",
                "has_scaler",
                "error",
            ]
        ].to_string(
            index=False
        )
    )


# ============================================================
# 11. SHOW ALL RUNS COMPACTLY
# ============================================================

print("\n")
print("=" * 100)
print("ALL 60 CHECKPOINTS")
print("=" * 100)


display_columns = [

    "index",

    "run_name",

    "checkpoint_exists",

    "completed_exists",

    "checkpoint_format",

    "checkpoint_epoch",

    "gate_shape",

    "gate_mean",

    "gate_min",

    "gate_max",

    "has_classifier_state_dict",

    "has_optimizer",

    "has_scaler",

]


print(
    audit_df[
        display_columns
    ].to_string(
        index=False
    )
)


# ============================================================
# 12. SAVE RAW CHECKPOINT KEY REPORT
# ============================================================

KEY_REPORT_PATH = os.path.join(
    GATE_DIR,
    "checkpoint_key_report_60_runs.txt"
)


with open(
    KEY_REPORT_PATH,
    "w"
) as f:

    for row in audit_rows:

        f.write(
            "\n"
            +
            "=" * 90
            +
            "\n"
        )

        f.write(
            f"Run: {row['run_name']}\n"
        )

        f.write(
            f"Checkpoint: "
            f"{row['checkpoint_exists']}\n"
        )

        f.write(
            f"Format: "
            f"{row['checkpoint_format']}\n"
        )

        f.write(
            f"Epoch: "
            f"{row['checkpoint_epoch']}\n"
        )

        f.write(
            f"Keys: "
            f"{row['checkpoint_keys']}\n"
        )

        f.write(
            f"Gate shape: "
            f"{row['gate_shape']}\n"
        )

        f.write(
            f"Error: "
            f"{row['error']}\n"
        )


# ============================================================
# 13. FINAL
# ============================================================

print("\n")
print("=" * 100)
print("RECOVERY CELL 8A COMPLETE")
print("=" * 100)

print(
    "\nAudit CSV:"
)

print(
    AUDIT_PATH
)

print(
    "\nKey report:"
)

print(
    KEY_REPORT_PATH
)

print(
    "\nPASS: All 60 run directories inspected"
)

print(
    "PASS: No training performed"
)

print(
    "PASS: No Siena data loaded"
)

print(
    "PASS: No checkpoints modified"
)

print(
    "PASS: No REVE forward pass performed"
)

print("=" * 100)
```

```python
# ============================================================
# RECOVERY CELL 8B — CONTINUE ONLY INCOMPLETE RUNS
# ============================================================
#
# IMPORTANT:
#   Existing completed runs 1–48 are NEVER rerun.
#
#   The checkpoint_latest.pt files currently present were
#   created using an older checkpoint format and do not contain
#   gate_state_dict / classifier_state_dict.
#
#   Therefore:
#       Runs 1–48 -> PRESERVE
#       Run 49    -> START FRESH
#       Runs 50–60 -> START FRESH
#
#   This cell does NOT attempt to resume the incompatible
#   old checkpoints.
#
# ============================================================

import os
import json
import time
import random
import copy

import numpy as np
import pandas as pd

import torch
import torch.nn as nn

from torch.utils.data import (
    DataLoader,
    TensorDataset
)


print("=" * 100)
print("RECOVERY CELL 8B — CONTINUE INCOMPLETE GATE RUNS")
print("=" * 100)


# ============================================================
# 1. DEVICE
# ============================================================

assert torch.cuda.is_available(), (
    "CUDA GPU is required for gate training."
)

DEVICE = torch.device("cuda")

print(
    "\nDevice:",
    DEVICE
)


# ============================================================
# 2. PATHS
# ============================================================

BASE_DIR = "/content/drive/MyDrive/Seinna_REVE_LOSO"

RAW_DIR = os.path.join(
    BASE_DIR,
    "subject_epochs_22ch"
)

FOLD_DIR = os.path.join(
    BASE_DIR,
    "novelty3_pruning",
    "cell66_gate_folds"
)

GATE_DIR = os.path.join(
    BASE_DIR,
    "novelty3_pruning",
    "gate_training"
)

RUN_DIR = os.path.join(
    GATE_DIR,
    "runs"
)

MANIFEST_PATH = os.path.join(
    GATE_DIR,
    "gate_training_60_runs.json"
)

FOLD_MANIFEST_PATH = os.path.join(
    FOLD_DIR,
    "gate_loso_folds.json"
)

RESULTS_PATH = os.path.join(
    GATE_DIR,
    "gate_training_results.csv"
)


# ============================================================
# 3. VERIFY REQUIRED OBJECTS FROM PREVIOUS CELLS
# ============================================================

required_objects = [
    "gated_reve",
    "position_model",
    "pos",
    "GATE_PARAMETER",
]


for object_name in required_objects:

    assert object_name in globals(), (
        f"Required object '{object_name}' "
        f"is missing. Run the REVE/gating setup cells first."
    )


print(
    "\nPASS: Gated REVE object found"
)

print(
    "PASS: Position model found"
)

print(
    "PASS: Position tensor found"
)

print(
    "PASS: Gate parameter found"
)


# ============================================================
# 4. LOAD MANIFESTS
# ============================================================

with open(
    MANIFEST_PATH,
    "r"
) as f:

    gate_manifest = json.load(f)


with open(
    FOLD_MANIFEST_PATH,
    "r"
) as f:

    fold_manifest = json.load(f)


# ------------------------------------------------------------
# Gate manifest
# ------------------------------------------------------------

if isinstance(
    gate_manifest,
    dict
) and "jobs" in gate_manifest:

    jobs = gate_manifest["jobs"]

elif isinstance(
    gate_manifest,
    list
):

    jobs = gate_manifest

else:

    raise ValueError(
        "Unexpected gate-training manifest format."
    )


# ------------------------------------------------------------
# Fold manifest
# ------------------------------------------------------------

if isinstance(
    fold_manifest,
    dict
) and "folds" in fold_manifest:

    folds = fold_manifest["folds"]

elif isinstance(
    fold_manifest,
    list
):

    folds = fold_manifest

else:

    raise ValueError(
        "Unexpected LOSO fold manifest format."
    )


assert len(jobs) == 60
assert len(folds) == 12


print(
    "\nPASS: 60 gate-training jobs loaded"
)

print(
    "PASS: 12 LOSO folds loaded"
)


# ============================================================
# 5. HYPERPARAMETERS — LOCKED
# ============================================================

MAX_EPOCHS = 20

PATIENCE = 5

BATCH_SIZE = 64

LEARNING_RATE = 1e-3

WEIGHT_DECAY = 1e-4

LAMBDA_GATE = 1e-3

SEEDS = [
    20,
    25,
    30,
    35,
    40
]


print("\nLocked hyperparameters:")

print(
    "  Max epochs:",
    MAX_EPOCHS
)

print(
    "  Patience:",
    PATIENCE
)

print(
    "  Batch size:",
    BATCH_SIZE
)

print(
    "  Learning rate:",
    LEARNING_RATE
)

print(
    "  Weight decay:",
    WEIGHT_DECAY
)

print(
    "  Gate penalty:",
    LAMBDA_GATE
)


# ============================================================
# 6. SUBJECT LOADER
# ============================================================

def load_subject(
    subject
):

    path = os.path.join(
        RAW_DIR,
        f"{subject}_epochs.npz"
    )

    assert os.path.exists(
        path
    ), (
        f"Missing subject file:\n{path}"
    )


    data = np.load(
        path
    )


    X = data["X"]

    y = data["y"]


    # Original saved format:
    #
    #   (N, 1000, 22)
    #
    # REVE input:
    #
    #   (N, 22, 1000)

    assert X.ndim == 3

    assert X.shape[2] == 22

    assert X.shape[1] == 1000


    X = np.transpose(
        X,
        (0, 2, 1)
    ).astype(
        np.float32
    )

    y = y.astype(
        np.int64
    )


    return X, y


# ============================================================
# 7. EXACT REVE POOLING
# ============================================================

def pool_reve_output(
    reve_output
):

    # REVE:
    #
    # (B, 22, 5, 512)
    #
    # mean over 5 patches
    #
    # (B, 22, 512)
    #
    # channel mean + channel std
    #
    # (B, 1024)

    assert reve_output.ndim == 4

    assert reve_output.shape[1] == 22

    assert reve_output.shape[2] == 5

    assert reve_output.shape[3] == 512


    x = reve_output.mean(
        dim=2
    )


    channel_mean = x.mean(
        dim=1
    )

    channel_std = x.std(
        dim=1
    )


    features = torch.cat(
        [
            channel_mean,
            channel_std
        ],
        dim=1
    )


    assert features.shape[1] == 1024


    return features


# ============================================================
# 8. CLASSIFIER
# ============================================================

def create_classifier():

    classifier = nn.Sequential(

        nn.Linear(
            1024,
            512
        ),

        nn.ReLU(),

        nn.Dropout(
            0.2
        ),

        nn.Linear(
            512,
            128
        ),

        nn.ReLU(),

        nn.Dropout(
            0.2
        ),

        nn.Linear(
            128,
            2
        )
    )


    return classifier.to(
        DEVICE
    )


# ============================================================
# 9. CREATE FRESH GATED REVE
# ============================================================

def create_fresh_gated_reve():

    model = copy.deepcopy(
        gated_reve
    )


    # Freeze everything first.

    for parameter in model.parameters():

        parameter.requires_grad = False


    # Locate the gate tensor.

    gate_candidates = []

    for name, parameter in model.named_parameters():

        if "gates" in name.lower():

            gate_candidates.append(
                (
                    name,
                    parameter
                )
            )


    assert len(
        gate_candidates
    ) == 1, (
        "Expected exactly one gate tensor."
    )


    gate_name, gate_parameter = (
        gate_candidates[0]
    )


    assert gate_parameter.numel() == 22


    # Reset all gates to 1.

    with torch.no_grad():

        gate_parameter.fill_(
            1.0
        )


    # Only gates are trainable.

    gate_parameter.requires_grad = True


    # Verify.

    trainable_count = sum(

        parameter.numel()

        for parameter in model.parameters()

        if parameter.requires_grad
    )


    assert trainable_count == 22


    model.eval()


    return (
        model,
        gate_name,
        gate_parameter
    )


# ============================================================
# 10. SET RANDOM SEEDS
# ============================================================

def set_seed(
    seed
):

    random.seed(
        seed
    )

    np.random.seed(
        seed
    )

    torch.manual_seed(
        seed
    )

    torch.cuda.manual_seed_all(
        seed
    )


# ============================================================
# 11. COMPLETED-RUN PROTECTION
# ============================================================

completed_runs = []

incomplete_runs = []


for job in jobs:

    run_name = job[
        "run_name"
    ]


    run_dir = job[
        "run_dir"
    ]


    if not os.path.isabs(
        run_dir
    ):

        run_dir = os.path.join(
            BASE_DIR,
            run_dir
        )


    completed_path = os.path.join(
        run_dir,
        "completed.json"
    )

    final_gates_path = os.path.join(
        run_dir,
        "final_gates.npy"
    )


    if (
        os.path.exists(
            completed_path
        )
        and
        os.path.exists(
            final_gates_path
        )
    ):

        completed_runs.append(
            run_name
        )

    else:

        incomplete_runs.append(
            job
        )


print("\n")
print("=" * 100)
print("RUN STATUS BEFORE CONTINUATION")
print("=" * 100)

print(
    "\nCompleted runs:",
    len(completed_runs)
)

print(
    "Incomplete runs:",
    len(incomplete_runs)
)


print(
    "\nIncomplete runs:"
)

for job in incomplete_runs:

    print(
        f"  {job['run_name']}"
    )


assert len(
    completed_runs
) == 48, (
    "Expected exactly 48 completed runs. "
    "Stopping to protect existing results."
)


assert len(
    incomplete_runs
) == 12, (
    "Expected exactly 12 incomplete runs."
)


# ============================================================
# 12. CRITICAL SAFETY CHECK
# ============================================================

protected_runs = set(
    completed_runs
)


for job in jobs:

    if (
        job["run_name"]
        in protected_runs
    ):

        run_dir = job[
            "run_dir"
        ]

        if not os.path.isabs(
            run_dir
        ):

            run_dir = os.path.join(
                BASE_DIR,
                run_dir
            )


        assert os.path.exists(
            os.path.join(
                run_dir,
                "final_gates.npy"
            )
        )


print(
    "\nPASS: All 48 completed runs protected"
)

print(
    "PASS: Existing final_gates.npy files will not be overwritten"
)


# ============================================================
# 13. TRAIN ONLY INCOMPLETE RUNS
# ============================================================

for run_number, job in enumerate(
    incomplete_runs,
    start=1
):

    run_name = job[
        "run_name"
    ]

    fold_number = int(
        job["fold"]
    )

    seed = int(
        job["seed"]
    )


    run_dir = job[
        "run_dir"
    ]


    if not os.path.isabs(
        run_dir
    ):

        run_dir = os.path.join(
            BASE_DIR,
            run_dir
        )


    os.makedirs(
        run_dir,
        exist_ok=True
    )


    print("\n")
    print("=" * 100)

    print(
        f"[INCOMPLETE {run_number}/12] "
        f"{run_name}"
    )

    print("=" * 100)


    # --------------------------------------------------------
    # Find fold
    # --------------------------------------------------------

    fold_info = None


    for candidate_fold in folds:

        if int(
            candidate_fold["fold"]
        ) == fold_number:

            fold_info = candidate_fold

            break


    assert fold_info is not None


    train_subjects = fold_info[
        "train_subjects"
    ]

    validation_subject = fold_info[
        "validation_subject"
    ]

    test_subject = fold_info[
        "test_subject"
    ]


    print(
        "\nTrain subjects:",
        len(train_subjects)
    )

    print(
        "Validation:",
        validation_subject
    )

    print(
        "Test:",
        test_subject
    )


    # --------------------------------------------------------
    # TEST SUBJECT IS NEVER LOADED
    # --------------------------------------------------------

    assert test_subject not in train_subjects

    assert test_subject != validation_subject


    # --------------------------------------------------------
    # Seed
    # --------------------------------------------------------

    set_seed(
        seed
    )


    # --------------------------------------------------------
    # Load training subjects
    # --------------------------------------------------------

    X_train_list = []

    y_train_list = []


    for subject in train_subjects:

        X_subject, y_subject = (
            load_subject(
                subject
            )
        )


        X_train_list.append(
            X_subject
        )

        y_train_list.append(
            y_subject
        )


    X_train = np.concatenate(
        X_train_list,
        axis=0
    )

    y_train = np.concatenate(
        y_train_list,
        axis=0
    )


    # --------------------------------------------------------
    # Validation only
    # --------------------------------------------------------

    X_val, y_val = load_subject(
        validation_subject
    )


    print(
        "\nTrain:",
        X_train.shape,
        y_train.shape
    )

    print(
        "Val:",
        X_val.shape,
        y_val.shape
    )


    # --------------------------------------------------------
    # Test subject protection
    # --------------------------------------------------------

    assert (
        X_train.shape[0]
        +
        X_val.shape[0]
        <=
        1007
    )


    # --------------------------------------------------------
    # Tensor datasets
    # --------------------------------------------------------

    train_dataset = TensorDataset(

        torch.from_numpy(
            X_train
        ),

        torch.from_numpy(
            y_train
        )
    )


    val_dataset = TensorDataset(

        torch.from_numpy(
            X_val
        ),

        torch.from_numpy(
            y_val
        )
    )


    train_loader = DataLoader(

        train_dataset,

        batch_size=BATCH_SIZE,

        shuffle=True,

        num_workers=0,

        pin_memory=True
    )


    val_loader = DataLoader(

        val_dataset,

        batch_size=BATCH_SIZE,

        shuffle=False,

        num_workers=0,

        pin_memory=True
    )


    # --------------------------------------------------------
    # Fresh model
    # --------------------------------------------------------

    current_model, gate_name, gate_parameter = (
        create_fresh_gated_reve()
    )


    classifier = create_classifier()


    criterion = nn.CrossEntropyLoss()


    optimizer = torch.optim.AdamW(

        [
            {
                "params": [
                    gate_parameter
                ]
            },

            {
                "params":
                    classifier.parameters()
            }
        ],

        lr=LEARNING_RATE,

        weight_decay=WEIGHT_DECAY
    )


    scaler = torch.amp.GradScaler(
        "cuda",
        enabled=True
    )


    # --------------------------------------------------------
    # Training
    # --------------------------------------------------------

    best_val_loss = float(
        "inf"
    )

    best_epoch = 0

    patience_counter = 0

    history = []


    start_time = time.time()


    for epoch in range(
        1,
        MAX_EPOCHS + 1
    ):


        current_model.eval()

        classifier.train()


        train_loss_total = 0.0

        train_samples = 0


        for X_batch, y_batch in train_loader:

            X_batch = X_batch.to(
                DEVICE,
                non_blocking=True
            )

            y_batch = y_batch.to(
                DEVICE,
                non_blocking=True
            )


            optimizer.zero_grad(
                set_to_none=True
            )


            with torch.amp.autocast(
                "cuda",
                enabled=True
            ):

                reve_output = current_model(
                    X_batch,
                    pos
                )


                features = pool_reve_output(
                    reve_output
                )


                logits = classifier(
                    features
                )


                classification_loss = criterion(
                    logits,
                    y_batch
                )


                gate_penalty = (
                    gate_parameter.abs().mean()
                )


                loss = (
                    classification_loss
                    +
                    LAMBDA_GATE
                    *
                    gate_penalty
                )


            scaler.scale(
                loss
            ).backward()


            scaler.step(
                optimizer
            )

            scaler.update()


            train_loss_total += (
                loss.item()
                *
                X_batch.size(0)
            )

            train_samples += (
                X_batch.size(0)
            )


        train_loss = (
            train_loss_total
            /
            train_samples
        )


        # ----------------------------------------------------
        # Validation
        # ----------------------------------------------------

        current_model.eval()

        classifier.eval()


        val_loss_total = 0.0

        val_samples = 0


        with torch.no_grad():

            for X_batch, y_batch in val_loader:

                X_batch = X_batch.to(
                    DEVICE,
                    non_blocking=True
                )

                y_batch = y_batch.to(
                    DEVICE,
                    non_blocking=True
                )


                with torch.amp.autocast(
                    "cuda",
                    enabled=True
                ):

                    reve_output = current_model(
                        X_batch,
                        pos
                    )


                    features = pool_reve_output(
                        reve_output
                    )


                    logits = classifier(
                        features
                    )


                    classification_loss = criterion(
                        logits,
                        y_batch
                    )


                    gate_penalty = (
                        gate_parameter.abs().mean()
                    )


                    val_loss_tensor = (
                        classification_loss
                        +
                        LAMBDA_GATE
                        *
                        gate_penalty
                    )


                val_loss_total += (
                    val_loss_tensor.item()
                    *
                    X_batch.size(0)
                )

                val_samples += (
                    X_batch.size(0)
                )


        val_loss = (
            val_loss_total
            /
            val_samples
        )


        # ----------------------------------------------------
        # Gate statistics
        # ----------------------------------------------------

        gate_cpu = (
            gate_parameter
            .detach()
            .float()
            .cpu()
        )


        gate_mean = (
            gate_cpu.mean().item()
        )

        gate_min = (
            gate_cpu.min().item()
        )

        gate_max = (
            gate_cpu.max().item()
        )


        # ----------------------------------------------------
        # Save history
        # ----------------------------------------------------

        history.append({

            "epoch":
                epoch,

            "train_loss":
                train_loss,

            "val_loss":
                val_loss,

            "gate_mean":
                gate_mean,

            "gate_min":
                gate_min,

            "gate_max":
                gate_max
        })


        # ----------------------------------------------------
        # Best model
        # ----------------------------------------------------

        if val_loss < best_val_loss:

            best_val_loss = val_loss

            best_epoch = epoch

            patience_counter = 0


            best_gate_state = (
                gate_parameter
                .detach()
                .cpu()
                .clone()
            )


            best_classifier_state = copy.deepcopy(
                classifier.state_dict()
            )


            print(
                f"Epoch {epoch:02d} "
                f"Train {train_loss:.5f} "
                f"Val {val_loss:.5f} "
                f"Gate {gate_mean:.4f} "
                f"BEST"
            )

        else:

            patience_counter += 1


            print(
                f"Epoch {epoch:02d} "
                f"Train {train_loss:.5f} "
                f"Val {val_loss:.5f} "
                f"Gate {gate_mean:.4f} "
                f"patience "
                f"{patience_counter}/{PATIENCE}"
            )


        # ----------------------------------------------------
        # Current checkpoint
        #
        # IMPORTANT:
        # This is the NEW checkpoint format.
        # ----------------------------------------------------

        checkpoint = {

            "checkpoint_version":
                2,

            "run_name":
                run_name,

            "fold":
                fold_number,

            "seed":
                seed,

            "epoch":
                epoch,

            "best_epoch":
                best_epoch,

            "best_val_loss":
                best_val_loss,

            "patience_counter":
                patience_counter,

            "gate_name":
                gate_name,

            "gate_state_dict":
                {
                    "gates":
                        gate_parameter
                        .detach()
                        .cpu()
                        .clone()
                },

            "classifier_state_dict":
                copy.deepcopy(
                    classifier.state_dict()
                ),

            "optimizer_state_dict":
                optimizer.state_dict(),

            "scaler_state_dict":
                scaler.state_dict(),

            "best_gate_state":
                best_gate_state,

            "best_classifier_state":
                best_classifier_state,

            "history":
                history,
        }


        torch.save(
            checkpoint,
            os.path.join(
                run_dir,
                "checkpoint_latest.pt"
            )
        )


        # ----------------------------------------------------
        # Early stopping
        # ----------------------------------------------------

        if patience_counter >= PATIENCE:

            print(
                f"Early stopping at epoch "
                f"{epoch}"
            )

            break


    # ========================================================
    # RESTORE BEST STATE
    # ========================================================

    with torch.no_grad():

        gate_parameter.copy_(
            best_gate_state.to(
                DEVICE
            )
        )


    classifier.load_state_dict(
        best_classifier_state
    )


    final_gates = (
        gate_parameter
        .detach()
        .float()
        .cpu()
        .numpy()
    )


    assert final_gates.shape == (
        22,
    )

    assert np.isfinite(
        final_gates
    ).all()


    # ========================================================
    # SAVE FINAL GATES
    # ========================================================

    np.save(
        os.path.join(
            run_dir,
            "final_gates.npy"
        ),
        final_gates
    )


    pd.DataFrame({

        "block":
            np.arange(
                1,
                23
            ),

        "gate":
            final_gates

    }).to_csv(

        os.path.join(
            run_dir,
            "final_gates.csv"
        ),

        index=False
    )


    # ========================================================
    # SAVE HISTORY
    # ========================================================

    pd.DataFrame(
        history
    ).to_csv(

        os.path.join(
            run_dir,
            "history.csv"
        ),

        index=False
    )


    # ========================================================
    # SAVE BEST CHECKPOINT
    # ========================================================

    best_checkpoint = {

        "checkpoint_version":
            2,

        "run_name":
            run_name,

        "fold":
            fold_number,

        "seed":
            seed,

        "best_epoch":
            best_epoch,

        "best_val_loss":
            best_val_loss,

        "gate_name":
            gate_name,

        "gate_state_dict":
            {
                "gates":
                    torch.tensor(
                        final_gates,
                        dtype=torch.float32
                    )
            },

        "classifier_state_dict":
            copy.deepcopy(
                classifier.state_dict()
            ),

        "best_gate_state":
            torch.tensor(
                final_gates,
                dtype=torch.float32
            ),

        "best_classifier_state":
            copy.deepcopy(
                classifier.state_dict()
            ),

        "history":
            history,
    }


    torch.save(
        best_checkpoint,
        os.path.join(
            run_dir,
            "best_checkpoint.pt"
        )
    )


    # ========================================================
    # COMPLETED MARKER
    # ========================================================

    elapsed = (
        time.time()
        -
        start_time
    )


    completed_info = {

        "run_name":
            run_name,

        "fold":
            fold_number,

        "seed":
            seed,

        "best_epoch":
            best_epoch,

        "best_val_loss":
            best_val_loss,

        "runtime_seconds":
            elapsed,

        "final_gate_mean":
            float(
                final_gates.mean()
            ),

        "final_gate_min":
            float(
                final_gates.min()
            ),

        "final_gate_max":
            float(
                final_gates.max()
            ),

        "checkpoint_version":
            2,

        "test_subject_used":
            False,
    }


    with open(
        os.path.join(
            run_dir,
            "completed.json"
        ),
        "w"
    ) as f:

        json.dump(
            completed_info,
            f,
            indent=2
        )


    print("\n")
    print(
        f"COMPLETED: {run_name}"
    )

    print(
        f"Best epoch: {best_epoch}"
    )

    print(
        f"Best validation loss: "
        f"{best_val_loss:.6f}"
    )

    print(
        f"Gate mean: "
        f"{final_gates.mean():.6f}"
    )

    print(
        f"Gate min: "
        f"{final_gates.min():.6f}"
    )

    print(
        f"Gate max: "
        f"{final_gates.max():.6f}"
    )

    print(
        f"Runtime: "
        f"{elapsed / 60:.2f} minutes"
    )


    # ========================================================
    # CLEAN MEMORY
    # ========================================================

    del current_model
    del classifier
    del optimizer
    del scaler

    del train_loader
    del val_loader

    del X_train
    del y_train
    del X_val
    del y_val

    torch.cuda.empty_cache()


# ============================================================
# 14. FINAL ACCOUNTING
# ============================================================

print("\n")
print("=" * 100)
print("FINAL GATE-RUN ACCOUNTING")
print("=" * 100)


final_completed = []

final_incomplete = []


for job in jobs:

    run_name = job[
        "run_name"
    ]

    run_dir = job[
        "run_dir"
    ]


    if not os.path.isabs(
        run_dir
    ):

        run_dir = os.path.join(
            BASE_DIR,
            run_dir
        )


    completed_path = os.path.join(
        run_dir,
        "completed.json"
    )

    final_gates_path = os.path.join(
        run_dir,
        "final_gates.npy"
    )


    if (
        os.path.exists(
            completed_path
        )
        and
        os.path.exists(
            final_gates_path
        )
    ):

        final_completed.append(
            run_name
        )

    else:

        final_incomplete.append(
            run_name
        )


print(
    "\nCompleted:",
    len(final_completed),
    "/ 60"
)

print(
    "Incomplete:",
    len(final_incomplete),
    "/ 60"
)


if final_incomplete:

    print(
        "\nStill incomplete:"
    )

    for run_name in final_incomplete:

        print(
            " ",
            run_name
        )


assert len(
    final_completed
) == 60, (
    "Not all 60 gate runs completed."
)


assert len(
    final_incomplete
) == 0


# ============================================================
# 15. BUILD 60 × 22 GATE MATRIX
# ============================================================

gate_matrix = []


for job in jobs:

    run_name = job[
        "run_name"
    ]

    run_dir = job[
        "run_dir"
    ]


    if not os.path.isabs(
        run_dir
    ):

        run_dir = os.path.join(
            BASE_DIR,
            run_dir
        )


    gate_path = os.path.join(
        run_dir,
        "final_gates.npy"
    )


    gates = np.load(
        gate_path
    )


    assert gates.shape == (
        22,
    )

    assert np.isfinite(
        gates
    ).all()


    gate_matrix.append(
        gates
    )


gate_matrix = np.stack(
    gate_matrix,
    axis=0
)


assert gate_matrix.shape == (
    60,
    22
)


GATE_MATRIX_PATH = os.path.join(
    GATE_DIR,
    "gate_matrix_60x22.npy"
)


GATE_MATRIX_CSV = os.path.join(
    GATE_DIR,
    "gate_matrix_60x22.csv"
)


np.save(
    GATE_MATRIX_PATH,
    gate_matrix
)


pd.DataFrame(

    gate_matrix,

    index=[
        job["run_name"]
        for job in jobs
    ],

    columns=[
        f"Block_{i}"
        for i in range(
            1,
            23
        )
    ]

).to_csv(
    GATE_MATRIX_CSV
)


# ============================================================
# 16. FINAL PASS
# ============================================================

print("\n")
print("=" * 100)
print("RECOVERY CELL 8B COMPLETE")
print("=" * 100)

print(
    "\nResults:"
)

print(
    "  Completed runs:",
    len(final_completed)
)

print(
    "  Gate matrix:",
    gate_matrix.shape
)

print(
    "  Gate matrix file:",
    GATE_MATRIX_PATH
)

print(
    "  Gate matrix CSV:",
    GATE_MATRIX_CSV
)

print()

print(
    "PASS: Existing 48 completed runs preserved"
)

print(
    "PASS: Only incomplete runs were trained"
)

print(
    "PASS: Run 49 restarted with current checkpoint format"
)

print(
    "PASS: Runs 50–60 trained"
)

print(
    "PASS: 60/60 gate vectors available"
)

print(
    "PASS: Gate matrix = (60, 22)"
)

print(
    "PASS: All gate values finite"
)

print(
    "PASS: No test subjects used during gate training"
)

print(
    "PASS: New checkpoints contain gate_state_dict"
)

print("=" * 100)
```

```python
# ============================================================
# CELL 9 — AGGREGATE 60 GATE-TRAINING RUNS
# ============================================================

import os
import json
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 1. Paths
# ------------------------------------------------------------

BASE_DIR = "/content/drive/MyDrive/Seinna_REVE_LOSO"

GATE_DIR = os.path.join(
    BASE_DIR,
    "novelty3_pruning",
    "gate_training"
)

RUN_DIR = os.path.join(
    GATE_DIR,
    "runs"
)

MANIFEST_PATH = os.path.join(
    GATE_DIR,
    "gate_training_60_runs.json"
)

GATE_MATRIX_PATH = os.path.join(
    GATE_DIR,
    "gate_matrix_60x22.npy"
)

GATE_MATRIX_CSV = os.path.join(
    GATE_DIR,
    "gate_matrix_60x22.csv"
)

# ------------------------------------------------------------
# 2. Load manifest
# ------------------------------------------------------------

with open(MANIFEST_PATH, "r") as f:
    gate_manifest = json.load(f)

jobs = gate_manifest["jobs"]

print("Total manifest jobs:", len(jobs))

assert len(jobs) == 60, "Expected exactly 60 gate-training runs."

# ------------------------------------------------------------
# 3. Load final gates from every run
# ------------------------------------------------------------

gate_rows = []
run_names = []

missing_runs = []

for job in jobs:

    run_name = job["run_name"]

    run_path = os.path.join(
        RUN_DIR,
        run_name
    )

    gate_path = os.path.join(
        run_path,
        "final_gates.npy"
    )

    if not os.path.exists(gate_path):
        missing_runs.append(run_name)
        continue

    gates = np.load(gate_path)

    gates = np.asarray(gates, dtype=np.float32).reshape(-1)

    if gates.shape[0] != 22:
        raise ValueError(
            f"{run_name}: expected 22 gates, "
            f"found {gates.shape[0]}"
        )

    if not np.all(np.isfinite(gates)):
        raise ValueError(
            f"{run_name}: NaN/Inf detected in gates."
        )

    gate_rows.append(gates)
    run_names.append(run_name)

# ------------------------------------------------------------
# 4. Verify all 60 runs
# ------------------------------------------------------------

print("\nGate files found:", len(gate_rows))
print("Missing runs:", len(missing_runs))

if missing_runs:
    print("\nMissing gate files:")
    for name in missing_runs:
        print("  ", name)

assert len(gate_rows) == 60, (
    f"Expected 60 completed gate runs, "
    f"but found {len(gate_rows)}."
)

# ------------------------------------------------------------
# 5. Construct 60 × 22 gate matrix
# ------------------------------------------------------------

gate_matrix = np.stack(gate_rows, axis=0)

print("\nGate matrix shape:", gate_matrix.shape)
print("Expected shape:   (60, 22)")

assert gate_matrix.shape == (60, 22)

# ------------------------------------------------------------
# 6. Save matrix
# ------------------------------------------------------------

np.save(
    GATE_MATRIX_PATH,
    gate_matrix
)

gate_columns = [
    f"B{i}"
    for i in range(1, 23)
]

gate_df = pd.DataFrame(
    gate_matrix,
    index=run_names,
    columns=gate_columns
)

gate_df.index.name = "run"

gate_df.to_csv(
    GATE_MATRIX_CSV
)

# ------------------------------------------------------------
# 7. Aggregate gate statistics
# ------------------------------------------------------------

gate_mean = np.mean(
    gate_matrix,
    axis=0
)

gate_std = np.std(
    gate_matrix,
    axis=0,
    ddof=1
)

gate_min = np.min(
    gate_matrix,
    axis=0
)

gate_max = np.max(
    gate_matrix,
    axis=0
)

gate_median = np.median(
    gate_matrix,
    axis=0
)

gate_statistics = pd.DataFrame({
    "block": np.arange(1, 23),
    "gate_mean": gate_mean,
    "gate_std": gate_std,
    "gate_median": gate_median,
    "gate_min": gate_min,
    "gate_max": gate_max
})

GATE_STATS_CSV = os.path.join(
    GATE_DIR,
    "gate_aggregate_statistics.csv"
)

gate_statistics.to_csv(
    GATE_STATS_CSV,
    index=False
)

# ------------------------------------------------------------
# 8. Print summary
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("CELL 9 COMPLETE")
print("=" * 60)

print("Runs aggregated:", gate_matrix.shape[0])
print("Blocks:", gate_matrix.shape[1])

print("\nOverall gate statistics:")
print(
    f"Mean   : {gate_matrix.mean():.6f}"
)
print(
    f"Std    : {gate_matrix.std(ddof=1):.6f}"
)
print(
    f"Min    : {gate_matrix.min():.6f}"
)
print(
    f"Max    : {gate_matrix.max():.6f}"
)

print("\nPer-block gate means:")
for block, mean_value in zip(
    gate_statistics["block"],
    gate_statistics["gate_mean"]
):
    print(
        f"B{block:02d}: {mean_value:.6f}"
    )

print("\nSaved:")
print(GATE_MATRIX_PATH)
print(GATE_MATRIX_CSV)
print(GATE_STATS_CSV)

print("\nPASS: 60 × 22 gate matrix successfully created.")
print("PASS: All gate vectors contain 22 finite values.")
```

```python
# ============================================================
# CELL 10A — VERIFY NOVELTY 1 FUNCTION SCORE
# No files are modified.
# ============================================================

import os
import numpy as np
import pandas as pd

BASE_DIR = "/content/drive/MyDrive/Seinna_REVE_LOSO"

N1_FILE = os.path.join(
    BASE_DIR,
    "novelty1_probes",
    "novelty1_statistics",
    "novelty1_all_probe_results.csv"
)

print("=" * 80)
print("NOVELTY 1 VERIFICATION")
print("=" * 80)

# ------------------------------------------------------------
# 1. Load
# ------------------------------------------------------------

assert os.path.exists(N1_FILE), (
    f"Novelty 1 file not found:\n{N1_FILE}"
)

n1 = pd.read_csv(N1_FILE)

print("\nFile:")
print(N1_FILE)

print("\nShape:")
print(n1.shape)

# ------------------------------------------------------------
# 2. Basic structure
# ------------------------------------------------------------

print("\nColumns:")
for i, col in enumerate(n1.columns, 1):
    print(f"{i:02d}. {col}")

# ------------------------------------------------------------
# 3. Functions
# ------------------------------------------------------------

n1["function_normalized"] = (
    n1["function"]
    .astype(str)
    .str.strip()
    .str.lower()
)

print("\nFunctions:")
print(
    n1["function_normalized"]
    .value_counts()
    .sort_index()
)

# ------------------------------------------------------------
# 4. Number of unique blocks
# ------------------------------------------------------------

print("\nUnique blocks:")
print(
    sorted(
        n1["block"]
        .dropna()
        .unique()
        .tolist()
    )
)

print(
    "Number of blocks:",
    n1["block"].nunique()
)

# ------------------------------------------------------------
# 5. Check quantitative columns
# ------------------------------------------------------------

candidate_columns = [
    "mean_R2",
    "mean_r2",
    "median_R2",
    "median_r2",
    "min_r2",
    "max_r2",
    "accuracy",
    "balanced_accuracy",
    "f1",
    "auroc",
    "score"
]

print("\nCandidate quantitative columns:")

for col in candidate_columns:

    if col not in n1.columns:
        print(f"  {col:20s} : NOT PRESENT")
        continue

    values = pd.to_numeric(
        n1[col],
        errors="coerce"
    )

    print(
        f"  {col:20s} : "
        f"present | "
        f"non-null={values.notna().sum()} | "
        f"min={values.min():.6f} | "
        f"max={values.max():.6f} | "
        f"mean={values.mean():.6f}"
    )

# ------------------------------------------------------------
# 6. Examine each function separately
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("FUNCTION-WISE SUMMARY")
print("=" * 80)

for function_name in sorted(
    n1["function_normalized"].unique()
):

    subset = n1[
        n1["function_normalized"] == function_name
    ]

    print(
        f"\n--- {function_name.upper()} ---"
    )

    print(
        "Rows:",
        len(subset)
    )

    print(
        "Blocks:",
        subset["block"].nunique()
    )

    for col in [
        "mean_R2",
        "mean_r2",
        "accuracy",
        "balanced_accuracy",
        "f1",
        "auroc",
        "score"
    ]:

        if col not in subset.columns:
            continue

        values = pd.to_numeric(
            subset[col],
            errors="coerce"
        ).dropna()

        if len(values) == 0:
            continue

        print(
            f"{col:20s}: "
            f"mean={values.mean():.6f}, "
            f"median={values.median():.6f}, "
            f"min={values.min():.6f}, "
            f"max={values.max():.6f}"
        )

# ------------------------------------------------------------
# 7. Check whether "score" behaves differently by function
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("SCORE COLUMN CHECK")
print("=" * 80)

if "score" in n1.columns:

    score_summary = (
        n1.groupby(
            "function_normalized"
        )["score"]
        .agg(
            ["count", "mean", "median", "min", "max"]
        )
    )

    print(score_summary)

else:

    print(
        "No 'score' column found."
    )

# ------------------------------------------------------------
# 8. Show representative rows
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("REPRESENTATIVE ROWS")
print("=" * 80)

show_columns = [
    "model",
    "block",
    "H",
    "test_subject",
    "function",
    "layer",
    "mean_R2",
    "mean_r2",
    "accuracy",
    "balanced_accuracy",
    "f1",
    "auroc",
    "score"
]

show_columns = [
    c for c in show_columns
    if c in n1.columns
]

print(
    n1[show_columns]
    .head(20)
    .to_string(index=False)
)

# ------------------------------------------------------------
# 9. Check for duplicate block/function combinations
# ------------------------------------------------------------

duplicates = (
    n1.groupby(
        ["block", "function_normalized"]
    )
    .size()
    .reset_index(name="count")
)

print("\n" + "=" * 80)
print("BLOCK × FUNCTION STRUCTURE")
print("=" * 80)

print(
    "Unique block/function combinations:",
    len(duplicates)
)

print(
    "Expected:",
    22 * 4
)

print(
    "\nRows per block/function combination:"
)

print(
    duplicates["count"]
    .value_counts()
    .sort_index()
)

# ------------------------------------------------------------
# 10. Final diagnostic
# ------------------------------------------------------------

print("\n" + "=" * 80)
print("CELL 10A COMPLETE")
print("=" * 80)

print(
    "No data modified."
)

print(
    "No Ours score calculated."
)

print(
    "Next step: choose the empirically supported "
    "Novelty 1 function representation from the output above."
)
```

```python
# ============================================================
# CELL 10B — COMPLETE NOVELTY 2 VERIFICATION
# NO DATA MODIFICATION
# ============================================================

import os
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 1. Path
# ------------------------------------------------------------

BASE_DIR = "/content/drive/MyDrive/Seinna_REVE_LOSO"

N2_FILE = os.path.join(
    BASE_DIR,
    "novelty2_results",
    "novelty2_primary_auroc_necessity.csv"
)

print("=" * 90)
print("NOVELTY 2 — COMPLETE VERIFICATION")
print("=" * 90)

# ------------------------------------------------------------
# 2. File existence
# ------------------------------------------------------------

if not os.path.exists(N2_FILE):
    raise FileNotFoundError(
        f"Novelty 2 file not found:\n{N2_FILE}"
    )

print("\nPASS: Novelty 2 file exists")
print(N2_FILE)

# ------------------------------------------------------------
# 3. Load
# ------------------------------------------------------------

n2 = pd.read_csv(N2_FILE)

print("\nShape:")
print(n2.shape)

# ------------------------------------------------------------
# 4. Exact columns
# ------------------------------------------------------------

print("\nColumns:")
for i, col in enumerate(n2.columns, 1):
    print(f"{i:02d}. {col}")

# ------------------------------------------------------------
# 5. Required columns
# ------------------------------------------------------------

required_columns = [
    "removed_block",
    "removed_layer",
    "mean_damage_auroc",
    "sd_damage_auroc",
    "median_damage_auroc",
    "wilcoxon_W_damage_auroc",
    "p_damage_auroc",
    "q_damage_auroc_FDR",
    "rank_biserial_damage_auroc",
    "ci95_low_damage_auroc",
    "ci95_high_damage_auroc",
    "significant_damage_auroc",
    "ci_excludes_zero_damage_auroc"
]

print("\nRequired-column verification:")

missing_columns = []

for col in required_columns:

    if col in n2.columns:
        print(f"PASS: {col}")
    else:
        print(f"MISSING: {col}")
        missing_columns.append(col)

if missing_columns:
    raise ValueError(
        "Novelty 2 is missing required columns:\n"
        + "\n".join(missing_columns)
    )

# ------------------------------------------------------------
# 6. Verify row count
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("ROW / BLOCK STRUCTURE")
print("=" * 90)

print("Rows:", len(n2))

if len(n2) != 22:
    raise ValueError(
        f"Expected 22 rows, found {len(n2)}."
    )

print("PASS: Exactly 22 rows.")

# ------------------------------------------------------------
# 7. Verify removed_block
# ------------------------------------------------------------

print("\nremoved_block values:")

print(
    n2["removed_block"]
    .tolist()
)

# Convert safely
removed_blocks_numeric = pd.to_numeric(
    n2["removed_block"],
    errors="coerce"
)

if removed_blocks_numeric.isna().any():
    raise ValueError(
        "removed_block contains non-numeric values."
    )

removed_blocks_numeric = (
    removed_blocks_numeric.astype(int)
)

expected_blocks = set(range(1, 23))

actual_blocks = set(
    removed_blocks_numeric.tolist()
)

print(
    "\nExpected blocks:",
    sorted(expected_blocks)
)

print(
    "Actual blocks:",
    sorted(actual_blocks)
)

if actual_blocks != expected_blocks:
    print(
        "Missing:",
        sorted(expected_blocks - actual_blocks)
    )

    print(
        "Extra:",
        sorted(actual_blocks - expected_blocks)
    )

    raise ValueError(
        "Novelty 2 does not contain exactly blocks 1–22."
    )

print(
    "PASS: All 22 blocks B1–B22 present."
)

# ------------------------------------------------------------
# 8. Duplicate block check
# ------------------------------------------------------------

duplicate_blocks = (
    n2["removed_block"]
    .duplicated()
)

if duplicate_blocks.any():
    print(
        "\nDuplicate rows:"
    )
    print(
        n2.loc[
            duplicate_blocks,
            ["removed_block", "removed_layer"]
        ]
    )

    raise ValueError(
        "Duplicate removed_block entries detected."
    )

print(
    "PASS: Exactly one necessity result per block."
)

# ------------------------------------------------------------
# 9. Verify layer identifiers
# ------------------------------------------------------------

print("\nRemoved layers:")

print(
    n2[
        [
            "removed_block",
            "removed_layer"
        ]
    ].to_string(index=False)
)

# ------------------------------------------------------------
# 10. Verify primary AUROC damage
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("PRIMARY AUROC NECESSITY MEASURE")
print("=" * 90)

damage = pd.to_numeric(
    n2["mean_damage_auroc"],
    errors="coerce"
)

if damage.isna().any():
    raise ValueError(
        "mean_damage_auroc contains NaN/non-numeric values."
    )

if not np.all(
    np.isfinite(
        damage.to_numpy()
    )
):
    raise ValueError(
        "mean_damage_auroc contains NaN/Inf."
    )

print(
    "Column: mean_damage_auroc"
)

print(
    f"Mean   : {damage.mean():.6f}"
)

print(
    f"Median : {damage.median():.6f}"
)

print(
    f"Min    : {damage.min():.6f}"
)

print(
    f"Max    : {damage.max():.6f}"
)

print(
    "\nBlock-level AUROC damage:"
)

for block, value in zip(
    removed_blocks_numeric,
    damage
):

    print(
        f"B{block:02d}: {value:.6f}"
    )

print(
    "\nPASS: Primary AUROC damage is finite."
)

# ------------------------------------------------------------
# 11. Verify statistical columns
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("STATISTICAL COLUMN VERIFICATION")
print("=" * 90)

numeric_columns = [
    "mean_damage_auroc",
    "sd_damage_auroc",
    "median_damage_auroc",
    "wilcoxon_W_damage_auroc",
    "p_damage_auroc",
    "q_damage_auroc_FDR",
    "rank_biserial_damage_auroc",
    "ci95_low_damage_auroc",
    "ci95_high_damage_auroc"
]

for col in numeric_columns:

    values = pd.to_numeric(
        n2[col],
        errors="coerce"
    )

    finite = np.isfinite(
        values.to_numpy()
    ).all()

    print(
        f"{col:35s} | "
        f"non-null={values.notna().sum():2d}/22 | "
        f"finite={finite}"
    )

    if values.isna().any():
        raise ValueError(
            f"{col} contains NaN/non-numeric values."
        )

    if not finite:
        raise ValueError(
            f"{col} contains NaN/Inf."
        )

# ------------------------------------------------------------
# 12. Verify significance columns
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("SIGNIFICANCE VERIFICATION")
print("=" * 90)

for col in [
    "significant_damage_auroc",
    "ci_excludes_zero_damage_auroc"
]:

    print(
        f"\n{col}:"
    )

    print(
        n2[col].value_counts(
            dropna=False
        )
    )

# ------------------------------------------------------------
# 13. Verify expected Novelty 2 statistical result
# ------------------------------------------------------------

significant_values = (
    n2[
        "significant_damage_auroc"
    ]
    .astype(str)
    .str.strip()
    .str.lower()
)

ci_values = (
    n2[
        "ci_excludes_zero_damage_auroc"
    ]
    .astype(str)
    .str.strip()
    .str.lower()
)

print(
    "\nSignificant entries:",
    (
        significant_values == "true"
    ).sum()
)

print(
    "CI excludes zero entries:",
    (
        ci_values == "true"
    ).sum()
)

# ------------------------------------------------------------
# 14. Display complete block-level table
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("NOVELTY 2 COMPLETE BLOCK TABLE")
print("=" * 90)

display_columns = [
    "removed_block",
    "removed_layer",
    "mean_damage_auroc",
    "sd_damage_auroc",
    "median_damage_auroc",
    "p_damage_auroc",
    "q_damage_auroc_FDR",
    "rank_biserial_damage_auroc",
    "ci95_low_damage_auroc",
    "ci95_high_damage_auroc",
    "significant_damage_auroc",
    "ci_excludes_zero_damage_auroc"
]

print(
    n2[
        display_columns
    ].to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)

# ------------------------------------------------------------
# 15. Final verification
# ------------------------------------------------------------

assert len(n2) == 22

assert (
    set(
        removed_blocks_numeric.tolist()
    )
    == expected_blocks
)

assert (
    n2["removed_block"]
    .duplicated()
    .sum()
    == 0
)

assert np.all(
    np.isfinite(
        damage.to_numpy()
    )
)

print("\n" + "=" * 90)
print("CELL 10B COMPLETE")
print("=" * 90)

print("PASS: Novelty 2 file exists.")
print("PASS: 22 rows verified.")
print("PASS: Blocks 1–22 verified.")
print("PASS: One row per removed block.")
print("PASS: removed_block verified.")
print("PASS: removed_layer verified.")
print("PASS: mean_damage_auroc verified.")
print("PASS: Statistical columns verified.")
print("PASS: No NaN/Inf in numeric fields.")
print("PASS: Novelty 2 is ready for Cell 10.")
print("\nNO DATA WAS MODIFIED.")
```

```python
# ============================================================
# CELL 10 — FINAL "OURS" SCORE
#
# Ours = Function + Task Necessity + Learned Gates
#
# Function:
#   Mean(pretrained - random) across 4 probe functions
#
# Task:
#   Novelty 2 mean AUROC damage
#
# Gates:
#   Novelty 3 Cell 9 mean learned gate
#
# Each component is min-max normalized across B1-B22.
# Equal weights are used: 1/3 + 1/3 + 1/3.
#
# NO DATA IS MODIFIED UNTIL ALL VERIFICATIONS PASS.
# ============================================================

import os
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 1. Paths
# ------------------------------------------------------------

BASE_DIR = "/content/drive/MyDrive/Seinna_REVE_LOSO"

N1_FILE = os.path.join(
    BASE_DIR,
    "novelty1_probes",
    "novelty1_statistics",
    "novelty1_all_probe_results.csv"
)

N2_FILE = os.path.join(
    BASE_DIR,
    "novelty2_results",
    "novelty2_primary_auroc_necessity.csv"
)

GATE_STATS_FILE = os.path.join(
    BASE_DIR,
    "novelty3_pruning",
    "gate_training",
    "gate_aggregate_statistics.csv"
)

OUTPUT_FILE = os.path.join(
    BASE_DIR,
    "novelty3_pruning",
    "gate_training",
    "ours_block_scores.csv"
)

# ------------------------------------------------------------
# 2. VERIFY FILES BEFORE LOADING
# ------------------------------------------------------------

print("=" * 100)
print("CELL 10 — PRE-RUN VERIFICATION")
print("=" * 100)

input_files = {
    "Novelty 1": N1_FILE,
    "Novelty 2": N2_FILE,
    "Gate statistics": GATE_STATS_FILE
}

for name, path in input_files.items():

    if not os.path.isfile(path):
        raise FileNotFoundError(
            f"{name} file not found:\n{path}"
        )

    print(f"PASS: {name}")
    print(f"      {path}")

# ------------------------------------------------------------
# 3. LOAD FILES
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("LOADING INPUTS")
print("=" * 100)

n1 = pd.read_csv(N1_FILE)
n2 = pd.read_csv(N2_FILE)
gate_stats = pd.read_csv(GATE_STATS_FILE)

print(
    f"Novelty 1 shape       : {n1.shape}"
)

print(
    f"Novelty 2 shape       : {n2.shape}"
)

print(
    f"Gate statistics shape : {gate_stats.shape}"
)

# ------------------------------------------------------------
# 4. VERIFY NOVELTY 1 SCHEMA
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("NOVELTY 1 VERIFICATION")
print("=" * 100)

required_n1_columns = [
    "model",
    "block",
    "function",
    "score"
]

missing_n1 = [
    col
    for col in required_n1_columns
    if col not in n1.columns
]

if missing_n1:
    raise ValueError(
        "Novelty 1 missing columns:\n"
        + "\n".join(missing_n1)
    )

print("PASS: Required Novelty 1 columns present.")

# Normalize text
n1["model_normalized"] = (
    n1["model"]
    .astype(str)
    .str.strip()
    .str.lower()
)

n1["function_normalized"] = (
    n1["function"]
    .astype(str)
    .str.strip()
    .str.lower()
)

# ------------------------------------------------------------
# 5. VERIFY NOVELTY 1 MODELS
# ------------------------------------------------------------

n1_models = set(
    n1["model_normalized"].unique()
)

print(
    "\nNovelty 1 models:",
    sorted(n1_models)
)

required_models = {
    "pretrained",
    "random"
}

if not required_models.issubset(
    n1_models
):
    raise ValueError(
        "Novelty 1 does not contain both "
        "'pretrained' and 'random'."
    )

print(
    "PASS: pretrained + random models present."
)

# ------------------------------------------------------------
# 6. VERIFY NOVELTY 1 FUNCTIONS
# ------------------------------------------------------------

required_functions = {
    "spatial",
    "temporal",
    "spectral",
    "task"
}

n1_functions = set(
    n1["function_normalized"].unique()
)

print(
    "\nNovelty 1 functions:",
    sorted(n1_functions)
)

if n1_functions != required_functions:
    raise ValueError(
        "Unexpected Novelty 1 function set.\n"
        f"Expected: {sorted(required_functions)}\n"
        f"Found:    {sorted(n1_functions)}"
    )

print(
    "PASS: Four required probe functions present."
)

# ------------------------------------------------------------
# 7. VERIFY NOVELTY 1 BLOCKS
# ------------------------------------------------------------

n1_blocks = pd.to_numeric(
    n1["block"],
    errors="coerce"
)

if n1_blocks.isna().any():
    raise ValueError(
        "Novelty 1 contains non-numeric block values."
    )

n1["block"] = (
    n1_blocks.astype(int)
)

expected_blocks = set(
    range(1, 23)
)

actual_n1_blocks = set(
    n1["block"].unique()
)

print(
    "\nNovelty 1 block count:",
    len(actual_n1_blocks)
)

if actual_n1_blocks != expected_blocks:
    raise ValueError(
        "Novelty 1 block mismatch.\n"
        f"Missing: {sorted(expected_blocks - actual_n1_blocks)}\n"
        f"Extra:   {sorted(actual_n1_blocks - expected_blocks)}"
    )

print(
    "PASS: Novelty 1 contains B1-B22."
)

# ------------------------------------------------------------
# 8. VERIFY NOVELTY 1 SCORE
# ------------------------------------------------------------

n1["score"] = pd.to_numeric(
    n1["score"],
    errors="coerce"
)

if n1["score"].isna().any():
    raise ValueError(
        "Novelty 1 score contains NaN/non-numeric values."
    )

if not np.all(
    np.isfinite(
        n1["score"].to_numpy()
    )
):
    raise ValueError(
        "Novelty 1 score contains NaN/Inf."
    )

print(
    "PASS: Novelty 1 score values are finite."
)

# ------------------------------------------------------------
# 9. VERIFY 88 BLOCK × FUNCTION COMBINATIONS
# ------------------------------------------------------------

n1_structure = (
    n1.groupby(
        [
            "block",
            "function_normalized"
        ]
    )
    .size()
    .reset_index(name="count")
)

print(
    "\nUnique block/function combinations:",
    len(n1_structure)
)

if len(n1_structure) != 88:
    raise ValueError(
        "Expected 88 block/function combinations."
    )

counts = (
    n1_structure["count"]
    .unique()
)

print(
    "Rows per block/function:",
    counts.tolist()
)

if len(counts) != 1:
    raise ValueError(
        "Block/function combinations do not have "
        "consistent row counts."
    )

print(
    "PASS: 22 × 4 = 88 block/function combinations."
)

# ------------------------------------------------------------
# 10. CALCULATE FUNCTIONAL PRETRAINED-RANDOM DIFFERENCES
# ------------------------------------------------------------

function_means = (
    n1.groupby(
        [
            "block",
            "function_normalized",
            "model_normalized"
        ]
    )["score"]
    .mean()
    .reset_index()
)

function_pivot = (
    function_means
    .pivot(
        index=[
            "block",
            "function_normalized"
        ],
        columns="model_normalized",
        values="score"
    )
    .reset_index()
)

# Remove pivot column name
function_pivot.columns.name = None

if "pretrained" not in function_pivot.columns:
    raise ValueError(
        "Pretrained scores missing after pivot."
    )

if "random" not in function_pivot.columns:
    raise ValueError(
        "Random scores missing after pivot."
    )

function_pivot[
    "pretrained_minus_random"
] = (
    function_pivot["pretrained"]
    -
    function_pivot["random"]
)

if not np.all(
    np.isfinite(
        function_pivot[
            "pretrained_minus_random"
        ].to_numpy()
    )
):
    raise ValueError(
        "Non-finite pretrained-random differences."
    )

print(
    "\nPASS: 88 pretrained-vs-random functional "
    "comparisons calculated."
)

# ------------------------------------------------------------
# 11. VERIFY EXACTLY FOUR FUNCTIONS PER BLOCK
# ------------------------------------------------------------

function_counts = (
    function_pivot
    .groupby("block")
    ["function_normalized"]
    .nunique()
)

if not np.all(
    function_counts.to_numpy() == 4
):
    raise ValueError(
        "Not every block has exactly four functions."
    )

print(
    "PASS: Every block has all four functional probes."
)

# ------------------------------------------------------------
# 12. FUNCTION COMPONENT
#
# Average the four pretrained-random differences
# for each block.
# ------------------------------------------------------------

functional = (
    function_pivot
    .groupby("block")[
        "pretrained_minus_random"
    ]
    .mean()
    .reset_index()
)

functional = functional.rename(
    columns={
        "pretrained_minus_random":
        "function_raw"
    }
)

if len(functional) != 22:
    raise ValueError(
        "Function component does not contain 22 blocks."
    )

print(
    "\nFunction component calculated for 22 blocks."
)

# ------------------------------------------------------------
# 13. VERIFY NOVELTY 2 SCHEMA
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("NOVELTY 2 VERIFICATION")
print("=" * 100)

required_n2_columns = [
    "removed_block",
    "removed_layer",
    "mean_damage_auroc"
]

missing_n2 = [
    col
    for col in required_n2_columns
    if col not in n2.columns
]

if missing_n2:
    raise ValueError(
        "Novelty 2 missing columns:\n"
        + "\n".join(missing_n2)
    )

print(
    "PASS: removed_block present."
)

print(
    "PASS: removed_layer present."
)

print(
    "PASS: mean_damage_auroc present."
)

# ------------------------------------------------------------
# 14. VERIFY NOVELTY 2 ROWS / BLOCKS
# ------------------------------------------------------------

if len(n2) != 22:
    raise ValueError(
        f"Expected 22 Novelty 2 rows, found {len(n2)}."
    )

n2["removed_block"] = pd.to_numeric(
    n2["removed_block"],
    errors="coerce"
)

if n2["removed_block"].isna().any():
    raise ValueError(
        "Novelty 2 removed_block contains invalid values."
    )

n2["removed_block"] = (
    n2["removed_block"].astype(int)
)

actual_n2_blocks = set(
    n2["removed_block"].unique()
)

if actual_n2_blocks != expected_blocks:
    raise ValueError(
        "Novelty 2 does not contain exactly B1-B22."
    )

if (
    n2["removed_block"]
    .duplicated()
    .any()
):
    raise ValueError(
        "Duplicate Novelty 2 removed_block values found."
    )

print(
    "PASS: Novelty 2 has exactly one result for B1-B22."
)

# ------------------------------------------------------------
# 15. VERIFY NOVELTY 2 LAYERS
# ------------------------------------------------------------

expected_layers = [
    f"H{i}"
    for i in range(22)
]

actual_layers = (
    n2.sort_values("removed_block")
    ["removed_layer"]
    .astype(str)
    .str.strip()
    .tolist()
)

if actual_layers != expected_layers:
    raise ValueError(
        "Novelty 2 layer mapping is not H0-H21."
    )

print(
    "PASS: B1-H0 through B22-H21 mapping verified."
)

# ------------------------------------------------------------
# 16. VERIFY NOVELTY 2 AUROC DAMAGE
# ------------------------------------------------------------

n2["mean_damage_auroc"] = pd.to_numeric(
    n2["mean_damage_auroc"],
    errors="coerce"
)

if n2["mean_damage_auroc"].isna().any():
    raise ValueError(
        "Novelty 2 mean_damage_auroc contains NaN."
    )

if not np.all(
    np.isfinite(
        n2["mean_damage_auroc"].to_numpy()
    )
):
    raise ValueError(
        "Novelty 2 mean_damage_auroc contains NaN/Inf."
    )

print(
    "PASS: Novelty 2 AUROC damage is finite."
)

# ------------------------------------------------------------
# 17. TASK COMPONENT
# ------------------------------------------------------------

task = (
    n2[
        [
            "removed_block",
            "mean_damage_auroc"
        ]
    ]
    .copy()
    .rename(
        columns={
            "removed_block": "block",
            "mean_damage_auroc": "task_raw"
        }
    )
)

# ------------------------------------------------------------
# 18. VERIFY GATE STATISTICS
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("CELL 9 GATE VERIFICATION")
print("=" * 100)

required_gate_columns = [
    "block",
    "gate_mean",
    "gate_std",
    "gate_median",
    "gate_min",
    "gate_max"
]

missing_gate = [
    col
    for col in required_gate_columns
    if col not in gate_stats.columns
]

if missing_gate:
    raise ValueError(
        "Gate statistics missing columns:\n"
        + "\n".join(missing_gate)
    )

print(
    "PASS: All Cell 9 gate-statistics columns present."
)

# ------------------------------------------------------------
# 19. VERIFY 22 GATE ROWS
# ------------------------------------------------------------

if len(gate_stats) != 22:
    raise ValueError(
        f"Expected 22 gate rows, found {len(gate_stats)}."
    )

gate_stats["block"] = pd.to_numeric(
    gate_stats["block"],
    errors="coerce"
)

if gate_stats["block"].isna().any():
    raise ValueError(
        "Gate statistics block contains invalid values."
    )

gate_stats["block"] = (
    gate_stats["block"].astype(int)
)

actual_gate_blocks = set(
    gate_stats["block"].unique()
)

if actual_gate_blocks != expected_blocks:
    raise ValueError(
        "Gate statistics does not contain exactly B1-B22."
    )

if (
    gate_stats["block"]
    .duplicated()
    .any()
):
    raise ValueError(
        "Duplicate gate block values found."
    )

print(
    "PASS: Gate statistics has exactly one row per block."
)

# ------------------------------------------------------------
# 20. VERIFY GATE VALUES
# ------------------------------------------------------------

gate_stats["gate_mean"] = pd.to_numeric(
    gate_stats["gate_mean"],
    errors="coerce"
)

if gate_stats["gate_mean"].isna().any():
    raise ValueError(
        "gate_mean contains NaN."
    )

if not np.all(
    np.isfinite(
        gate_stats["gate_mean"].to_numpy()
    )
):
    raise ValueError(
        "gate_mean contains NaN/Inf."
    )

print(
    "PASS: Gate means are finite."
)

# ------------------------------------------------------------
# 21. GATE COMPONENT
# ------------------------------------------------------------

gates = (
    gate_stats[
        [
            "block",
            "gate_mean"
        ]
    ]
    .copy()
    .rename(
        columns={
            "gate_mean": "gate_raw"
        }
    )
)

# ------------------------------------------------------------
# 22. MERGE ALL THREE COMPONENTS
# ------------------------------------------------------------

print("\n" + "=" * 100)
print("MERGING FUNCTION + TASK + GATES")
print("=" * 100)

ours = (
    functional
    .merge(
        task,
        on="block",
        how="inner"
    )
    .merge(
        gates,
        on="block",
        how="inner"
    )
)

ours = (
    ours
    .sort_values("block")
    .reset_index(drop=True)
)

print(
    "Merged rows:",
    len(ours)
)

if len(ours) != 22:
    raise ValueError(
        "Merged result does not contain exactly 22 blocks."
    )

if set(ours["block"]) != expected_blocks:
    raise ValueError(
        "Merged result does not contain B1-B22."
    )

print(
    "PASS: All 22 blocks have Function + Task + Gate."
)

# ------------------------------------------------------------
# 23. VERIFY ALL RAW COMPONENTS
# ------------------------------------------------------------

for column in [
    "function_raw",
    "task_raw",
    "gate_raw"
]:

    values = pd.to_numeric(
        ours[column],
        errors="coerce"
    )

    if values.isna().any():
        raise ValueError(
            f"{column} contains NaN."
        )

    if not np.all(
        np.isfinite(
            values.to_numpy()
        )
    ):
        raise ValueError(
            f"{column} contains NaN/Inf."
        )

print(
    "PASS: All three raw components are finite."
)

# ------------------------------------------------------------
# 24. MIN-MAX NORMALIZATION
# ------------------------------------------------------------

def minmax_normalize(values):

    values = np.asarray(
        values,
        dtype=np.float64
    )

    vmin = np.min(values)
    vmax = np.max(values)

    if not np.isfinite(vmin):
        raise ValueError(
            "Minimum is not finite."
        )

    if not np.isfinite(vmax):
        raise ValueError(
            "Maximum is not finite."
        )

    if np.isclose(
        vmax,
        vmin
    ):
        return np.ones_like(values)

    return (
        (values - vmin)
        /
        (vmax - vmin)
    )

ours["function_score"] = minmax_normalize(
    ours["function_raw"]
)

ours["task_score"] = minmax_normalize(
    ours["task_raw"]
)

ours["gate_score"] = minmax_normalize(
    ours["gate_raw"]
)

# ------------------------------------------------------------
# 25. VERIFY NORMALIZED COMPONENTS
# ------------------------------------------------------------

for column in [
    "function_score",
    "task_score",
    "gate_score"
]:

    values = ours[column].to_numpy()

    if not np.all(
        np.isfinite(values)
    ):
        raise ValueError(
            f"{column} contains NaN/Inf."
        )

    if np.any(values < -1e-10):
        raise ValueError(
            f"{column} contains values below 0."
        )

    if np.any(values > 1 + 1e-10):
        raise ValueError(
            f"{column} contains values above 1."
        )

print(
    "PASS: All components normalized to [0, 1]."
)

# ------------------------------------------------------------
# 26. FINAL OURS FORMULATION
# ------------------------------------------------------------

FUNCTION_WEIGHT = 1.0 / 3.0
TASK_WEIGHT = 1.0 / 3.0
GATE_WEIGHT = 1.0 / 3.0

ours["ours_score"] = (
    FUNCTION_WEIGHT
    * ours["function_score"]
    +
    TASK_WEIGHT
    * ours["task_score"]
    +
    GATE_WEIGHT
    * ours["gate_score"]
)

# ------------------------------------------------------------
# 27. VERIFY FINAL SCORE
# ------------------------------------------------------------

if not np.all(
    np.isfinite(
        ours["ours_score"].to_numpy()
    )
):
    raise ValueError(
        "Final Ours score contains NaN/Inf."
    )

if np.any(
    ours["ours_score"].to_numpy()
    < -1e-10
):
    raise ValueError(
        "Final Ours score contains values below 0."
    )

if np.any(
    ours["ours_score"].to_numpy()
    > 1 + 1e-10
):
    raise ValueError(
        "Final Ours score contains values above 1."
    )

print(
    "PASS: Final Ours scores are finite and in [0,1]."
)

# ------------------------------------------------------------
# 28. FINAL RANKING
# ------------------------------------------------------------

ours["ours_rank"] = (
    ours["ours_score"]
    .rank(
        ascending=False,
        method="min"
    )
    .astype(int)
)

if sorted(
    ours["ours_rank"].tolist()
) != list(range(1, 23)):
    raise ValueError(
        "Ranking does not contain exactly ranks 1-22."
    )

ours = (
    ours
    .sort_values(
        "ours_score",
        ascending=False
    )
    .reset_index(drop=True)
)

# ------------------------------------------------------------
# 29. SAVE OUTPUT
# ------------------------------------------------------------

ours.to_csv(
    OUTPUT_FILE,
    index=False
)

# ------------------------------------------------------------
# 30. DISPLAY FINAL RESULT
# ------------------------------------------------------------

display_columns = [
    "block",
    "function_raw",
    "task_raw",
    "gate_raw",
    "function_score",
    "task_score",
    "gate_score",
    "ours_score",
    "ours_rank"
]

print("\n" + "=" * 110)
print("FINAL OURS RANKING")
print("=" * 110)

print(
    ours[
        display_columns
    ].to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)

# ------------------------------------------------------------
# 31. FINAL COMPLETE VERIFICATION
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("CELL 10 COMPLETE")
print("=" * 110)

print(
    "Function component : pretrained − random across 4 probes"
)

print(
    "Task component     : Novelty 2 mean_damage_auroc"
)

print(
    "Gate component     : Cell 9 gate_mean"
)

print(
    "\nFinal formulation:"
)

print(
    "Ours = (Function + Task + Gates) / 3"
)

print(
    "\nWeights:"
)

print(
    f"Function = {FUNCTION_WEIGHT:.6f}"
)

print(
    f"Task     = {TASK_WEIGHT:.6f}"
)

print(
    f"Gates    = {GATE_WEIGHT:.6f}"
)

print(
    "\nFinal blocks ranked:",
    len(ours)
)

print(
    "\nSaved:"
)

print(
    OUTPUT_FILE
)

print(
    "\nPASS: Novelty 1 verified."
)

print(
    "PASS: Novelty 2 verified."
)

print(
    "PASS: Cell 9 gates verified."
)

print(
    "PASS: Function + Task + Gates combined."
)

print(
    "PASS: 22 blocks ranked."
)

print(
    "PASS: No retention term introduced."
)
```

```python
# ============================================================
# CELL 11 — PRUNING PLAN GENERATION
#
# Uses the Cell 10 "Ours" ranking.
#
# Higher Ours score = more removable.
#
# Pruning levels:
#   10% -> remove 2
#   20% -> remove 4
#   30% -> remove 7
#   40% -> remove 9
#   50% -> remove 11
#   60% -> remove 13
#   70% -> remove 15
#   80% -> remove 18
#   90% -> remove 20
#
# NO MODEL IS RUN IN THIS CELL.
# NO EEG DATA IS MODIFIED.
# ============================================================

import os
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 1. Paths
# ------------------------------------------------------------

BASE_DIR = "/content/drive/MyDrive/Seinna_REVE_LOSO"

OURS_FILE = os.path.join(
    BASE_DIR,
    "novelty3_pruning",
    "gate_training",
    "ours_block_scores.csv"
)

OUTPUT_FILE = os.path.join(
    BASE_DIR,
    "novelty3_pruning",
    "pruning_plans.csv"
)

# ------------------------------------------------------------
# 2. VERIFY INPUT FILE
# ------------------------------------------------------------

print("=" * 100)
print("CELL 11 — PRE-RUN VERIFICATION")
print("=" * 100)

if not os.path.isfile(OURS_FILE):
    raise FileNotFoundError(
        f"Cell 10 output not found:\n{OURS_FILE}"
    )

print("PASS: Cell 10 output exists.")
print(OURS_FILE)

# ------------------------------------------------------------
# 3. LOAD CELL 10 RESULTS
# ------------------------------------------------------------

ours = pd.read_csv(OURS_FILE)

print("\nCell 10 file shape:", ours.shape)

# ------------------------------------------------------------
# 4. VERIFY REQUIRED COLUMNS
# ------------------------------------------------------------

required_columns = [
    "block",
    "function_raw",
    "task_raw",
    "gate_raw",
    "function_score",
    "task_score",
    "gate_score",
    "ours_score",
    "ours_rank"
]

missing_columns = [
    col
    for col in required_columns
    if col not in ours.columns
]

if missing_columns:
    raise ValueError(
        "Cell 10 output is missing required columns:\n"
        + "\n".join(missing_columns)
    )

print("PASS: All Cell 10 columns present.")

# ------------------------------------------------------------
# 5. VERIFY EXACTLY 22 BLOCKS
# ------------------------------------------------------------

ours["block"] = pd.to_numeric(
    ours["block"],
    errors="coerce"
)

if ours["block"].isna().any():
    raise ValueError(
        "Block column contains invalid values."
    )

ours["block"] = (
    ours["block"]
    .astype(int)
)

expected_blocks = set(range(1, 23))

actual_blocks = set(
    ours["block"].tolist()
)

if actual_blocks != expected_blocks:
    raise ValueError(
        "Cell 10 blocks are not exactly B1-B22.\n"
        f"Missing: {sorted(expected_blocks - actual_blocks)}\n"
        f"Extra:   {sorted(actual_blocks - expected_blocks)}"
    )

if ours["block"].duplicated().any():
    raise ValueError(
        "Duplicate block entries found in Cell 10 output."
    )

print("PASS: Exactly one entry for each block B1-B22.")

# ------------------------------------------------------------
# 6. VERIFY Ours SCORES
# ------------------------------------------------------------

ours["ours_score"] = pd.to_numeric(
    ours["ours_score"],
    errors="coerce"
)

if ours["ours_score"].isna().any():
    raise ValueError(
        "ours_score contains NaN."
    )

if not np.all(
    np.isfinite(
        ours["ours_score"].to_numpy()
    )
):
    raise ValueError(
        "ours_score contains NaN/Inf."
    )

print("PASS: Ours scores are finite.")

# ------------------------------------------------------------
# 7. VERIFY RANKS
# ------------------------------------------------------------

ours["ours_rank"] = pd.to_numeric(
    ours["ours_rank"],
    errors="coerce"
)

if ours["ours_rank"].isna().any():
    raise ValueError(
        "ours_rank contains invalid values."
    )

ours["ours_rank"] = (
    ours["ours_rank"]
    .astype(int)
)

expected_ranks = set(range(1, 23))

actual_ranks = set(
    ours["ours_rank"].tolist()
)

if actual_ranks != expected_ranks:
    raise ValueError(
        "Ranks are not exactly 1-22."
    )

print("PASS: Ranks 1-22 verified.")

# ------------------------------------------------------------
# 8. VERIFY RANK ↔ SCORE ORDER
# ------------------------------------------------------------

ours_by_rank = (
    ours
    .sort_values("ours_rank")
    .reset_index(drop=True)
)

score_values = (
    ours_by_rank["ours_score"]
    .to_numpy()
)

if not np.all(
    score_values[:-1]
    >=
    score_values[1:]
):
    raise ValueError(
        "Ours ranking is not sorted in descending score order."
    )

print(
    "PASS: Rank 1 has the highest Ours score "
    "and rank 22 the lowest."
)

# ------------------------------------------------------------
# 9. DEFINE PRUNING LEVELS
# ------------------------------------------------------------

pruning_levels = {
    10: 2,
    20: 4,
    30: 7,
    40: 9,
    50: 11,
    60: 13,
    70: 15,
    80: 18,
    90: 20
}

print("\n" + "=" * 100)
print("PRUNING LEVEL VERIFICATION")
print("=" * 100)

for percentage, remove_count in pruning_levels.items():

    expected_remove = int(
        round(
            22 * percentage / 100
        )
    )

    # Explicitly enforce the previously defined plan.
    if remove_count != expected_remove:

        # 30%, 40%, 50%, etc. are intentionally rounded
        # to the predefined integer pruning counts.
        print(
            f"{percentage:>3}% -> remove {remove_count} "
            f"(predefined plan)"
        )

    else:

        print(
            f"{percentage:>3}% -> remove {remove_count}"
        )

# ------------------------------------------------------------
# 10. GENERATE PLANS
# ------------------------------------------------------------

plan_rows = []

for percentage, remove_count in pruning_levels.items():

    selected = (
        ours_by_rank
        .iloc[:remove_count]
        .copy()
    )

    remove_blocks = (
        selected["block"]
        .astype(int)
        .tolist()
    )

    remove_ranks = (
        selected["ours_rank"]
        .astype(int)
        .tolist()
    )

    remove_scores = (
        selected["ours_score"]
        .astype(float)
        .tolist()
    )

    keep_blocks = [
        block
        for block in range(1, 23)
        if block not in remove_blocks
    ]

    # Store one row per pruning level.
    plan_rows.append(
        {
            "pruning_percent": percentage,
            "remove_count": remove_count,
            "keep_count": 22 - remove_count,

            "removed_blocks": ",".join(
                f"B{b}"
                for b in remove_blocks
            ),

            "removed_ranks": ",".join(
                str(r)
                for r in remove_ranks
            ),

            "removed_ours_scores": ",".join(
                f"{s:.6f}"
                for s in remove_scores
            ),

            "kept_blocks": ",".join(
                f"B{b}"
                for b in keep_blocks
            )
        }
    )

plans = pd.DataFrame(
    plan_rows
)

# ------------------------------------------------------------
# 11. VERIFY PLAN COUNT
# ------------------------------------------------------------

if len(plans) != len(
    pruning_levels
):
    raise ValueError(
        "Incorrect number of pruning plans."
    )

print(
    "\nPASS:",
    len(plans),
    "pruning plans generated."
)

# ------------------------------------------------------------
# 12. VERIFY REMOVE + KEEP COUNTS
# ------------------------------------------------------------

for _, row in plans.iterrows():

    percentage = int(
        row["pruning_percent"]
    )

    remove_count = int(
        row["remove_count"]
    )

    keep_count = int(
        row["keep_count"]
    )

    if remove_count + keep_count != 22:
        raise ValueError(
            f"{percentage}% plan does not sum to 22 blocks."
        )

print(
    "PASS: Remove + keep counts = 22 for every plan."
)

# ------------------------------------------------------------
# 13. VERIFY REMOVED BLOCKS
# ------------------------------------------------------------

for _, row in plans.iterrows():

    percentage = int(
        row["pruning_percent"]
    )

    remove_count = int(
        row["remove_count"]
    )

    removed = [
        x
        for x in row["removed_blocks"].split(",")
        if x
    ]

    if len(removed) != remove_count:
        raise ValueError(
            f"{percentage}% plan has "
            f"{len(removed)} removed blocks, "
            f"expected {remove_count}."
        )

    if len(set(removed)) != remove_count:
        raise ValueError(
            f"{percentage}% plan contains duplicate blocks."
        )

print(
    "PASS: Every pruning plan contains the "
    "correct number of unique removed blocks."
)

# ------------------------------------------------------------
# 14. VERIFY NESTED PRUNING STRUCTURE
#
# Every higher pruning level should contain all blocks
# removed at the previous level, because we are progressively
# removing blocks according to the same Ours ranking.
# ------------------------------------------------------------

previous_removed = set()

for _, row in plans.sort_values(
    "pruning_percent"
).iterrows():

    percentage = int(
        row["pruning_percent"]
    )

    removed = set(
        row["removed_blocks"].split(",")
    )

    if not previous_removed.issubset(
        removed
    ):
        raise ValueError(
            f"{percentage}% plan is not nested "
            "inside the previous pruning plan."
        )

    previous_removed = removed

print(
    "PASS: Pruning plans are nested according "
    "to the Ours ranking."
)

# ------------------------------------------------------------
# 15. VERIFY FIRST / LAST BLOCKS
# ------------------------------------------------------------

rank1_block = (
    ours_by_rank
    .iloc[0]["block"]
)

rank22_block = (
    ours_by_rank
    .iloc[-1]["block"]
)

if rank1_block != 8:
    raise ValueError(
        f"Expected rank-1 block B8, found B{rank1_block}."
    )

if rank22_block != 1:
    raise ValueError(
        f"Expected rank-22 block B1, found B{rank22_block}."
    )

print(
    "PASS: Rank-1 B8 and rank-22 B1 verified."
)

# ------------------------------------------------------------
# 16. SAVE PRUNING PLANS
# ------------------------------------------------------------

plans.to_csv(
    OUTPUT_FILE,
    index=False
)

print(
    "\nSaved:"
)

print(
    OUTPUT_FILE
)

# ------------------------------------------------------------
# 17. DISPLAY FINAL PLANS
# ------------------------------------------------------------

print("\n" + "=" * 120)
print("FINAL PRUNING PLANS")
print("=" * 120)

display(
    plans[
        [
            "pruning_percent",
            "remove_count",
            "keep_count",
            "removed_blocks",
            "removed_ranks"
        ]
    ]
)

# ------------------------------------------------------------
# 18. DISPLAY REMOVAL ORDER
# ------------------------------------------------------------

print("\n" + "=" * 120)
print("REMOVAL ORDER FROM CELL 10")
print("=" * 120)

removal_order = (
    ours_by_rank[
        [
            "ours_rank",
            "block",
            "ours_score"
        ]
    ]
    .copy()
)

removal_order["block"] = (
    removal_order["block"]
    .apply(lambda x: f"B{x}")
)

print(
    removal_order.to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)

# ------------------------------------------------------------
# 19. FINAL CHECK
# ------------------------------------------------------------

print("\n" + "=" * 120)
print("CELL 11 COMPLETE")
print("=" * 120)

print(
    "PASS: Cell 10 ranking loaded and verified."
)

print(
    "PASS: 22 blocks verified."
)

print(
    "PASS: 9 pruning levels generated."
)

print(
    "PASS: Progressive removal follows Ours ranking."
)

print(
    "PASS: No model evaluation performed."
)

print(
    "PASS: No EEG data modified."
)

print(
    "PASS: Pruning plans saved."
)
```

```python
# ============================================================
# CELL 12 — PRUNING PLAN VALIDATION
#
# Verification-only cell.
#
# This cell DOES NOT:
#   - load REVE
#   - load EEG data
#   - run a forward pass
#   - train a classifier
#   - perform pruning
#
# It only verifies that:
#   1. Cell 10 ranking is intact.
#   2. Cell 11 pruning plans are correct.
#   3. Removed/kept blocks are valid.
#   4. Every pruning level is nested.
#   5. Every plan exactly follows the Ours ranking.
#
# NO DATA IS MODIFIED.
# ============================================================

import os
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

BASE_DIR = "/content/drive/MyDrive/Seinna_REVE_LOSO"

OURS_FILE = os.path.join(
    BASE_DIR,
    "novelty3_pruning",
    "gate_training",
    "ours_block_scores.csv"
)

PLANS_FILE = os.path.join(
    BASE_DIR,
    "novelty3_pruning",
    "pruning_plans.csv"
)

# ------------------------------------------------------------
# 2. EXPECTED PRUNING CONFIGURATION
# ------------------------------------------------------------

EXPECTED_PLANS = {
    10: 2,
    20: 4,
    30: 7,
    40: 9,
    50: 11,
    60: 13,
    70: 15,
    80: 18,
    90: 20
}

EXPECTED_BLOCKS = set(range(1, 23))
EXPECTED_RANKS = set(range(1, 23))

# ------------------------------------------------------------
# 3. PRE-RUN FILE VERIFICATION
# ------------------------------------------------------------

print("=" * 110)
print("CELL 12 — PRE-RUN VERIFICATION")
print("=" * 110)

if not os.path.isfile(OURS_FILE):
    raise FileNotFoundError(
        "Cell 10 output not found:\n"
        + OURS_FILE
    )

if not os.path.isfile(PLANS_FILE):
    raise FileNotFoundError(
        "Cell 11 pruning plan file not found:\n"
        + PLANS_FILE
    )

print("PASS: Cell 10 output exists.")
print(OURS_FILE)

print("\nPASS: Cell 11 pruning-plan file exists.")
print(PLANS_FILE)

# ------------------------------------------------------------
# 4. LOAD FILES
# ------------------------------------------------------------

ours = pd.read_csv(OURS_FILE)
plans = pd.read_csv(PLANS_FILE)

print("\n" + "=" * 110)
print("LOADED FILES")
print("=" * 110)

print(
    f"Cell 10 Ours file shape : {ours.shape}"
)

print(
    f"Cell 11 plans shape     : {plans.shape}"
)

# ------------------------------------------------------------
# 5. VERIFY CELL 10 SCHEMA
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("CELL 10 RANKING VERIFICATION")
print("=" * 110)

required_ours_columns = [
    "block",
    "ours_score",
    "ours_rank"
]

missing_ours = [
    col
    for col in required_ours_columns
    if col not in ours.columns
]

if missing_ours:
    raise ValueError(
        "Cell 10 output missing columns:\n"
        + "\n".join(missing_ours)
    )

print(
    "PASS: Required Cell 10 ranking columns present."
)

# ------------------------------------------------------------
# 6. VERIFY CELL 10 BLOCKS
# ------------------------------------------------------------

ours["block"] = pd.to_numeric(
    ours["block"],
    errors="coerce"
)

if ours["block"].isna().any():
    raise ValueError(
        "Cell 10 block contains invalid values."
    )

ours["block"] = (
    ours["block"]
    .astype(int)
)

if set(ours["block"]) != EXPECTED_BLOCKS:
    raise ValueError(
        "Cell 10 does not contain exactly B1-B22."
    )

if ours["block"].duplicated().any():
    raise ValueError(
        "Cell 10 contains duplicate blocks."
    )

print(
    "PASS: Cell 10 contains exactly one row for B1-B22."
)

# ------------------------------------------------------------
# 7. VERIFY CELL 10 RANKS
# ------------------------------------------------------------

ours["ours_rank"] = pd.to_numeric(
    ours["ours_rank"],
    errors="coerce"
)

if ours["ours_rank"].isna().any():
    raise ValueError(
        "Cell 10 ours_rank contains invalid values."
    )

ours["ours_rank"] = (
    ours["ours_rank"]
    .astype(int)
)

if set(ours["ours_rank"]) != EXPECTED_RANKS:
    raise ValueError(
        "Cell 10 ranks are not exactly 1-22."
    )

if ours["ours_rank"].duplicated().any():
    raise ValueError(
        "Cell 10 contains duplicate ranks."
    )

print(
    "PASS: Cell 10 ranks 1-22 verified."
)

# ------------------------------------------------------------
# 8. RECONSTRUCT AUTHORITATIVE RANKING
# ------------------------------------------------------------

ours_ranked = (
    ours
    .sort_values("ours_rank")
    .reset_index(drop=True)
)

authoritative_order = (
    ours_ranked["block"]
    .astype(int)
    .tolist()
)

authoritative_rank = {
    block: rank
    for rank, block
    in enumerate(
        authoritative_order,
        start=1
    )
}

print(
    "\nAuthoritative removal order from Cell 10:"
)

print(
    " → ".join(
        f"B{block}"
        for block in authoritative_order
    )
)

# ------------------------------------------------------------
# 9. VERIFY CELL 11 PLAN SCHEMA
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("CELL 11 PLAN SCHEMA VERIFICATION")
print("=" * 110)

required_plan_columns = [
    "pruning_percent",
    "remove_count",
    "keep_count",
    "removed_blocks",
    "removed_ranks",
    "removed_ours_scores",
    "kept_blocks"
]

missing_plan_columns = [
    col
    for col in required_plan_columns
    if col not in plans.columns
]

if missing_plan_columns:
    raise ValueError(
        "Cell 11 plan file missing columns:\n"
        + "\n".join(missing_plan_columns)
    )

print(
    "PASS: All Cell 11 pruning-plan columns present."
)

# ------------------------------------------------------------
# 10. VERIFY EXACTLY 9 PLANS
# ------------------------------------------------------------

if len(plans) != 9:
    raise ValueError(
        f"Expected 9 pruning plans, found {len(plans)}."
    )

print(
    "PASS: Exactly 9 pruning plans found."
)

# ------------------------------------------------------------
# 11. VERIFY PRUNING PERCENTAGES
# ------------------------------------------------------------

plans["pruning_percent"] = pd.to_numeric(
    plans["pruning_percent"],
    errors="coerce"
)

if plans["pruning_percent"].isna().any():
    raise ValueError(
        "pruning_percent contains invalid values."
    )

plans["pruning_percent"] = (
    plans["pruning_percent"]
    .astype(int)
)

actual_percentages = set(
    plans["pruning_percent"].tolist()
)

expected_percentages = set(
    EXPECTED_PLANS.keys()
)

if actual_percentages != expected_percentages:
    raise ValueError(
        "Pruning percentages do not match expected levels.\n"
        f"Expected: {sorted(expected_percentages)}\n"
        f"Found:    {sorted(actual_percentages)}"
    )

if plans["pruning_percent"].duplicated().any():
    raise ValueError(
        "Duplicate pruning percentages found."
    )

print(
    "PASS: 10%, 20%, ..., 90% all present exactly once."
)

# ------------------------------------------------------------
# 12. VERIFY REMOVE / KEEP COUNTS
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("REMOVE / KEEP COUNT VERIFICATION")
print("=" * 110)

plans_sorted = (
    plans
    .sort_values("pruning_percent")
    .reset_index(drop=True)
)

for _, row in plans_sorted.iterrows():

    percentage = int(
        row["pruning_percent"]
    )

    expected_remove = EXPECTED_PLANS[
        percentage
    ]

    expected_keep = 22 - expected_remove

    remove_count = int(
        row["remove_count"]
    )

    keep_count = int(
        row["keep_count"]
    )

    if remove_count != expected_remove:
        raise ValueError(
            f"{percentage}%: expected "
            f"{expected_remove} removed, "
            f"found {remove_count}."
        )

    if keep_count != expected_keep:
        raise ValueError(
            f"{percentage}%: expected "
            f"{expected_keep} kept, "
            f"found {keep_count}."
        )

    if remove_count + keep_count != 22:
        raise ValueError(
            f"{percentage}%: remove + keep != 22."
        )

    print(
        f"PASS: {percentage:>3}% → "
        f"remove={remove_count}, "
        f"keep={keep_count}"
    )

# ------------------------------------------------------------
# 13. PARSE BLOCK STRINGS
# ------------------------------------------------------------

def parse_block_string(value, field_name, percentage):

    if pd.isna(value):
        raise ValueError(
            f"{percentage}% {field_name} is NaN."
        )

    value = str(value).strip()

    if value == "":
        return []

    tokens = [
        token.strip()
        for token in value.split(",")
        if token.strip()
    ]

    parsed = []

    for token in tokens:

        if not token.startswith("B"):
            raise ValueError(
                f"{percentage}% {field_name} contains "
                f"invalid block token: {token}"
            )

        number_text = token[1:]

        if not number_text.isdigit():
            raise ValueError(
                f"{percentage}% {field_name} contains "
                f"invalid block token: {token}"
            )

        block = int(number_text)

        if block not in EXPECTED_BLOCKS:
            raise ValueError(
                f"{percentage}% {field_name} contains "
                f"invalid block B{block}."
            )

        parsed.append(block)

    return parsed


# ------------------------------------------------------------
# 14. VERIFY EACH REMOVED / KEPT BLOCK SET
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("BLOCK SET VERIFICATION")
print("=" * 110)

parsed_plans = {}

for _, row in plans_sorted.iterrows():

    percentage = int(
        row["pruning_percent"]
    )

    removed = parse_block_string(
        row["removed_blocks"],
        "removed_blocks",
        percentage
    )

    kept = parse_block_string(
        row["kept_blocks"],
        "kept_blocks",
        percentage
    )

    # Correct number of blocks.
    expected_remove = EXPECTED_PLANS[
        percentage
    ]

    expected_keep = 22 - expected_remove

    if len(removed) != expected_remove:
        raise ValueError(
            f"{percentage}% contains "
            f"{len(removed)} removed blocks; "
            f"expected {expected_remove}."
        )

    if len(kept) != expected_keep:
        raise ValueError(
            f"{percentage}% contains "
            f"{len(kept)} kept blocks; "
            f"expected {expected_keep}."
        )

    # No duplicates.
    if len(set(removed)) != len(removed):
        raise ValueError(
            f"{percentage}% has duplicate removed blocks."
        )

    if len(set(kept)) != len(kept):
        raise ValueError(
            f"{percentage}% has duplicate kept blocks."
        )

    removed_set = set(removed)
    kept_set = set(kept)

    # Removed and kept must be disjoint.
    if removed_set.intersection(kept_set):
        raise ValueError(
            f"{percentage}% has blocks simultaneously "
            "removed and kept."
        )

    # Together they must contain all 22 blocks.
    if removed_set.union(kept_set) != EXPECTED_BLOCKS:
        raise ValueError(
            f"{percentage}% removed + kept blocks "
            "do not cover exactly B1-B22."
        )

    parsed_plans[percentage] = {
        "removed": removed,
        "kept": kept
    }

    print(
        f"PASS: {percentage:>3}% → "
        f"{len(removed)} removed + "
        f"{len(kept)} kept = 22"
    )

# ------------------------------------------------------------
# 15. VERIFY REMOVAL RANKS
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("REMOVAL-RANK VERIFICATION")
print("=" * 110)

for _, row in plans_sorted.iterrows():

    percentage = int(
        row["pruning_percent"]
    )

    removed = parsed_plans[
        percentage
    ]["removed"]

    expected_remove = EXPECTED_PLANS[
        percentage
    ]

    expected_blocks = authoritative_order[
        :expected_remove
    ]

    if removed != expected_blocks:
        raise ValueError(
            f"{percentage}% removed blocks do not "
            "exactly follow Cell 10 Ours ranking.\n"
            f"Expected: {expected_blocks}\n"
            f"Found:    {removed}"
        )

    print(
        f"PASS: {percentage:>3}% removes exactly "
        f"Cell 10 ranks 1-{expected_remove}."
    )

# ------------------------------------------------------------
# 16. VERIFY REMOVED RANK NUMBERS
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("REMOVED-RANK NUMBER VERIFICATION")
print("=" * 110)

for _, row in plans_sorted.iterrows():

    percentage = int(
        row["pruning_percent"]
    )

    removed = parsed_plans[
        percentage
    ]["removed"]

    rank_text = str(
        row["removed_ranks"]
    ).strip()

    if rank_text == "":
        raise ValueError(
            f"{percentage}% removed_ranks is empty."
        )

    rank_tokens = [
        token.strip()
        for token in rank_text.split(",")
        if token.strip()
    ]

    try:
        ranks = [
            int(token)
            for token in rank_tokens
        ]
    except ValueError:
        raise ValueError(
            f"{percentage}% contains invalid removed_ranks."
        )

    expected_ranks = [
        authoritative_rank[block]
        for block in removed
    ]

    if ranks != expected_ranks:
        raise ValueError(
            f"{percentage}% removed ranks do not match "
            "the Cell 10 ranking.\n"
            f"Expected: {expected_ranks}\n"
            f"Found:    {ranks}"
        )

    print(
        f"PASS: {percentage:>3}% removed ranks verified."
    )

# ------------------------------------------------------------
# 17. VERIFY NESTED PRUNING
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("NESTED PRUNING VERIFICATION")
print("=" * 110)

previous_removed = set()

for percentage in sorted(parsed_plans):

    current_removed = set(
        parsed_plans[percentage]["removed"]
    )

    if not previous_removed.issubset(
        current_removed
    ):
        raise ValueError(
            f"{percentage}% pruning plan is not nested."
        )

    previous_removed = current_removed

    print(
        f"PASS: {percentage:>3}% contains all "
        "previously removed blocks."
    )

# ------------------------------------------------------------
# 18. VERIFY KEEP SETS
# ------------------------------------------------------------

print("\n" + "=" * 110)
print("KEEP-SET VERIFICATION")
print("=" * 110)

for percentage in sorted(parsed_plans):

    removed = set(
        parsed_plans[percentage]["removed"]
    )

    kept = set(
        parsed_plans[percentage]["kept"]
    )

    expected_kept = (
        EXPECTED_BLOCKS - removed
    )

    if kept != expected_kept:
        raise ValueError(
            f"{percentage}% kept-block set is incorrect."
        )

    print(
        f"PASS: {percentage:>3}% kept set = "
        "B1-B22 minus removed blocks."
    )

# ------------------------------------------------------------
# 19. VERIFY NO DATA WAS ALTERED
# ------------------------------------------------------------

# This cell only reads files and constructs temporary
# in-memory validation objects.

print("\n" + "=" * 110)
print("NON-MUTATION CHECK")
print("=" * 110)

print(
    "PASS: This cell performs verification only."
)

print(
    "PASS: No REVE model loaded."
)

print(
    "PASS: No EEG data loaded."
)

print(
    "PASS: No forward pass performed."
)

print(
    "PASS: No classifier training performed."
)

print(
    "PASS: No pruning performed."
)

# ------------------------------------------------------------
# 20. FINAL VALIDATION SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 120)
print("CELL 12 COMPLETE")
print("=" * 120)

print(
    "PASS: Cell 10 ranking verified."
)

print(
    "PASS: 22 blocks verified."
)

print(
    "PASS: 22 ranks verified."
)

print(
    "PASS: 9 pruning levels verified."
)

print(
    "PASS: Remove/keep counts verified."
)

print(
    "PASS: Removed blocks are valid and unique."
)

print(
    "PASS: Kept blocks are valid and unique."
)

print(
    "PASS: Removed and kept sets are disjoint."
)

print(
    "PASS: Every plan covers exactly B1-B22."
)

print(
    "PASS: Every pruning plan follows Cell 10 ranking."
)

print(
    "PASS: Removed ranks verified."
)

print(
    "PASS: Progressive pruning is nested."
)

print(
    "PASS: No model evaluation performed."
)

print(
    "PASS: No EEG data modified."
)

print(
    "\nREADY FOR CELL 13 — 2,700 PRUNING EVALUATIONS."
)
```

```python
# ============================================================
# CELL 13A — COMPLETE PREFLIGHT VERIFICATION
#             + 2700-JOB MANIFEST GENERATION
#
# IMPORTANT:
#   This cell DOES NOT:
#       - load REVE weights
#       - load EEG arrays
#       - run REVE forward passes
#       - prune any model
#       - train any classifier
#       - evaluate any subject
#
# It ONLY:
#       1. verifies all required inputs/configurations
#       2. verifies all five pruning methods
#       3. verifies the Ours ranking from Cell 10
#       4. verifies pruning plans from Cell 11/12
#       5. verifies Siena subject files
#       6. verifies experiment configuration
#       7. creates the complete 2700-job manifest
#
# The actual 2700 evaluations will be performed in CELL 13B.
# ============================================================

import os
import json
import glob
import hashlib
import numpy as np
import pandas as pd

print("=" * 120)
print("CELL 13A — COMPLETE PREFLIGHT VERIFICATION")
print("=" * 120)

# ============================================================
# 1. GLOBAL PATHS
# ============================================================

BASE_DIR = "/content/drive/MyDrive/Seinna_REVE_LOSO"

NOVELTY3_DIR = os.path.join(
    BASE_DIR,
    "novelty3_pruning"
)

GATE_DIR = os.path.join(
    NOVELTY3_DIR,
    "gate_training"
)

RANKING_DIR = os.path.join(
    NOVELTY3_DIR,
    "rankings"
)

OURS_FILE = os.path.join(
    GATE_DIR,
    "ours_block_scores.csv"
)

PLANS_FILE = os.path.join(
    NOVELTY3_DIR,
    "pruning_plans.csv"
)

# Existing ranking inputs.
DDF_FILE = os.path.join(
    RANKING_DIR,
    "ddf_block_ranking.csv"
)

SSN_FILE = os.path.join(
    RANKING_DIR,
    "ssn_block_ranking.csv"
)

RANDOM_FILE = os.path.join(
    RANKING_DIR,
    "random_rankings.json"
)

# ------------------------------------------------------------
# IMPORTANT:
# Functional ranking must be discovered/verified rather than
# silently assumed to be an old MDR/Ours ranking.
# ------------------------------------------------------------

print("\n" + "=" * 120)
print("PATH DISCOVERY")
print("=" * 120)

print("BASE_DIR:")
print(BASE_DIR)

print("\nNovelty 3 directory:")
print(NOVELTY3_DIR)

print("\nRanking directory:")
print(RANKING_DIR)

# ============================================================
# 2. VERIFY CORE CELL 10 / 11 INPUTS
# ============================================================

print("\n" + "=" * 120)
print("CORE NOVELTY 3 INPUT VERIFICATION")
print("=" * 120)

core_files = {
    "Cell 10 Ours ranking": OURS_FILE,
    "Cell 11 pruning plans": PLANS_FILE,
    "DDF ranking": DDF_FILE,
    "SSN ranking": SSN_FILE,
    "Random rankings": RANDOM_FILE,
}

for name, path in core_files.items():

    if not os.path.isfile(path):
        raise FileNotFoundError(
            f"\n{name} not found:\n{path}"
        )

    print(f"PASS: {name}")
    print(f"      {path}")

# ============================================================
# 3. DISCOVER FUNCTIONAL RANKING FILE
# ============================================================

print("\n" + "=" * 120)
print("FUNCTIONAL RANKING DISCOVERY")
print("=" * 120)

ranking_files = sorted(
    glob.glob(
        os.path.join(
            RANKING_DIR,
            "*"
        )
    )
)

if len(ranking_files) == 0:
    raise FileNotFoundError(
        "No files found in ranking directory:\n"
        + RANKING_DIR
    )

print("Files found in ranking directory:")

for path in ranking_files:
    print(
        "  ",
        os.path.basename(path)
    )

# Candidate names are deliberately broad only for discovery.
functional_candidates = []

for path in ranking_files:

    filename = os.path.basename(
        path
    ).lower()

    if (
        "functional" in filename
        or "function" in filename
    ):
        functional_candidates.append(
            path
        )

print(
    "\nFunctional-like files discovered:",
    len(functional_candidates)
)

for path in functional_candidates:
    print(
        "  ",
        os.path.basename(path)
    )

if len(functional_candidates) == 0:

    raise FileNotFoundError(
        "\nNO FUNCTIONAL RANKING FILE WAS DISCOVERED.\n"
        "Cell 13A will stop here rather than assuming an old "
        "Functional/MDR/Ours ranking."
    )

if len(functional_candidates) > 1:

    raise ValueError(
        "\nMULTIPLE FUNCTIONAL-LIKE RANKING FILES FOUND.\n"
        "The correct one must be identified explicitly before "
        "running 2700 evaluations.\n\n"
        + "\n".join(
            functional_candidates
        )
    )

FUNCTIONAL_FILE = functional_candidates[0]

print(
    "\nPASS: Exactly one Functional ranking candidate found:"
)

print(
    FUNCTIONAL_FILE
)

# ============================================================
# 4. LOAD CORE FILES
# ============================================================

print("\n" + "=" * 120)
print("LOADING RANKING / PLAN FILES")
print("=" * 120)

ours = pd.read_csv(
    OURS_FILE
)

plans = pd.read_csv(
    PLANS_FILE
)

ddf = pd.read_csv(
    DDF_FILE
)

ssn = pd.read_csv(
    SSN_FILE
)

functional = pd.read_csv(
    FUNCTIONAL_FILE
)

with open(
    RANDOM_FILE,
    "r"
) as f:

    random_data = json.load(f)

print(
    f"Ours shape       : {ours.shape}"
)

print(
    f"Plans shape      : {plans.shape}"
)

print(
    f"DDF shape        : {ddf.shape}"
)

print(
    f"SSN shape        : {ssn.shape}"
)

print(
    f"Functional shape : {functional.shape}"
)

print(
    "Random object type:",
    type(random_data).__name__
)

# ============================================================
# 5. VERIFY OURS RANKING
# ============================================================

print("\n" + "=" * 120)
print("OURS RANKING VERIFICATION")
print("=" * 120)

required_ours_columns = [
    "block",
    "ours_score",
    "ours_rank"
]

missing = [
    c
    for c in required_ours_columns
    if c not in ours.columns
]

if missing:
    raise ValueError(
        "Ours file missing columns:\n"
        + "\n".join(missing)
    )

ours["block"] = pd.to_numeric(
    ours["block"],
    errors="coerce"
)

ours["ours_rank"] = pd.to_numeric(
    ours["ours_rank"],
    errors="coerce"
)

ours["ours_score"] = pd.to_numeric(
    ours["ours_score"],
    errors="coerce"
)

if (
    ours[[
        "block",
        "ours_rank",
        "ours_score"
    ]]
    .isna()
    .any()
    .any()
):

    raise ValueError(
        "Ours ranking contains NaN/invalid values."
    )

ours["block"] = (
    ours["block"]
    .astype(int)
)

ours["ours_rank"] = (
    ours["ours_rank"]
    .astype(int)
)

if len(ours) != 22:
    raise ValueError(
        f"Ours ranking must contain 22 rows; "
        f"found {len(ours)}."
    )

if set(ours["block"]) != set(range(1, 23)):
    raise ValueError(
        "Ours ranking does not contain exactly B1-B22."
    )

if set(ours["ours_rank"]) != set(range(1, 23)):
    raise ValueError(
        "Ours ranking does not contain ranks 1-22."
    )

if ours["block"].duplicated().any():
    raise ValueError(
        "Duplicate blocks in Ours ranking."
    )

if ours["ours_rank"].duplicated().any():
    raise ValueError(
        "Duplicate ranks in Ours ranking."
    )

ours_ranked = (
    ours
    .sort_values("ours_rank")
    .reset_index(drop=True)
)

ours_order = (
    ours_ranked["block"]
    .tolist()
)

print(
    "PASS: Ours contains exactly 22 blocks."
)

print(
    "PASS: Ours ranks 1-22 are unique."
)

print(
    "Ours removal order:"
)

print(
    " -> ".join(
        f"B{x}"
        for x in ours_order
    )
)

# ============================================================
# 6. VERIFY PRUNING PLANS
# ============================================================

print("\n" + "=" * 120)
print("PRUNING PLAN VERIFICATION")
print("=" * 120)

expected_plans = {
    10: 2,
    20: 4,
    30: 7,
    40: 9,
    50: 11,
    60: 13,
    70: 15,
    80: 18,
    90: 20
}

required_plan_columns = [
    "pruning_percent",
    "remove_count",
    "keep_count",
    "removed_blocks",
    "removed_ranks",
    "kept_blocks"
]

missing = [
    c
    for c in required_plan_columns
    if c not in plans.columns
]

if missing:
    raise ValueError(
        "Pruning plan file missing columns:\n"
        + "\n".join(missing)
    )

if len(plans) != 9:
    raise ValueError(
        f"Expected 9 pruning plans; found {len(plans)}."
    )

plans["pruning_percent"] = pd.to_numeric(
    plans["pruning_percent"],
    errors="coerce"
).astype(int)

if set(plans["pruning_percent"]) != set(
    expected_plans.keys()
):
    raise ValueError(
        "Incorrect pruning percentages."
    )

plans = (
    plans
    .sort_values("pruning_percent")
    .reset_index(drop=True)
)

print(
    "PASS: Nine pruning levels found."
)

for _, row in plans.iterrows():

    pct = int(
        row["pruning_percent"]
    )

    remove_count = int(
        row["remove_count"]
    )

    keep_count = int(
        row["keep_count"]
    )

    if remove_count != expected_plans[pct]:
        raise ValueError(
            f"{pct}% remove count incorrect."
        )

    if remove_count + keep_count != 22:
        raise ValueError(
            f"{pct}% remove + keep != 22."
        )

    print(
        f"PASS: {pct:>3}% -> "
        f"remove {remove_count}, "
        f"keep {keep_count}"
    )

# ============================================================
# 7. VERIFY OURS PLAN EXACTLY MATCHES CELL 10
# ============================================================

print("\n" + "=" * 120)
print("OURS ↔ PRUNING PLAN CONSISTENCY")
print("=" * 120)

for _, row in plans.iterrows():

    pct = int(
        row["pruning_percent"]
    )

    remove_count = expected_plans[pct]

    expected_removed = (
        ours_order[:remove_count]
    )

    actual_removed = [
        int(
            str(token).strip()[1:]
        )
        for token
        in str(
            row["removed_blocks"]
        ).split(",")
        if str(token).strip()
    ]

    if actual_removed != expected_removed:
        raise ValueError(
            f"{pct}% pruning plan does not "
            "match Cell 10 Ours ranking."
        )

    print(
        f"PASS: {pct:>3}% exactly matches "
        f"Ours ranks 1-{remove_count}."
    )

# ============================================================
# 8. GENERIC RANKING VERIFICATION FUNCTION
# ============================================================

def detect_ranking_columns(
    df,
    method_name
):

    block_candidates = [
        "block",
        "removed_block",
        "pruned_block"
    ]

    rank_candidates = [
        "rank",
        "ranking",
        "ours_rank",
        "functional_rank",
        "ddf_rank",
        "ssn_rank",
        "random_rank"
    ]

    block_col = None

    for col in block_candidates:

        if col in df.columns:
            block_col = col
            break

    if block_col is None:
        raise ValueError(
            f"{method_name}: could not identify "
            "block column.\n"
            f"Columns: {df.columns.tolist()}"
        )

    rank_col = None

    for col in rank_candidates:

        if col in df.columns:
            rank_col = col
            break

    if rank_col is None:

        # If there is no explicit rank column,
        # inspect whether the rows themselves represent
        # a complete ordered ranking.
        if len(df) == 22:
            rank_col = None
        else:
            raise ValueError(
                f"{method_name}: could not identify "
                "ranking column.\n"
                f"Columns: {df.columns.tolist()}"
            )

    return block_col, rank_col


# ============================================================
# 9. VERIFY FUNCTIONAL / DDF / SSN RANKINGS
# ============================================================

def verify_single_ranking(
    df,
    method_name
):

    print(
        f"\n--- {method_name} ---"
    )

    block_col, rank_col = (
        detect_ranking_columns(
            df,
            method_name
        )
    )

    print(
        f"Block column : {block_col}"
    )

    print(
        f"Rank column  : {rank_col}"
    )

    blocks = pd.to_numeric(
        df[block_col],
        errors="coerce"
    )

    if blocks.isna().any():
        raise ValueError(
            f"{method_name}: invalid block values."
        )

    blocks = (
        blocks
        .astype(int)
        .tolist()
    )

    if len(blocks) != 22:
        raise ValueError(
            f"{method_name}: expected 22 rows; "
            f"found {len(blocks)}."
        )

    if set(blocks) != set(range(1, 23)):
        raise ValueError(
            f"{method_name}: blocks are not exactly B1-B22."
        )

    if len(set(blocks)) != 22:
        raise ValueError(
            f"{method_name}: duplicate blocks."
        )

    if rank_col is not None:

        ranks = pd.to_numeric(
            df[rank_col],
            errors="coerce"
        )

        if ranks.isna().any():
            raise ValueError(
                f"{method_name}: invalid rank values."
            )

        ranks = (
            ranks
            .astype(int)
            .tolist()
        )

        if set(ranks) != set(range(1, 23)):
            raise ValueError(
                f"{method_name}: ranks are not 1-22."
            )

        if len(set(ranks)) != 22:
            raise ValueError(
                f"{method_name}: duplicate ranks."
            )

        order = (
            pd.DataFrame({
                "block": blocks,
                "rank": ranks
            })
            .sort_values("rank")
            ["block"]
            .tolist()
        )

    else:

        # No explicit rank column means row order is the
        # only possible ranking representation.
        order = blocks

    print(
        f"PASS: {method_name} contains a complete "
        "B1-B22 ranking."
    )

    print(
        f"{method_name} order:"
    )

    print(
        " -> ".join(
            f"B{x}"
            for x in order
        )
    )

    return order


functional_order = verify_single_ranking(
    functional,
    "Functional"
)

ddf_order = verify_single_ranking(
    ddf,
    "DDF"
)

ssn_order = verify_single_ranking(
    ssn,
    "SSN"
)

# ============================================================
# 10. RANDOM RANKING VERIFICATION
# ============================================================

print("\n" + "=" * 120)
print("RANDOM RANKING VERIFICATION")
print("=" * 120)

random_orders = {}

# ------------------------------------------------------------
# Case A: dictionary representation
# ------------------------------------------------------------

if isinstance(
    random_data,
    dict
):

    print(
        "Random representation: dictionary"
    )

    print(
        "Random keys:"
    )

    for key in random_data.keys():
        print(
            "  ",
            key
        )

    # Attempt to identify seed -> ranking mappings.
    for key, value in random_data.items():

        try:
            seed = int(key)
        except Exception:
            continue

        if isinstance(
            value,
            list
        ):

            candidate = value

        elif isinstance(
            value,
            dict
        ):

            candidate = None

            for possible_key in [
                "blocks",
                "ranking",
                "order",
                "block_order"
            ]:

                if possible_key in value:
                    candidate = value[
                        possible_key
                    ]
                    break

            if candidate is None:
                continue

        else:
            continue

        try:

            candidate = [
                int(x)
                for x in candidate
            ]

        except Exception:
            continue

        if (
            len(candidate) == 22
            and set(candidate)
            == set(range(1,23))
        ):

            random_orders[
                seed
            ] = candidate

# ------------------------------------------------------------
# Case B: list representation
# ------------------------------------------------------------

elif isinstance(
    random_data,
    list
):

    print(
        "Random representation: list"
    )

    # Possible list of ranking records.
    for item in random_data:

        if not isinstance(
            item,
            dict
        ):
            continue

        seed_value = (
            item.get("seed")
        )

        ranking_value = None

        for possible_key in [
            "blocks",
            "ranking",
            "order",
            "block_order"
        ]:

            if possible_key in item:
                ranking_value = item[
                    possible_key
                ]
                break

        if (
            seed_value is None
            or ranking_value is None
        ):
            continue

        try:

            seed = int(
                seed_value
            )

            candidate = [
                int(x)
                for x in ranking_value
            ]

        except Exception:
            continue

        if (
            len(candidate) == 22
            and set(candidate)
            == set(range(1,23))
        ):

            random_orders[
                seed
            ] = candidate

else:

    raise ValueError(
        "Unsupported random_rankings.json format:\n"
        + type(random_data).__name__
    )

if len(random_orders) == 0:

    raise ValueError(
        "Could not extract any valid random B1-B22 ranking."
    )

print(
    "\nExtracted random rankings:",
    sorted(random_orders.keys())
)

for seed, order in sorted(
    random_orders.items()
):

    print(
        f"PASS: Random seed {seed} "
        "contains exactly B1-B22."
    )

    print(
        "      "
        + " -> ".join(
            f"B{x}"
            for x in order
        )
    )

# ------------------------------------------------------------
# Determine whether expected classifier seeds are represented.
# ------------------------------------------------------------

CLASSIFIER_SEEDS = [
    20,
    25,
    30,
    35,
    40
]

random_seed_overlap = (
    sorted(
        set(
            CLASSIFIER_SEEDS
        ).intersection(
            random_orders.keys()
        )
    )
)

print(
    "\nRandom rankings matching classifier seeds:",
    random_seed_overlap
)

# We do NOT automatically require random ranking seeds to equal
# classifier seeds. We first report the actual representation.
# Cell 13B will use the verified random-ranking protocol.
#
# If random rankings have exactly five entries but unrelated
# seeds, those seeds will be preserved rather than renamed.

# ============================================================
# 11. VERIFY FUNCTIONAL RANKING IS NOT THE OLD OURS RANKING
# ============================================================

print("\n" + "=" * 120)
print("FUNCTIONAL ↔ OURS INDEPENDENCE CHECK")
print("=" * 120)

if functional_order == ours_order:

    raise ValueError(
        "\nFunctional ranking is IDENTICAL to the Ours ranking.\n"
        "This may indicate that an old Ours/MDR ranking was "
        "mistakenly supplied as Functional.\n"
        "Cell 13A stops before any evaluation."
    )

print(
    "PASS: Functional ranking is not identical "
    "to the Cell 10 Ours ranking."
)

# ============================================================
# 12. VERIFY DDF / SSN ARE NOT ACCIDENTALLY OURS
# ============================================================

if ddf_order == ours_order:

    print(
        "WARNING: DDF ordering is identical to Ours."
    )

    print(
        "This is not automatically an error, but must be "
        "reported for methodological inspection."
    )

else:

    print(
        "PASS: DDF ordering differs from Ours."
    )

if ssn_order == ours_order:

    print(
        "WARNING: SSN ordering is identical to Ours."
    )

    print(
        "This is not automatically an error, but must be "
        "reported for methodological inspection."
    )

else:

    print(
        "PASS: SSN ordering differs from Ours."
    )

# ============================================================
# 13. SIENA DATA DISCOVERY
# ============================================================

print("\n" + "=" * 120)
print("SIENA DATA DISCOVERY")
print("=" * 120)

DATA_DIR = os.path.join(
    BASE_DIR,
    "subject_epochs_22ch"
)

if not os.path.isdir(DATA_DIR):
    raise FileNotFoundError(
        "Siena subject directory not found:\n"
        + DATA_DIR
    )

print(
    "PASS: Siena subject directory exists."
)

print(
    DATA_DIR
)

EXPECTED_SUBJECTS = [
    "PN00",
    "PN01",
    "PN06",
    "PN07",
    "PN09",
    "PN10",
    "PN11",
    "PN12",
    "PN13",
    "PN14",
    "PN16",
    "PN17"
]

print(
    "\nExpected LOSO subjects:"
)

print(
    EXPECTED_SUBJECTS
)

# ------------------------------------------------------------
# Discover subject files without loading arrays.
# ------------------------------------------------------------

all_data_files = sorted(
    glob.glob(
        os.path.join(
            DATA_DIR,
            "*"
        )
    )
)

print(
    "\nFiles discovered:",
    len(all_data_files)
)

for path in all_data_files:
    print(
        "  ",
        os.path.basename(path)
    )

# ------------------------------------------------------------
# Match subject IDs to files.
# ------------------------------------------------------------

subject_files = {}

for subject in EXPECTED_SUBJECTS:

    matches = [
        path
        for path in all_data_files
        if subject.lower()
        in os.path.basename(
            path
        ).lower()
    ]

    if len(matches) == 0:

        raise FileNotFoundError(
            f"No data file found for {subject}."
        )

    if len(matches) > 1:

        raise ValueError(
            f"Multiple candidate files found for {subject}:\n"
            + "\n".join(matches)
        )

    subject_files[
        subject
    ] = matches[0]

    print(
        f"PASS: {subject} -> "
        f"{os.path.basename(matches[0])}"
    )

# ============================================================
# 14. DATA FORMAT VERIFICATION
# ============================================================

print("\n" + "=" * 120)
print("SIENA DATA FORMAT VERIFICATION")
print("=" * 120)

print(
    "This section reads metadata/arrays only for verification."
)

subject_metadata = {}

for subject in EXPECTED_SUBJECTS:

    path = subject_files[
        subject
    ]

    if not path.lower().endswith(
        ".npz"
    ):

        raise ValueError(
            f"{subject}: expected NPZ file, found:\n{path}"
        )

    with np.load(
        path,
        allow_pickle=False
    ) as data:

        keys = list(
            data.keys()
        )

        if "X" not in keys:
            raise ValueError(
                f"{subject}: NPZ missing X."
            )

        if "y" not in keys:
            raise ValueError(
                f"{subject}: NPZ missing y."
            )

        X = data["X"]
        y = data["y"]

        if X.ndim != 3:
            raise ValueError(
                f"{subject}: expected X.ndim=3; "
                f"found {X.ndim}."
            )

        if X.shape[1:] != (
            1000,
            22
        ):

            raise ValueError(
                f"{subject}: expected X shape "
                "(N,1000,22); "
                f"found {X.shape}."
            )

        if y.ndim != 1:
            raise ValueError(
                f"{subject}: expected y.ndim=1; "
                f"found {y.ndim}."
            )

        if len(X) != len(y):
            raise ValueError(
                f"{subject}: X/y length mismatch."
            )

        if not np.all(
            np.isfinite(X)
        ):

            raise ValueError(
                f"{subject}: X contains NaN/Inf."
            )

        if not np.all(
            np.isfinite(y)
        ):

            raise ValueError(
                f"{subject}: y contains NaN/Inf."
            )

        unique_y = sorted(
            np.unique(y).tolist()
        )

        if not set(unique_y).issubset(
            {0,1}
        ):

            raise ValueError(
                f"{subject}: unexpected labels "
                f"{unique_y}."
            )

        subject_metadata[
            subject
        ] = {
            "path": path,
            "n_epochs": int(
                len(X)
            ),
            "shape": tuple(
                X.shape
            ),
            "label_values": unique_y
        }

        print(
            f"PASS: {subject} | "
            f"X={X.shape} | "
            f"labels={unique_y}"
        )

print(
    "\nPASS: All 12 Siena files have "
    "the expected (N,1000,22) structure."
)

# ============================================================
# 15. VERIFY TOTAL SIENA DATASET
# ============================================================

total_epochs = sum(
    item["n_epochs"]
    for item
    in subject_metadata.values()
)

print(
    "\nTotal epochs across 12 subjects:",
    total_epochs
)

if total_epochs != 1007:

    print(
        "WARNING: Total epoch count differs from "
        "the previously established 1007."
    )

    print(
        "This is reported rather than silently corrected."
    )

else:

    print(
        "PASS: Total epochs = 1007."
    )

# ============================================================
# 16. CHANNEL CONFIGURATION
# ============================================================

print("\n" + "=" * 120)
print("REVE CHANNEL CONFIGURATION VERIFICATION")
print("=" * 120)

EEG_CHANNELS_22 = [
    "Fp1",
    "F3",
    "C3",
    "P3",
    "O1",
    "F7",
    "T5",
    "Fc1",
    "Fc5",
    "Cp1",
    "Cp5",
    "F9",
    "Fz",
    "Pz",
    "F4",
    "C4",
    "P4",
    "O2",
    "F8",
    "T4",
    "T6",
    "Fc2"
]

if len(
    EEG_CHANNELS_22
) != 22:

    raise ValueError(
        "Expected exactly 22 EEG channels."
    )

if len(
    set(
        EEG_CHANNELS_22
    )
) != 22:

    raise ValueError(
        "Duplicate EEG channel names."
    )

print(
    "PASS: Exactly 22 unique EEG channels."
)

print(
    EEG_CHANNELS_22
)

# ============================================================
# 17. REVE CONFIGURATION — NO MODEL LOAD
# ============================================================

print("\n" + "=" * 120)
print("REVE CONFIGURATION VERIFICATION")
print("=" * 120)

REVE_MODEL_NAME = (
    "brain-bzh/reve-base"
)

REVE_POSITION_MODEL_NAME = (
    "brain-bzh/reve-positions"
)

EXPECTED_REVE_BLOCKS = 22
EXPECTED_REVE_EMBED_DIM = 512
EXPECTED_PATCHES = 5
EXPECTED_SAMPLE_RATE = 200
EXPECTED_EPOCH_SECONDS = 5

print(
    "REVE model:",
    REVE_MODEL_NAME
)

print(
    "Position model:",
    REVE_POSITION_MODEL_NAME
)

print(
    "Expected blocks:",
    EXPECTED_REVE_BLOCKS
)

print(
    "Expected embedding dimension:",
    EXPECTED_REVE_EMBED_DIM
)

print(
    "Expected patches:",
    EXPECTED_PATCHES
)

print(
    "Expected sample rate:",
    EXPECTED_SAMPLE_RATE
)

print(
    "Expected epoch duration:",
    EXPECTED_EPOCH_SECONDS,
    "seconds"
)

if EXPECTED_REVE_BLOCKS != 22:
    raise ValueError(
        "REVE block configuration incorrect."
    )

if EXPECTED_REVE_EMBED_DIM != 512:
    raise ValueError(
        "REVE embedding dimension incorrect."
    )

if EXPECTED_PATCHES != 5:
    raise ValueError(
        "Expected five REVE output patches."
    )

print(
    "\nPASS: Locked REVE configuration matches "
    "the established experiment."
)

# ============================================================
# 18. CLASSIFIER CONFIGURATION
# ============================================================

print("\n" + "=" * 120)
print("CLASSIFIER CONFIGURATION VERIFICATION")
print("=" * 120)

CLASSIFIER_INPUT = 1024
CLASSIFIER_HIDDEN_1 = 512
CLASSIFIER_HIDDEN_2 = 128
CLASSIFIER_OUTPUT = 2

DROPOUT = 0.2
LEARNING_RATE = 1e-3
WEIGHT_DECAY = 1e-4
BATCH_SIZE = 32
MAX_EPOCHS = 100
PATIENCE = 15

CLASSIFIER_SEEDS = [
    20,
    25,
    30,
    35,
    40
]

print(
    "Classifier:",
    "Linear(1024,512) -> ReLU -> Dropout(0.2)"
)

print(
    "           ",
    "-> Linear(512,128) -> ReLU -> Dropout(0.2)"
)

print(
    "           ",
    "-> Linear(128,2)"
)

print(
    "\nBatch size:",
    BATCH_SIZE
)

print(
    "Learning rate:",
    LEARNING_RATE
)

print(
    "Weight decay:",
    WEIGHT_DECAY
)

print(
    "Max epochs:",
    MAX_EPOCHS
)

print(
    "Patience:",
    PATIENCE
)

print(
    "Classifier seeds:",
    CLASSIFIER_SEEDS
)

if CLASSIFIER_INPUT != 1024:
    raise ValueError(
        "Classifier input must be 1024."
    )

if CLASSIFIER_OUTPUT != 2:
    raise ValueError(
        "Classifier output must be 2."
    )

if DROPOUT != 0.2:
    raise ValueError(
        "Dropout must be 0.2."
    )

if CLASSIFIER_SEEDS != [
    20,
    25,
    30,
    35,
    40
]:

    raise ValueError(
        "Classifier seeds changed."
    )

# ------------------------------------------------------------
# Parameter-count verification
# ------------------------------------------------------------

expected_params = (
    (1024 * 512 + 512)
    + (512 * 128 + 128)
    + (128 * 2 + 2)
)

print(
    "\nExpected classifier parameters:",
    expected_params
)

if expected_params != 590722:

    raise ValueError(
        "Classifier parameter count does not equal "
        "the locked 590,722."
    )

print(
    "PASS: Classifier parameter count = 590,722."
)

# ============================================================
# 19. FEATURE REPRESENTATION VERIFICATION
# ============================================================

print("\n" + "=" * 120)
print("FEATURE REPRESENTATION VERIFICATION")
print("=" * 120)

print(
    "Expected REVE output:"
)

print(
    "  (B,22,5,512)"
)

print(
    "Expected pooling:"
)

print(
    "  mean over 5 patches -> (B,22,512)"
)

print(
    "  channel mean + channel std -> (B,1024)"
)

print(
    "Expected classifier input:",
    "(B,1024)"
)

print(
    "\nPASS: Feature representation configuration locked."
)

# ============================================================
# 20. LOSO CONFIGURATION
# ============================================================

print("\n" + "=" * 120)
print("LOSO CONFIGURATION VERIFICATION")
print("=" * 120)

if len(
    EXPECTED_SUBJECTS
) != 12:

    raise ValueError(
        "Expected exactly 12 LOSO subjects."
    )

if len(
    set(
        EXPECTED_SUBJECTS
    )
) != 12:

    raise ValueError(
        "Duplicate LOSO subject."
    )

print(
    "PASS: Exactly 12 LOSO subjects."
)

for subject in EXPECTED_SUBJECTS:
    print(
        "  ",
        subject
    )

# ============================================================
# 21. METHOD COUNT
# ============================================================

print("\n" + "=" * 120)
print("FIVE-METHOD EXPERIMENT VERIFICATION")
print("=" * 120)

METHODS = [
    "Random",
    "Functional",
    "DDF",
    "SSN",
    "Ours"
]

if len(METHODS) != 5:
    raise ValueError(
        "Expected exactly five pruning methods."
    )

print(
    "Methods:"
)

for method in METHODS:
    print(
        "  ",
        method
    )

print(
    "\nPASS: Five pruning methods confirmed."
)

# ============================================================
# 22. VERIFY ALL METHOD ORDERS
# ============================================================

method_orders = {
    "Random": random_orders,
    "Functional": functional_order,
    "DDF": ddf_order,
    "SSN": ssn_order,
    "Ours": ours_order
}

print("\n" + "=" * 120)
print("ALL METHOD ORDER VERIFICATION")
print("=" * 120)

print(
    "PASS: Random ranking(s) extracted."
)

print(
    "PASS: Functional ranking extracted."
)

print(
    "PASS: DDF ranking extracted."
)

print(
    "PASS: SSN ranking extracted."
)

print(
    "PASS: Ours ranking extracted."
)

# ============================================================
# 23. EXPERIMENT COUNT
# ============================================================

print("\n" + "=" * 120)
print("2700-JOB COUNT VERIFICATION")
print("=" * 120)

N_METHODS = 5
N_PRUNING_LEVELS = 9
N_SUBJECTS = 12
N_CLASSIFIER_SEEDS = 5

TOTAL_JOBS = (
    N_METHODS
    * N_PRUNING_LEVELS
    * N_SUBJECTS
    * N_CLASSIFIER_SEEDS
)

print(
    "Methods:",
    N_METHODS
)

print(
    "Pruning levels:",
    N_PRUNING_LEVELS
)

print(
    "LOSO subjects:",
    N_SUBJECTS
)

print(
    "Classifier seeds:",
    N_CLASSIFIER_SEEDS
)

print(
    "\nTotal:"
)

print(
    f"{N_METHODS} × "
    f"{N_PRUNING_LEVELS} × "
    f"{N_SUBJECTS} × "
    f"{N_CLASSIFIER_SEEDS}"
    f" = {TOTAL_JOBS}"
)

if TOTAL_JOBS != 2700:

    raise ValueError(
        f"Expected 2700 jobs; calculated {TOTAL_JOBS}."
    )

print(
    "\nPASS: Exactly 2700 evaluations."
)

# ============================================================
# 24. BUILD JOB MANIFEST
# ============================================================

print("\n" + "=" * 120)
print("BUILDING 2700-JOB MANIFEST")
print("=" * 120)

manifest_rows = []

job_id = 0

for method in METHODS:

    for _, plan_row in plans.iterrows():

        pruning_percent = int(
            plan_row["pruning_percent"]
        )

        remove_count = expected_plans[
            pruning_percent
        ]

        # ----------------------------------------------------
        # Ours uses Cell 10 order.
        # Other deterministic methods use their verified order.
        # Random requires explicit handling below.
        # ----------------------------------------------------

        if method == "Ours":

            order_list = ours_order

            selected_order_id = (
                "ours_cell10"
            )

            method_rank_seed = None

            method_rank_type = (
                "deterministic"
            )

            order_to_use = (
                order_list
            )

            subject_specific_orders = {
                "ALL": order_to_use
            }

        elif method == "Functional":

            order_to_use = (
                functional_order
            )

            selected_order_id = (
                "functional_verified"
            )

            method_rank_seed = None

            method_rank_type = (
                "deterministic"
            )

            subject_specific_orders = {
                "ALL": order_to_use
            }

        elif method == "DDF":

            order_to_use = (
                ddf_order
            )

            selected_order_id = (
                "ddf_verified"
            )

            method_rank_seed = None

            method_rank_type = (
                "deterministic"
            )

            subject_specific_orders = {
                "ALL": order_to_use
            }

        elif method == "SSN":

            order_to_use = (
                ssn_order
            )

            selected_order_id = (
                "ssn_verified"
            )

            method_rank_seed = None

            method_rank_type = (
                "deterministic"
            )

            subject_specific_orders = {
                "ALL": order_to_use
            }

        elif method == "Random":

            # Do not silently choose a random order.
            #
            # A separate job will be created for each verified
            # random ranking only if its protocol is established.
            #
            # Since the requested experiment has exactly
            # 5 classifier seeds, we require five random rankings
            # so each classifier seed can be paired deterministically
            # with one random ranking.
            #
            # If the file contains another number of rankings,
            # stop rather than silently changing the experiment.

            if len(
                random_orders
            ) != 5:

                raise ValueError(
                    "\nRandom ranking file contains "
                    f"{len(random_orders)} rankings.\n"
                    "Expected exactly 5 for the planned "
                    "5-seed random-pruning protocol.\n"
                    "Cell 13A stops rather than assuming "
                    "a mapping."
                )

            method_rank_type = (
                "seed_specific"
            )

            selected_order_id = (
                "random_seed_specific"
            )

            method_rank_seed = None

            order_to_use = None

            subject_specific_orders = None

        else:

            raise ValueError(
                f"Unknown method: {method}"
            )

        # ----------------------------------------------------
        # Create classifier-seed jobs.
        # ----------------------------------------------------

        for subject in EXPECTED_SUBJECTS:

            for classifier_seed in CLASSIFIER_SEEDS:

                job_id += 1

                if method == "Random":

                    random_seed_keys = sorted(
                        random_orders.keys()
                    )

                    # Deterministic mapping:
                    # classifier seed position -> random
                    # ranking position.
                    #
                    # This does NOT rename random seeds.
                    # It records both explicitly.

                    random_index = (
                        CLASSIFIER_SEEDS.index(
                            classifier_seed
                        )
                    )

                    random_ranking_seed = (
                        random_seed_keys[
                            random_index
                        ]
                    )

                    ranking_order = (
                        random_orders[
                            random_ranking_seed
                        ]
                    )

                    removed_blocks = (
                        ranking_order[
                            :remove_count
                        ]
                    )

                else:

                    ranking_order = (
                        order_to_use
                    )

                    removed_blocks = (
                        ranking_order[
                            :remove_count
                        ]
                    )

                    random_ranking_seed = None

                kept_blocks = [
                    block
                    for block in range(
                        1,
                        23
                    )
                    if block
                    not in set(
                        removed_blocks
                    )
                ]

                manifest_rows.append(
                    {
                        "job_id": job_id,

                        "method": method,

                        "pruning_percent":
                            pruning_percent,

                        "remove_count":
                            remove_count,

                        "keep_count":
                            22 - remove_count,

                        "subject":
                            subject,

                        "classifier_seed":
                            classifier_seed,

                        "random_ranking_seed":
                            random_ranking_seed,

                        "ranking_type":
                            method_rank_type,

                        "ranking_source":
                            selected_order_id,

                        "removed_blocks":
                            ",".join(
                                f"B{x}"
                                for x
                                in removed_blocks
                            ),

                        "kept_blocks":
                            ",".join(
                                f"B{x}"
                                for x
                                in kept_blocks
                            )
                    }
                )

# ============================================================
# 25. MANIFEST DATAFRAME
# ============================================================

manifest = pd.DataFrame(
    manifest_rows
)

print(
    "Manifest shape:",
    manifest.shape
)

if len(manifest) != 2700:

    raise ValueError(
        f"Expected 2700 manifest rows; "
        f"found {len(manifest)}."
    )

print(
    "PASS: Manifest contains exactly 2700 jobs."
)

# ============================================================
# 26. MANIFEST JOB ID VERIFICATION
# ============================================================

print("\n" + "=" * 120)
print("MANIFEST JOB-ID VERIFICATION")
print("=" * 120)

if manifest["job_id"].duplicated().any():

    raise ValueError(
        "Duplicate job IDs found."
    )

if set(
    manifest["job_id"]
) != set(
    range(
        1,
        2701
    )
):

    raise ValueError(
        "Job IDs are not exactly 1-2700."
    )

print(
    "PASS: Job IDs 1-2700 are unique and complete."
)

# ============================================================
# 27. MANIFEST METHOD COUNTS
# ============================================================

print("\n" + "=" * 120)
print("MANIFEST METHOD COUNT VERIFICATION")
print("=" * 120)

method_counts = (
    manifest[
        "method"
    ]
    .value_counts()
    .sort_index()
)

print(
    method_counts
)

expected_per_method = (
    9
    * 12
    * 5
)

for method in METHODS:

    count = int(
        method_counts.get(
            method,
            0
        )
    )

    if count != expected_per_method:

        raise ValueError(
            f"{method}: expected "
            f"{expected_per_method} jobs; "
            f"found {count}."
        )

    print(
        f"PASS: {method} = {count} jobs."
    )

# ============================================================
# 28. MANIFEST PRUNING-LEVEL COUNTS
# ============================================================

print("\n" + "=" * 120)
print("MANIFEST PRUNING-LEVEL VERIFICATION")
print("=" * 120)

level_counts = (
    manifest
    .groupby(
        "pruning_percent"
    )
    .size()
)

expected_per_level = (
    5
    * 12
    * 5
)

for pct in expected_plans:

    count = int(
        level_counts.get(
            pct,
            0
        )
    )

    if count != expected_per_level:

        raise ValueError(
            f"{pct}%: expected "
            f"{expected_per_level} jobs; "
            f"found {count}."
        )

    print(
        f"PASS: {pct:>3}% = {count} jobs."
    )

# ============================================================
# 29. VERIFY EACH JOB'S REMOVED BLOCK COUNT
# ============================================================

print("\n" + "=" * 120)
print("MANIFEST BLOCK-COUNT VERIFICATION")
print("=" * 120)

for _, row in manifest.iterrows():

    pct = int(
        row["pruning_percent"]
    )

    removed = [
        token
        for token
        in str(
            row["removed_blocks"]
        ).split(",")
        if token
    ]

    kept = [
        token
        for token
        in str(
            row["kept_blocks"]
        ).split(",")
        if token
    ]

    expected_remove = (
        expected_plans[pct]
    )

    expected_keep = (
        22 - expected_remove
    )

    if len(removed) != expected_remove:

        raise ValueError(
            f"Job {row['job_id']}: "
            f"wrong removed count."
        )

    if len(kept) != expected_keep:

        raise ValueError(
            f"Job {row['job_id']}: "
            f"wrong kept count."
        )

    if set(removed).intersection(
        set(kept)
    ):

        raise ValueError(
            f"Job {row['job_id']}: "
            "removed/kept overlap."
        )

print(
    "PASS: All 2700 jobs have correct "
    "remove/keep counts."
)

# ============================================================
# 30. VERIFY ALL JOBS COVER B1-B22
# ============================================================

for _, row in manifest.iterrows():

    removed = set(
        str(
            row["removed_blocks"]
        ).split(",")
    )

    kept = set(
        str(
            row["kept_blocks"]
        ).split(",")
    )

    all_blocks = {
        f"B{x}"
        for x in range(
            1,
            23
        )
    }

    if (
        removed.union(kept)
        !=
        all_blocks
    ):

        raise ValueError(
            f"Job {row['job_id']}: "
            "removed + kept does not cover B1-B22."
        )

print(
    "PASS: Every job covers exactly B1-B22."
)

# ============================================================
# 31. VERIFY RANDOM MAPPING
# ============================================================

print("\n" + "=" * 120)
print("RANDOM RANKING ↔ CLASSIFIER SEED MAPPING")
print("=" * 120)

random_seed_keys = sorted(
    random_orders.keys()
)

for index, classifier_seed in enumerate(
    CLASSIFIER_SEEDS
):

    mapped_random_seed = (
        random_seed_keys[index]
    )

    print(
        f"Classifier seed {classifier_seed}"
        f" -> random ranking seed "
        f"{mapped_random_seed}"
    )

print(
    "\nPASS: Random ranking mapping is deterministic "
    "for the manifest."
)

# ============================================================
# 32. VERIFY NO RANDOM RANKING CHANGES BY SUBJECT
# ============================================================

random_manifest = manifest[
    manifest["method"] == "Random"
]

for classifier_seed in CLASSIFIER_SEEDS:

    subset = random_manifest[
        random_manifest[
            "classifier_seed"
        ] == classifier_seed
    ]

    unique_random_seeds = (
        subset[
            "random_ranking_seed"
        ]
        .unique()
        .tolist()
    )

    if len(
        unique_random_seeds
    ) != 1:

        raise ValueError(
            "Random ranking assignment changes "
            "across subjects for classifier seed "
            f"{classifier_seed}."
        )

print(
    "PASS: Random ranking assignment is consistent "
    "across all 12 LOSO subjects."
)

# ============================================================
# 33. MANIFEST FILE PATHS
# ============================================================

MANIFEST_FILE = os.path.join(
    NOVELTY3_DIR,
    "pruning_2700_job_manifest.csv"
)

MANIFEST_JSON = os.path.join(
    NOVELTY3_DIR,
    "pruning_2700_job_manifest.json"
)

# ============================================================
# 34. SAVE MANIFEST
# ============================================================

print("\n" + "=" * 120)
print("SAVING 2700-JOB MANIFEST")
print("=" * 120)

manifest.to_csv(
    MANIFEST_FILE,
    index=False
)

manifest.to_json(
    MANIFEST_JSON,
    orient="records",
    indent=2
)

print(
    "Saved CSV:"
)

print(
    MANIFEST_FILE
)

print(
    "\nSaved JSON:"
)

print(
    MANIFEST_JSON
)

# ============================================================
# 35. MANIFEST HASH
# ============================================================

print("\n" + "=" * 120)
print("MANIFEST INTEGRITY")
print("=" * 120)

with open(
    MANIFEST_FILE,
    "rb"
) as f:

    manifest_hash = hashlib.sha256(
        f.read()
    ).hexdigest()

print(
    "SHA-256:"
)

print(
    manifest_hash
)

# ============================================================
# 36. FINAL PREFLIGHT SUMMARY
# ============================================================

print("\n" + "=" * 130)
print("CELL 13A — FINAL PREFLIGHT SUMMARY")
print("=" * 130)

print(
    "PASS: Cell 10 Ours ranking verified."
)

print(
    "PASS: Cell 11/12 pruning plans verified."
)

print(
    "PASS: Functional ranking verified."
)

print(
    "PASS: DDF ranking verified."
)

print(
    "PASS: SSN ranking verified."
)

print(
    "PASS: Random ranking(s) verified."
)

print(
    "PASS: Functional ranking is independent "
    "of Ours ranking."
)

print(
    "PASS: 12 LOSO subjects verified."
)

print(
    "PASS: Siena X shape verified as (N,1000,22)."
)

print(
    "PASS: Siena labels verified."
)

print(
    "PASS: 22-channel REVE configuration verified."
)

print(
    "PASS: 22 REVE blocks configured."
)

print(
    "PASS: 512-dimensional REVE embedding configured."
)

print(
    "PASS: Classifier architecture verified."
)

print(
    "PASS: Classifier parameter count = 590,722."
)

print(
    "PASS: Classifier seeds = [20,25,30,35,40]."
)

print(
    "PASS: Five pruning methods verified."
)

print(
    "PASS: Nine pruning levels verified."
)

print(
    "PASS: 12 LOSO subjects verified."
)

print(
    "PASS: 2700 = 5 × 9 × 12 × 5."
)

print(
    "PASS: Manifest contains exactly 2700 unique jobs."
)

print(
    "PASS: Every job has valid removed/kept blocks."
)

print(
    "PASS: No model evaluation performed."
)

print(
    "PASS: No REVE forward pass performed."
)

print(
    "PASS: No classifier training performed."
)

print(
    "PASS: No pruning performed."
)

print(
    "PASS: Manifest saved."
)

print("\n" + "=" * 130)
print("CELL 13A COMPLETE — WAIT FOR REVIEW BEFORE CELL 13B")
print("=" * 130)

print(
    "\n2700-job manifest:"
)

print(
    MANIFEST_FILE
)

print(
    "\nManifest SHA-256:"
)

print(
    manifest_hash
)
```

```python
# ============================================================
# REPLACEMENT FOR extract_features()
# Official brain-bzh/reve-base API
# ============================================================

@torch.no_grad()
def extract_features(
    eeg,
    removed_blocks
):
    """
    Official REVE forward with temporary structured
    block replacement.

    Input:
        eeg: (B, 22, 1000)

    removed_blocks:
        1-indexed REVE blocks, e.g. [8, 10, 9]

    Output:
        (B, 1024)

    No REVE parameters are modified permanently.
    No REVE parameters are trained.
    """

    if not torch.is_tensor(eeg):
        raise TypeError(
            "eeg must be a torch.Tensor."
        )

    if eeg.ndim != 3:
        raise ValueError(
            f"Expected EEG shape (B,22,1000), "
            f"got {tuple(eeg.shape)}."
        )

    if eeg.shape[1] != 22:
        raise ValueError(
            f"Expected 22 channels, "
            f"got {eeg.shape[1]}."
        )

    if eeg.shape[2] != 1000:
        raise ValueError(
            f"Expected 1000 samples, "
            f"got {eeg.shape[2]}."
        )

    # --------------------------------------------------------
    # Validate requested blocks.
    # --------------------------------------------------------

    removed_blocks = [
        int(x)
        for x in removed_blocks
    ]

    if len(
        set(removed_blocks)
    ) != len(removed_blocks):

        raise ValueError(
            "Duplicate removed blocks."
        )

    if not all(
        1 <= x <= 22
        for x in removed_blocks
    ):

        raise ValueError(
            f"Invalid REVE block numbers: "
            f"{removed_blocks}"
        )

    removed_set = set(
        removed_blocks
    )

    transformer = reve_model.transformer

    if not hasattr(
        transformer,
        "layers"
    ):

        raise AttributeError(
            "Official REVE transformer "
            "does not expose layers."
        )

    original_layers = transformer.layers

    if len(
        original_layers
    ) != 22:

        raise ValueError(
            f"Expected 22 REVE blocks, "
            f"found {len(original_layers)}."
        )

    # --------------------------------------------------------
    # Create temporary pruned ModuleList.
    #
    # REVE block structure:
    #     [attention, feed_forward]
    #
    # A removed block becomes:
    #     [Identity(), Identity()]
    #
    # Because the official transformer applies the residual
    # connection around these operations, this corresponds
    # to removing the block contribution while retaining the
    # residual stream.
    # --------------------------------------------------------

    temporary_layers = nn.ModuleList()

    for block_number, layer in enumerate(
        original_layers,
        start=1
    ):

        if block_number in removed_set:

            temporary_layers.append(
                nn.ModuleList([
                    nn.Identity(),
                    nn.Identity()
                ])
            )

        else:

            temporary_layers.append(
                layer
            )

    transformer.layers = temporary_layers

    try:

        # ----------------------------------------------------
        # Official REVE API:
        #
        #     model(eeg, positions)
        #
        # NOT:
        #     model(eeg=..., pos=...)
        #     model(..., return_out_layers=True)
        # ----------------------------------------------------

        output = reve_model(
            eeg,
            positions
        )

    finally:

        # ----------------------------------------------------
        # ALWAYS restore the original architecture.
        # ----------------------------------------------------

        transformer.layers = original_layers

    # --------------------------------------------------------
    # Verify official output.
    # --------------------------------------------------------

    if not torch.is_tensor(
        output
    ):

        raise TypeError(
            "Official REVE forward did not return "
            "a tensor."
        )

    expected_shape = (
        eeg.shape[0],
        22,
        5,
        512
    )

    if tuple(
        output.shape
    ) != expected_shape:

        raise ValueError(
            "\nUnexpected REVE output shape.\n"
            f"Expected: {expected_shape}\n"
            f"Found:    {tuple(output.shape)}\n"
            f"Removed blocks: {removed_blocks}"
        )

    # --------------------------------------------------------
    # REVE representation:
    #
    # (B,22,5,512)
    #
    # Mean across 5 temporal patches:
    #
    # (B,22,512)
    # --------------------------------------------------------

    patch_mean = output.mean(
        dim=2
    )

    if tuple(
        patch_mean.shape
    ) != (
        eeg.shape[0],
        22,
        512
    ):

        raise ValueError(
            "Unexpected patch-pooled shape: "
            f"{tuple(patch_mean.shape)}"
        )

    # --------------------------------------------------------
    # Channel statistics:
    #
    # mean -> 512
    # std  -> 512
    #
    # concatenated -> 1024
    # --------------------------------------------------------

    channel_mean = patch_mean.mean(
        dim=1
    )

    channel_std = patch_mean.std(
        dim=1,
        unbiased=False
    )

    features = torch.cat(
        [
            channel_mean,
            channel_std
        ],
        dim=1
    )

    if tuple(
        features.shape
    ) != (
        eeg.shape[0],
        1024
    ):

        raise ValueError(
            "Unexpected classifier feature shape: "
            f"{tuple(features.shape)}"
        )

    if not torch.all(
        torch.isfinite(
            features
        )
    ):

        raise ValueError(
            "REVE features contain NaN/Inf."
        )

    return features
```

```python
# ============================================================
# CELL 13B-TEST — REAL REVE PRUNING FORWARD VERIFICATION
# ============================================================

print("=" * 120)
print("13B TEST — OFFICIAL REVE STRUCTURED PRUNING FORWARD")
print("=" * 120)

TEST_SUBJECT = "PN00"

X_np, y_np = load_subject_data(
    TEST_SUBJECT
)

print(
    "Loaded:",
    TEST_SUBJECT
)

print(
    "Stored/reformatted X shape:",
    X_np.shape
)

print(
    "Labels shape:",
    y_np.shape
)

# ------------------------------------------------------------
# Use only the first 2 epochs.
# ------------------------------------------------------------

test_eeg = torch.from_numpy(
    X_np[:2]
).to(
    DEVICE
)

print(
    "\nTest EEG shape:",
    tuple(
        test_eeg.shape
    )
)

# ------------------------------------------------------------
# Test 1 — no pruning
# ------------------------------------------------------------

print(
    "\nTest 1: 0% pruning"
)

features_0 = extract_features(
    test_eeg,
    []
)

print(
    "Output:",
    tuple(
        features_0.shape
    )
)

assert tuple(
    features_0.shape
) == (
    2,
    1024
)

# ------------------------------------------------------------
# Test 2 — remove B8
# ------------------------------------------------------------

print(
    "\nTest 2: remove B8"
)

features_b8 = extract_features(
    test_eeg,
    [8]
)

print(
    "Output:",
    tuple(
        features_b8.shape
    )
)

assert tuple(
    features_b8.shape
) == (
    2,
    1024
)

# ------------------------------------------------------------
# Test 3 — remove the actual Ours 10% set:
#           B8, B10
# ------------------------------------------------------------

print(
    "\nTest 3: Ours 10% pruning"
)

features_ours_10 = extract_features(
    test_eeg,
    [8, 10]
)

print(
    "Output:",
    tuple(
        features_ours_10.shape
    )
)

assert tuple(
    features_ours_10.shape
) == (
    2,
    1024
)

# ------------------------------------------------------------
# Verify outputs are finite.
# ------------------------------------------------------------

assert torch.all(
    torch.isfinite(
        features_0
    )
)

assert torch.all(
    torch.isfinite(
        features_b8
    )
)

assert torch.all(
    torch.isfinite(
        features_ours_10
    )
)

# ------------------------------------------------------------
# Verify original REVE architecture was restored.
# ------------------------------------------------------------

assert len(
    reve_model.transformer.layers
) == 22

for layer in reve_model.transformer.layers:

    assert isinstance(
        layer,
        (tuple, list, nn.ModuleList)
    )

    assert len(layer) == 2

print(
    "\nPASS: Original 22-block REVE architecture restored."
)

# ------------------------------------------------------------
# Check that pruning actually changes the representation.
# ------------------------------------------------------------

difference_b8 = torch.mean(
    torch.abs(
        features_0
        - features_b8
    )
).item()

difference_ours = torch.mean(
    torch.abs(
        features_0
        - features_ours_10
    )
).item()

print(
    "\nMean absolute difference:"
)

print(
    f"0% vs B8-pruned   : {difference_b8:.10f}"
)

print(
    f"0% vs Ours-10%    : {difference_ours:.10f}"
)

if difference_b8 == 0.0:

    print(
        "\nWARNING: Removing B8 produced exactly "
        "the same representation."
    )

else:

    print(
        "\nPASS: Removing B8 changes the representation."
    )

if difference_ours == 0.0:

    print(
        "WARNING: Ours 10% pruning produced "
        "exactly the same representation."
    )

else:

    print(
        "PASS: Ours 10% pruning changes the representation."
    )

# ------------------------------------------------------------
# Cleanup
# ------------------------------------------------------------

del test_eeg
del features_0
del features_b8
del features_ours_10

if DEVICE.type == "cuda":
    torch.cuda.empty_cache()

print("\n" + "=" * 120)
print("13B TEST COMPLETE — DO NOT START 2700 JOBS UNTIL THIS PASSES")
print("=" * 120)
```

```python
# =================================================================================================
# CELL 13A-F — FUNCTIONAL RANKING VERIFICATION
# FULLY CORRECTED VERSION
# =================================================================================================

import os
from pathlib import Path

import numpy as np
import pandas as pd

print("=" * 120)
print("CELL 13A-F — FUNCTIONAL RANKING VERIFICATION")
print("=" * 120)

# -------------------------------------------------------------------------------------------------
# PATHS
# -------------------------------------------------------------------------------------------------

BASE_DIR = Path("/content/drive/MyDrive/Seinna_REVE_LOSO")

FUNCTIONAL_SOURCE = (
    BASE_DIR
    / "novelty1_probes"
    / "novelty1_statistics"
    / "novelty1_all_probe_results.csv"
)

RANKING_DIR = (
    BASE_DIR
    / "novelty3_pruning"
    / "rankings"
)

RANKING_DIR.mkdir(
    parents=True,
    exist_ok=True
)

FUNCTIONAL_OUTPUT = (
    RANKING_DIR
    / "functional_block_ranking.csv"
)

# -------------------------------------------------------------------------------------------------
# 1. CHECK SOURCE
# -------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("1. CHECKING FUNCTIONAL SOURCE")
print("-" * 120)

assert FUNCTIONAL_SOURCE.exists(), (
    f"Functional source does not exist:\n{FUNCTIONAL_SOURCE}"
)

print("PASS:")
print(FUNCTIONAL_SOURCE)

# -------------------------------------------------------------------------------------------------
# 2. LOAD SOURCE
# -------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("2. SOURCE SCHEMA")
print("-" * 120)

df = pd.read_csv(FUNCTIONAL_SOURCE)

print("Shape:", df.shape)

required_columns = [
    "model",
    "block",
    "function",
    "score"
]

missing_columns = [
    c for c in required_columns
    if c not in df.columns
]

assert not missing_columns, (
    f"Missing required columns: {missing_columns}"
)

print("Required columns present.")

# -------------------------------------------------------------------------------------------------
# 3. VERIFY EXACT SOURCE DIMENSIONS
# -------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("3. VERIFYING SOURCE STRUCTURE")
print("-" * 120)

assert df.shape == (2112, 22), (
    f"Expected source shape (2112, 22), got {df.shape}"
)

EXPECTED_MODELS = [
    "pretrained",
    "random"
]

EXPECTED_FUNCTIONS = [
    "Spatial",
    "Spectral",
    "Task",
    "Temporal"
]

EXPECTED_BLOCKS = list(range(1, 23))

models = sorted(
    df["model"].dropna().unique().tolist()
)

functions = sorted(
    df["function"].dropna().unique().tolist()
)

blocks = sorted(
    df["block"].dropna().astype(int).unique().tolist()
)

print("Models:", models)
print("Functions:", functions)
print("Blocks:", blocks)

assert models == EXPECTED_MODELS, (
    f"Unexpected models: {models}"
)

assert functions == EXPECTED_FUNCTIONS, (
    f"Unexpected functions: {functions}"
)

assert blocks == EXPECTED_BLOCKS, (
    f"Unexpected blocks: {blocks}"
)

print("PASS: Models, functions and blocks verified.")

# -------------------------------------------------------------------------------------------------
# 4. VERIFY DATASET MATHEMATICALLY
# -------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("4. VERIFYING DATASET DIMENSIONS MATHEMATICALLY")
print("-" * 120)

n_models = len(EXPECTED_MODELS)
n_functions = len(EXPECTED_FUNCTIONS)
n_blocks = len(EXPECTED_BLOCKS)

expected_groups = (
    n_models
    * n_functions
    * n_blocks
)

observations_per_group = (
    len(df)
    // expected_groups
)

print("Number of models:", n_models)
print("Number of functions:", n_functions)
print("Number of blocks:", n_blocks)
print(
    "Expected model × function × block groups:",
    expected_groups
)
print(
    "Observed rows:",
    len(df)
)
print(
    "Observations per model × function × block group:",
    observations_per_group
)

assert expected_groups == 176

assert len(df) == 2112

assert len(df) % expected_groups == 0

assert observations_per_group == 12, (
    f"Expected 12 observations per model/function/block group, "
    f"got {observations_per_group}"
)

print(
    "PASS: 2 × 4 × 22 × 12 = 2112 structure verified."
)

# -------------------------------------------------------------------------------------------------
# 5. VERIFY MODEL × FUNCTION COUNTS
# -------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("5. VERIFYING MODEL/FUNCTION COUNTS")
print("-" * 120)

counts = (
    df
    .groupby(
        [
            "model",
            "function"
        ]
    )
    .size()
)

print(counts)

expected_model_function_count = (
    n_blocks
    * observations_per_group
)

assert len(counts) == (
    n_models
    * n_functions
)

assert (
    counts == expected_model_function_count
).all(), (
    f"Every model/function combination must have exactly "
    f"{expected_model_function_count} rows."
)

print(
    f"PASS: Every model/function combination has "
    f"{expected_model_function_count} rows."
)

# -------------------------------------------------------------------------------------------------
# 6. VERIFY MODEL × FUNCTION × BLOCK COUNTS
# -------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("6. VERIFYING MODEL/FUNCTION/BLOCK COUNTS")
print("-" * 120)

block_counts = (
    df
    .groupby(
        [
            "model",
            "function",
            "block"
        ]
    )
    .size()
)

expected_block_groups = (
    n_models
    * n_functions
    * n_blocks
)

assert len(block_counts) == expected_block_groups

assert (
    block_counts == observations_per_group
).all(), (
    f"Every model/function/block combination must have exactly "
    f"{observations_per_group} observations."
)

print(
    f"PASS: All {expected_block_groups} "
    f"model/function/block groups have exactly "
    f"{observations_per_group} observations."
)

# -------------------------------------------------------------------------------------------------
# 7. VERIFY COMPLETE BLOCK COVERAGE
# -------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("7. VERIFYING COMPLETE BLOCK COVERAGE")
print("-" * 120)

for model in EXPECTED_MODELS:

    for function in EXPECTED_FUNCTIONS:

        subset = df[
            (df["model"] == model)
            &
            (df["function"] == function)
        ]

        subset_blocks = sorted(
            subset["block"]
            .astype(int)
            .unique()
            .tolist()
        )

        assert subset_blocks == EXPECTED_BLOCKS, (
            f"Incomplete block coverage for "
            f"{model}/{function}: {subset_blocks}"
        )

print(
    "PASS: All 22 blocks are present for every model/function pair."
)

# -------------------------------------------------------------------------------------------------
# 8. VERIFY SCORE
# -------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("8. VERIFYING SCORE VALUES")
print("-" * 120)

df["score"] = pd.to_numeric(
    df["score"],
    errors="coerce"
)

assert df["score"].notna().all(), (
    "Score contains missing or non-numeric values."
)

score_values = df["score"].to_numpy()

assert np.isfinite(score_values).all(), (
    "Score contains NaN or Inf."
)

print(
    f"Score range: "
    f"{df['score'].min():.6f} → "
    f"{df['score'].max():.6f}"
)

print("PASS: All score values are finite.")

# -------------------------------------------------------------------------------------------------
# 9. VERIFY SUBJECT COVERAGE
# -------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("9. VERIFYING SUBJECT-LEVEL COVERAGE")
print("-" * 120)

# Prefer test_subject because it is explicitly present in this source.
assert "test_subject" in df.columns, (
    "Expected 'test_subject' column is missing."
)

subjects = sorted(
    df["test_subject"].dropna().astype(str).unique().tolist()
)

print("Number of unique test subjects:", len(subjects))
print("Subjects:", subjects)

assert len(subjects) == 12, (
    f"Expected 12 LOSO test subjects, got {len(subjects)}"
)

# Every model/function/block should contain exactly 12 subjects.
subject_counts = (
    df
    .groupby(
        [
            "model",
            "function",
            "block"
        ]
    )["test_subject"]
    .nunique()
)

assert (
    subject_counts == 12
).all(), (
    "Every model/function/block must contain exactly "
    "12 unique test subjects."
)

print(
    "PASS: Every model/function/block contains "
    "all 12 LOSO subjects exactly once."
)

# -------------------------------------------------------------------------------------------------
# 10. VERIFY NO DUPLICATE MODEL/FUNCTION/BLOCK/SUBJECT ROWS
# -------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("10. VERIFYING UNIQUE SUBJECT OBSERVATIONS")
print("-" * 120)

duplicate_mask = df.duplicated(
    subset=[
        "model",
        "function",
        "block",
        "test_subject"
    ],
    keep=False
)

duplicate_count = int(
    duplicate_mask.sum()
)

print(
    "Duplicate model/function/block/test_subject rows:",
    duplicate_count
)

assert duplicate_count == 0, (
    "Duplicate subject-level observations detected."
)

print(
    "PASS: No duplicate subject-level observations."
)

# -------------------------------------------------------------------------------------------------
# 11. COMPUTE BLOCK/FUNCTION/MODEL MEANS
# -------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("11. COMPUTING BLOCK/FUNCTION/MODEL MEANS")
print("-" * 120)

group_means = (
    df
    .groupby(
        [
            "block",
            "function",
            "model"
        ],
        as_index=False
    )["score"]
    .mean()
)

assert len(group_means) == (
    n_blocks
    * n_functions
    * n_models
)

print(
    "Computed groups:",
    len(group_means)
)

# -------------------------------------------------------------------------------------------------
# 12. PIVOT PRETRAINED AND RANDOM
# -------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("12. MATCHING PRETRAINED AND RANDOM SCORES")
print("-" * 120)

comparison = (
    group_means
    .pivot(
        index=[
            "block",
            "function"
        ],
        columns="model",
        values="score"
    )
    .reset_index()
)

assert "pretrained" in comparison.columns
assert "random" in comparison.columns

assert len(comparison) == (
    n_blocks
    * n_functions
)

assert comparison[
    "pretrained"
].notna().all()

assert comparison[
    "random"
].notna().all()

comparison[
    "function_effect"
] = (
    comparison["pretrained"]
    -
    comparison["random"]
)

assert np.isfinite(
    comparison["function_effect"].to_numpy()
).all()

print(
    "PASS: Every block/function has both "
    "pretrained and random means."
)

# -------------------------------------------------------------------------------------------------
# 13. FUNCTION-SPECIFIC EFFECT TABLE
# -------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("13. FUNCTION-SPECIFIC PRETRAINED − RANDOM EFFECTS")
print("-" * 120)

function_effect_table = (
    comparison
    .pivot(
        index="block",
        columns="function",
        values="function_effect"
    )
    .reset_index()
)

for function in EXPECTED_FUNCTIONS:

    assert function in function_effect_table.columns

assert len(function_effect_table) == 22

print(
    function_effect_table.to_string(
        index=False
    )
)

# -------------------------------------------------------------------------------------------------
# 14. EQUAL-WEIGHT FUNCTIONAL SCORE
#
# F_b = 1/4 ×
#       [Spatial_effect +
#        Spectral_effect +
#        Task_effect +
#        Temporal_effect]
# -------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("14. COMPUTING EQUAL-WEIGHT FUNCTIONAL SCORE")
print("-" * 120)

functional_scores = (
    comparison
    .groupby(
        "block"
    )["function_effect"]
    .mean()
    .reset_index()
    .rename(
        columns={
            "function_effect":
                "functional_score"
        }
    )
)

assert len(functional_scores) == 22

assert sorted(
    functional_scores[
        "block"
    ].astype(int)
) == EXPECTED_BLOCKS

assert np.isfinite(
    functional_scores[
        "functional_score"
    ].to_numpy()
).all()

print(
    functional_scores.to_string(
        index=False
    )
)

print(
    "\nPASS: Functional score calculated for all 22 blocks."
)

# -------------------------------------------------------------------------------------------------
# 15. MIN-MAX NORMALIZATION
# -------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("15. MIN-MAX NORMALIZATION")
print("-" * 120)

fmin = functional_scores[
    "functional_score"
].min()

fmax = functional_scores[
    "functional_score"
].max()

assert np.isfinite(fmin)
assert np.isfinite(fmax)

assert fmax > fmin, (
    "Functional scores are identical; "
    "min-max normalization cannot be performed."
)

functional_scores[
    "functional_score_normalized"
] = (
    functional_scores[
        "functional_score"
    ]
    -
    fmin
) / (
    fmax
    -
    fmin
)

normalized_values = functional_scores[
    "functional_score_normalized"
].to_numpy()

assert np.isfinite(
    normalized_values
).all()

assert normalized_values.min() >= 0.0
assert normalized_values.max() <= 1.0

print(
    f"Raw range: {fmin:.6f} → {fmax:.6f}"
)

print(
    f"Normalized range: "
    f"{normalized_values.min():.6f} → "
    f"{normalized_values.max():.6f}"
)

print(
    "PASS: Functional scores normalized to [0,1]."
)

# -------------------------------------------------------------------------------------------------
# 16. CREATE DESCENDING RANKING
# -------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("16. CREATING FUNCTIONAL RANKING")
print("-" * 120)

functional_scores = (
    functional_scores
    .sort_values(
        [
            "functional_score",
            "block"
        ],
        ascending=[
            False,
            True
        ]
    )
    .reset_index(
        drop=True
    )
)

functional_scores[
    "rank"
] = np.arange(
    1,
    23
)

functional_scores[
    "block_label"
] = (
    "B"
    +
    functional_scores[
        "block"
    ].astype(int).astype(str)
)

# -------------------------------------------------------------------------------------------------
# 17. RANKING INTEGRITY
# -------------------------------------------------------------------------------------------------

assert functional_scores.shape[0] == 22

assert functional_scores[
    "rank"
].tolist() == list(range(1, 23))

computed_ranking = (
    functional_scores[
        "block"
    ]
    .astype(int)
    .tolist()
)

assert len(
    set(computed_ranking)
) == 22

assert sorted(
    computed_ranking
) == EXPECTED_BLOCKS

score_array = functional_scores[
    "functional_score"
].to_numpy()

assert np.all(
    score_array[:-1] >= score_array[1:]
)

print(
    "PASS: Ranking contains exactly 22 unique blocks "
    "in descending functional-score order."
)

# -------------------------------------------------------------------------------------------------
# 18. PRINT FINAL RANKING
# -------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("17. FINAL COMPUTED FUNCTIONAL RANKING")
print("-" * 120)

for _, row in functional_scores.iterrows():

    print(
        f"{int(row['rank']):>2}. "
        f"B{int(row['block']):02d}   "
        f"raw={row['functional_score']:.6f}   "
        f"normalized={row['functional_score_normalized']:.6f}"
    )

print("\nFunctional removal order:")

print(
    " → ".join(
        f"B{b}"
        for b in computed_ranking
    )
)

# -------------------------------------------------------------------------------------------------
# 19. SAVE VERIFIED RANKING
# -------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("18. SAVING FUNCTIONAL RANKING")
print("-" * 120)

output_columns = [
    "rank",
    "block",
    "block_label",
    "functional_score",
    "functional_score_normalized"
]

functional_scores[
    output_columns
].to_csv(
    FUNCTIONAL_OUTPUT,
    index=False
)

assert FUNCTIONAL_OUTPUT.exists()

print("Saved:")
print(FUNCTIONAL_OUTPUT)

# -------------------------------------------------------------------------------------------------
# 20. RELOAD AND VERIFY SAVED FILE
# -------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("19. RELOADING AND VERIFYING SAVED FILE")
print("-" * 120)

saved = pd.read_csv(
    FUNCTIONAL_OUTPUT
)

assert saved.shape == (
    22,
    5
)

assert saved[
    "rank"
].tolist() == list(range(1, 23))

saved_ranking = (
    saved[
        "block"
    ]
    .astype(int)
    .tolist()
)

assert saved_ranking == computed_ranking

assert len(
    set(saved_ranking)
) == 22

assert sorted(
    saved_ranking
) == EXPECTED_BLOCKS

assert np.isfinite(
    saved[
        "functional_score"
    ].to_numpy()
).all()

assert np.isfinite(
    saved[
        "functional_score_normalized"
    ].to_numpy()
).all()

print(
    "PASS: Saved ranking reloaded successfully."
)

print(
    "PASS: Saved ranking exactly matches computed ranking."
)

# -------------------------------------------------------------------------------------------------
# 21. FINAL STATUS
# -------------------------------------------------------------------------------------------------

print("\n" + "=" * 120)
print("CELL 13A-F COMPLETE — FUNCTIONAL RANKING VERIFIED")
print("=" * 120)

print(
    "\nFinal Functional order:"
)

print(
    " → ".join(
        f"B{b}"
        for b in saved_ranking
    )
)

print(
    "\nOutput file:"
)

print(
    FUNCTIONAL_OUTPUT
)

print(
    "\nNo REVE model was loaded."
)

print(
    "No EEG was loaded."
)

print(
    "No classifier was trained."
)

print(
    "No pruning evaluation was performed."
)

print(
    "\nREADY FOR REVIEW BEFORE CELL 13B."
)
```

```python
# ======================================================================================================================
# CELL 13B — SECTION 2
# VERIFIED SOURCE EEG FILE DISCOVERY + STRUCTURE CHECK
# ======================================================================================================================

import os
import glob
import numpy as np

print("\n" + "-" * 112)
print("2. VERIFYING SOURCE EEG FILES")
print("-" * 112)

SOURCE_EEG_DIR = "/content/drive/MyDrive/Seinna_REVE_LOSO/subject_epochs_22ch"

EXPECTED_SUBJECTS = [
    "PN00", "PN01", "PN06", "PN07", "PN09", "PN10",
    "PN11", "PN12", "PN13", "PN14", "PN16", "PN17"
]

# ----------------------------------------------------------------------------------------------------------------------
# 2.1 VERIFY SOURCE DIRECTORY
# ----------------------------------------------------------------------------------------------------------------------

assert os.path.isdir(SOURCE_EEG_DIR), (
    f"Source EEG directory does not exist:\n{SOURCE_EEG_DIR}"
)

print(f"Source directory exists:\n{SOURCE_EEG_DIR}")

# ----------------------------------------------------------------------------------------------------------------------
# 2.2 FIND ALL NPZ FILES
# ----------------------------------------------------------------------------------------------------------------------

all_npz_files = sorted(
    glob.glob(os.path.join(SOURCE_EEG_DIR, "*.npz"))
)

print(f"Total .npz files found: {len(all_npz_files)}")

assert len(all_npz_files) == len(EXPECTED_SUBJECTS), (
    f"Expected exactly {len(EXPECTED_SUBJECTS)} NPZ files "
    f"for the verified LOSO subjects, but found {len(all_npz_files)}.\n"
    f"Files found:\n{all_npz_files}"
)

# ----------------------------------------------------------------------------------------------------------------------
# 2.3 BUILD EXACT <SUBJECT>_epochs.npz INDEX
# ----------------------------------------------------------------------------------------------------------------------

SOURCE_EEG_FILES = {}

for path in all_npz_files:

    basename = os.path.basename(path)

    stem = os.path.splitext(basename)[0]

    # Required naming convention:
    # PN00_epochs.npz
    # PN01_epochs.npz
    # etc.

    assert stem.endswith("_epochs"), (
        f"Unexpected NPZ filename:\n{basename}\n"
        f"Expected naming convention: <SUBJECT>_epochs.npz"
    )

    subject = stem[:-len("_epochs")].upper()

    assert subject in EXPECTED_SUBJECTS, (
        f"Unexpected subject file:\n{basename}\n"
        f"Parsed subject: {subject}\n"
        f"Expected subjects: {EXPECTED_SUBJECTS}"
    )

    assert subject not in SOURCE_EEG_FILES, (
        f"Duplicate source file for subject {subject}."
    )

    SOURCE_EEG_FILES[subject] = path

# ----------------------------------------------------------------------------------------------------------------------
# 2.4 VERIFY EXACT SUBJECT COVERAGE
# ----------------------------------------------------------------------------------------------------------------------

resolved_subjects = sorted(SOURCE_EEG_FILES.keys())
expected_subjects_sorted = sorted(EXPECTED_SUBJECTS)

assert resolved_subjects == expected_subjects_sorted, (
    "Resolved subject set does not exactly match expected LOSO subjects.\n"
    f"Expected: {expected_subjects_sorted}\n"
    f"Resolved: {resolved_subjects}"
)

print("\nPASS: Exact 12-subject source coverage verified.")

# ----------------------------------------------------------------------------------------------------------------------
# 2.5 VERIFY EACH SOURCE FILE EXISTS AND IS NON-EMPTY
# ----------------------------------------------------------------------------------------------------------------------

print("\nResolved source files:")

for subject in EXPECTED_SUBJECTS:

    source_path = SOURCE_EEG_FILES[subject]

    assert os.path.isfile(source_path), (
        f"{subject}: source path is not a regular file:\n{source_path}"
    )

    file_size = os.path.getsize(source_path)

    assert file_size > 0, (
        f"{subject}: source file is empty:\n{source_path}"
    )

    print(
        f"PASS: {subject} -> "
        f"{os.path.basename(source_path)} "
        f"({file_size / (1024 ** 2):.2f} MB)"
    )

# ----------------------------------------------------------------------------------------------------------------------
# 2.6 VERIFY NPZ ARRAYS
# ----------------------------------------------------------------------------------------------------------------------

print("\nVerifying NPZ contents and array structure...")

SOURCE_EEG_METADATA = {}

for subject in EXPECTED_SUBJECTS:

    source_path = SOURCE_EEG_FILES[subject]

    with np.load(source_path, allow_pickle=False) as data:

        keys = sorted(data.files)

        assert "X" in data.files, (
            f"{subject}: required array 'X' not found.\n"
            f"Available keys: {keys}"
        )

        assert "y" in data.files, (
            f"{subject}: required array 'y' not found.\n"
            f"Available keys: {keys}"
        )

        X = data["X"]
        y = data["y"]

        # Expected verified saved format:
        #
        # X = (N, 1000, 22)
        # y = (N,)
        #
        # REVE input will later be converted in memory to:
        #
        # (N, 22, 1000)

        assert X.ndim == 3, (
            f"{subject}: X must be 3-dimensional.\n"
            f"Got: {X.shape}"
        )

        assert y.ndim == 1, (
            f"{subject}: y must be 1-dimensional.\n"
            f"Got: {y.shape}"
        )

        assert X.shape[1] == 1000, (
            f"{subject}: unexpected number of samples per epoch.\n"
            f"Expected: 1000\n"
            f"Got: {X.shape[1]}"
        )

        assert X.shape[2] == 22, (
            f"{subject}: unexpected number of EEG channels.\n"
            f"Expected: 22\n"
            f"Got: {X.shape[2]}"
        )

        assert X.shape[0] == y.shape[0], (
            f"{subject}: X/y epoch count mismatch.\n"
            f"X epochs: {X.shape[0]}\n"
            f"y epochs: {y.shape[0]}"
        )

        assert X.shape[0] > 0, (
            f"{subject}: zero epochs found."
        )

        assert np.issubdtype(X.dtype, np.number), (
            f"{subject}: X is not numeric.\n"
            f"dtype: {X.dtype}"
        )

        assert np.issubdtype(y.dtype, np.number), (
            f"{subject}: y is not numeric.\n"
            f"dtype: {y.dtype}"
        )

        assert np.isfinite(X).all(), (
            f"{subject}: X contains NaN or Inf values."
        )

        assert np.isfinite(y).all(), (
            f"{subject}: y contains NaN or Inf values."
        )

        unique_labels = np.unique(y)

        assert set(unique_labels.tolist()).issubset({0, 1}), (
            f"{subject}: labels are not binary 0/1.\n"
            f"Found labels: {unique_labels.tolist()}"
        )

        SOURCE_EEG_METADATA[subject] = {
            "path": source_path,
            "n_epochs": int(X.shape[0]),
            "n_samples": int(X.shape[1]),
            "n_channels": int(X.shape[2]),
            "x_dtype": str(X.dtype),
            "y_dtype": str(y.dtype),
            "labels": unique_labels.astype(int).tolist()
        }

        print(
            f"PASS: {subject} | "
            f"X={X.shape} | "
            f"y={y.shape} | "
            f"labels={unique_labels.astype(int).tolist()}"
        )

# ----------------------------------------------------------------------------------------------------------------------
# 2.7 FINAL SOURCE CHECK
# ----------------------------------------------------------------------------------------------------------------------

assert len(SOURCE_EEG_FILES) == 12
assert len(SOURCE_EEG_METADATA) == 12

print("\n" + "-" * 112)
print("SOURCE EEG VERIFICATION COMPLETE")
print("-" * 112)

print("PASS: 12/12 expected LOSO subjects have exactly one source EEG file.")
print("PASS: Filename convention <SUBJECT>_epochs.npz verified.")
print("PASS: Every file contains X and y.")
print("PASS: X format verified as (N, 1000, 22).")
print("PASS: y format verified as (N,).")
print("PASS: X/y epoch counts match for every subject.")
print("PASS: All EEG values are finite.")
print("PASS: All labels are binary 0/1.")
print("PASS: SOURCE_EEG_FILES is ready for Cell 13B.")

print("\nVerified source mapping:")

for subject in EXPECTED_SUBJECTS:
    print(f"{subject}: {SOURCE_EEG_FILES[subject]}")
```

```python
# ======================================================================================================================
# CELL 13B — COMPLETE FILE/DIRECTORY DIAGNOSTIC
# Finds existing files and reports ALL missing dependencies.
# NO assertions / NO REVE loading / NO EEG forward / NO classifier training.
# ======================================================================================================================

import os
import glob

print("\n" + "=" * 115)
print("CELL 13B — COMPLETE FILE/DIRECTORY DIAGNOSTIC")
print("=" * 115)

BASE_DIR = "/content/drive/MyDrive/Seinna_REVE_LOSO"
NOVELTY3_DIR = os.path.join(BASE_DIR, "novelty3_pruning")

print("\nBASE DIRECTORY:")
print(BASE_DIR)
print("Exists:", os.path.isdir(BASE_DIR))

print("\nNOVELTY 3 DIRECTORY:")
print(NOVELTY3_DIR)
print("Exists:", os.path.isdir(NOVELTY3_DIR))

# ----------------------------------------------------------------------------------------------------------------------
# 1. SHOW NOVELTY3 DIRECTORY TREE
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 115)
print("1. EXISTING NOVELTY3 FILES")
print("-" * 115)

if os.path.isdir(NOVELTY3_DIR):

    all_files = []

    for root, dirs, files in os.walk(NOVELTY3_DIR):
        for fname in files:
            full_path = os.path.join(root, fname)
            rel_path = os.path.relpath(full_path, NOVELTY3_DIR)
            size_mb = os.path.getsize(full_path) / (1024 ** 2)
            all_files.append((rel_path, size_mb))

    all_files.sort()

    print(f"Total files found: {len(all_files)}")

    for rel_path, size_mb in all_files:
        print(f"  {rel_path}  [{size_mb:.2f} MB]")

else:
    print("NOVELTY3 DIRECTORY DOES NOT EXIST.")

# ----------------------------------------------------------------------------------------------------------------------
# 2. SEARCH FOR PRUNING PLAN UNDER ENTIRE DRIVE
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 115)
print("2. SEARCHING FOR PRUNING PLAN FILES")
print("-" * 115)

search_patterns = [
    "pruning_plans.csv",
    "*pruning*plan*.csv",
    "*pruning*plans*.csv",
]

found_pruning = []

for pattern in search_patterns:

    matches = glob.glob(
        os.path.join(
            "/content/drive/MyDrive",
            "**",
            pattern
        ),
        recursive=True
    )

    for path in matches:
        if path not in found_pruning:
            found_pruning.append(path)

if found_pruning:

    for path in sorted(found_pruning):
        print("FOUND:", path)

else:
    print("NOT FOUND: Any pruning plan CSV")

# ----------------------------------------------------------------------------------------------------------------------
# 3. SEARCH FOR OURS SCORE FILE
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 115)
print("3. SEARCHING FOR OURS SCORE FILES")
print("-" * 115)

ours_patterns = [
    "ours_block_scores.csv",
    "*ours*score*.csv",
    "*ours*.csv",
]

found_ours = []

for pattern in ours_patterns:

    matches = glob.glob(
        os.path.join(
            "/content/drive/MyDrive",
            "**",
            pattern
        ),
        recursive=True
    )

    for path in matches:
        if path not in found_ours:
            found_ours.append(path)

if found_ours:

    for path in sorted(found_ours):
        print("FOUND:", path)

else:
    print("NOT FOUND: Any Ours CSV")

# ----------------------------------------------------------------------------------------------------------------------
# 4. SEARCH FOR ALL FIVE RANKINGS
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 115)
print("4. SEARCHING FOR ALL RANKING FILES")
print("-" * 115)

ranking_patterns = [
    "*ddf*ranking*.csv",
    "*ssn*ranking*.csv",
    "*functional*ranking*.csv",
    "*ours*ranking*.csv",
    "*random*ranking*.csv",
    "*block*ranking*.csv",
]

found_rankings = []

for pattern in ranking_patterns:

    matches = glob.glob(
        os.path.join(
            "/content/drive/MyDrive",
            "**",
            pattern
        ),
        recursive=True
    )

    for path in matches:
        if path not in found_rankings:
            found_rankings.append(path)

if found_rankings:

    for path in sorted(found_rankings):
        print("FOUND:", path)

else:
    print("NOT FOUND: Ranking CSVs")

# ----------------------------------------------------------------------------------------------------------------------
# 5. SEARCH FOR GATE FILES
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 115)
print("5. SEARCHING FOR GATE FILES")
print("-" * 115)

gate_patterns = [
    "*gate*matrix*.npy",
    "*gate*matrix*.csv",
    "*gate*aggregate*.csv",
    "*final*gates*.npy",
]

found_gates = []

for pattern in gate_patterns:

    matches = glob.glob(
        os.path.join(
            "/content/drive/MyDrive",
            "**",
            pattern
        ),
        recursive=True
    )

    for path in matches:
        if path not in found_gates:
            found_gates.append(path)

if found_gates:

    for path in sorted(found_gates):
        print("FOUND:", path)

else:
    print("NOT FOUND: Gate files")

# ----------------------------------------------------------------------------------------------------------------------
# 6. SEARCH FOR REVE ARCHIVES
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 115)
print("6. SEARCHING FOR REVE ARCHIVE FILES")
print("-" * 115)

reve_patterns = [
    "reve_base_state_dict.pt",
    "*reve*state*dict*.pt",
    "*reve*.pt",
]

found_reve = []

for pattern in reve_patterns:

    matches = glob.glob(
        os.path.join(
            "/content/drive/MyDrive",
            "**",
            pattern
        ),
        recursive=True
    )

    for path in matches:
        if path not in found_reve:
            found_reve.append(path)

if found_reve:

    for path in sorted(found_reve):
        print("FOUND:", path)

else:
    print("NOT FOUND: REVE checkpoint/state-dict files")

# ----------------------------------------------------------------------------------------------------------------------
# 7. SEARCH FOR POSITION ARCHIVE
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 115)
print("7. SEARCHING FOR POSITION MODEL ARCHIVE")
print("-" * 115)

position_patterns = [
    "reve_positions_state_dict.pt",
    "*position*state*dict*.pt",
    "*position*.pt",
]

found_positions = []

for pattern in position_patterns:

    matches = glob.glob(
        os.path.join(
            "/content/drive/MyDrive",
            "**",
            pattern
        ),
        recursive=True
    )

    for path in matches:
        if path not in found_positions:
            found_positions.append(path)

if found_positions:

    for path in sorted(found_positions):
        print("FOUND:", path)

else:
    print("NOT FOUND: Position model archive")

# ----------------------------------------------------------------------------------------------------------------------
# 8. SEARCH FOR EEG SOURCE DIRECTORY
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 115)
print("8. VERIFYING SOURCE EEG DIRECTORY")
print("-" * 115)

source_dir = os.path.join(
    BASE_DIR,
    "subject_epochs_22ch"
)

print("Source directory:", source_dir)
print("Exists:", os.path.isdir(source_dir))

if os.path.isdir(source_dir):

    source_files = sorted(
        glob.glob(
            os.path.join(source_dir, "*.npz")
        )
    )

    print("NPZ files:", len(source_files))

    for path in source_files:
        print(" ", os.path.basename(path))

# ----------------------------------------------------------------------------------------------------------------------
# 9. SEARCH FOR OLD/CURRENT CHECKPOINTS
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 115)
print("9. SEARCHING FOR EVALUATION CHECKPOINTS / RESULTS")
print("-" * 115)

evaluation_patterns = [
    "*job*.csv",
    "*metrics*.csv",
    "*evaluation*.csv",
    "*checkpoint*.pt",
    "*job*checkpoint*",
]

found_eval = []

for pattern in evaluation_patterns:

    matches = glob.glob(
        os.path.join(
            NOVELTY3_DIR,
            "**",
            pattern
        ),
        recursive=True
    )

    for path in matches:
        if path not in found_eval:
            found_eval.append(path)

if found_eval:

    for path in sorted(found_eval):
        print("FOUND:", path)

else:
    print("No evaluation-result/checkpoint files currently found.")

# ----------------------------------------------------------------------------------------------------------------------
# 10. CHECK EXACT CURRENT EXPECTED DEPENDENCIES
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "=" * 115)
print("10. EXACT DEPENDENCY STATUS")
print("=" * 115)

expected_dependencies = {
    "Gate matrix":
        os.path.join(
            NOVELTY3_DIR,
            "gate_training",
            "gate_matrix_60x22.npy"
        ),

    "Gate statistics":
        os.path.join(
            NOVELTY3_DIR,
            "gate_training",
            "gate_aggregate_statistics.csv"
        ),

    "Ours scores":
        os.path.join(
            NOVELTY3_DIR,
            "gate_training",
            "ours_block_scores.csv"
        ),

    "Pruning plans":
        os.path.join(
            NOVELTY3_DIR,
            "gate_training",
            "pruning_plans.csv"
        ),

    "DDF ranking":
        os.path.join(
            NOVELTY3_DIR,
            "rankings",
            "ddf_block_ranking.csv"
        ),

    "SSN ranking":
        os.path.join(
            NOVELTY3_DIR,
            "rankings",
            "ssn_block_ranking.csv"
        ),

    "Functional ranking":
        os.path.join(
            NOVELTY3_DIR,
            "rankings",
            "functional_block_ranking.csv"
        ),

    "REVE archive":
        os.path.join(
            NOVELTY3_DIR,
            "EXPERIMENT_ARCHIVE",
            "REVE_MODEL",
            "reve_base_state_dict.pt"
        ),

    "Position archive":
        os.path.join(
            NOVELTY3_DIR,
            "EXPERIMENT_ARCHIVE",
            "REVE_POSITION_MODEL",
            "reve_positions_state_dict.pt"
        ),
}

missing = []
present = []

for name, path in expected_dependencies.items():

    if os.path.isfile(path):
        present.append(name)
        print(f"PASS   {name}")
        print(f"       {path}")

    else:
        missing.append((name, path))
        print(f"MISSING {name}")
        print(f"       {path}")

# ----------------------------------------------------------------------------------------------------------------------
# 11. FINAL SUMMARY
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "=" * 115)
print("FINAL DIAGNOSTIC SUMMARY")
print("=" * 115)

print(f"Present dependencies : {len(present)}")
print(f"Missing dependencies : {len(missing)}")

if missing:

    print("\nMISSING FILES:")
    for name, path in missing:
        print(f"  - {name}")
        print(f"    {path}")

else:

    print("\nALL EXPECTED DEPENDENCIES EXIST.")

print("\nIMPORTANT:")
print("This cell did NOT run REVE.")
print("This cell did NOT load EEG.")
print("This cell did NOT train a classifier.")
print("This cell did NOT perform pruning.")
print("This cell did NOT start the 2700-job experiment.")

print("\n" + "=" * 115)
print("DIAGNOSTIC COMPLETE")
print("=" * 115)
```

```python
# ======================================================================================================================
# CELL 13B — FINAL CONTENT VERIFICATION
# Uses ACTUAL discovered paths.
# NO evaluation / NO classifier training.
# ======================================================================================================================

import os
import numpy as np
import pandas as pd

print("\n" + "=" * 115)
print("CELL 13B — FINAL CONTENT VERIFICATION")
print("=" * 115)

BASE_DIR = "/content/drive/MyDrive/Seinna_REVE_LOSO"
NOVELTY3_DIR = os.path.join(BASE_DIR, "novelty3_pruning")

# ACTUAL DISCOVERED PATHS
GATE_MATRIX = os.path.join(
    NOVELTY3_DIR,
    "gate_training",
    "gate_matrix_60x22.npy"
)

GATE_STATS = os.path.join(
    NOVELTY3_DIR,
    "gate_training",
    "gate_aggregate_statistics.csv"
)

OURS_SCORES = os.path.join(
    NOVELTY3_DIR,
    "gate_training",
    "ours_block_scores.csv"
)

PRUNING_PLANS = os.path.join(
    NOVELTY3_DIR,
    "pruning_plans.csv"
)

DDF_RANKING = os.path.join(
    NOVELTY3_DIR,
    "rankings",
    "ddf_block_ranking.csv"
)

SSN_RANKING = os.path.join(
    NOVELTY3_DIR,
    "rankings",
    "ssn_block_ranking.csv"
)

FUNCTIONAL_RANKING = os.path.join(
    NOVELTY3_DIR,
    "rankings",
    "functional_block_ranking.csv"
)

OURS_RANKING = os.path.join(
    NOVELTY3_DIR,
    "rankings",
    "ours_block_ranking.csv"
)

REVE_STATE = "/content/drive/MyDrive/Seinna_REVE_Novelty3_NEW/EXPERIMENT_ARCHIVE/REVE_MODEL/reve_base_state_dict.pt"

POSITION_STATE = "/content/drive/MyDrive/Seinna_REVE_Novelty3_NEW/EXPERIMENT_ARCHIVE/REVE_POSITION_MODEL/reve_positions_state_dict.pt"

SOURCE_DIR = os.path.join(
    BASE_DIR,
    "subject_epochs_22ch"
)

# ----------------------------------------------------------------------------------------------------------------------
# 1. FILE EXISTENCE
# ----------------------------------------------------------------------------------------------------------------------

print("\n1. FILE EXISTENCE")

paths = {
    "Gate matrix": GATE_MATRIX,
    "Gate statistics": GATE_STATS,
    "Ours scores": OURS_SCORES,
    "Pruning plans": PRUNING_PLANS,
    "DDF ranking": DDF_RANKING,
    "SSN ranking": SSN_RANKING,
    "Functional ranking": FUNCTIONAL_RANKING,
    "Ours ranking": OURS_RANKING,
    "REVE state": REVE_STATE,
    "Position state": POSITION_STATE,
}

missing = []

for name, path in paths.items():

    if os.path.isfile(path) and os.path.getsize(path) > 0:
        print(f"PASS: {name}")
    else:
        print(f"MISSING: {name}")
        print(path)
        missing.append(name)

# ----------------------------------------------------------------------------------------------------------------------
# 2. GATE MATRIX
# ----------------------------------------------------------------------------------------------------------------------

print("\n2. GATE MATRIX")

gates = np.load(GATE_MATRIX)

print("Shape:", gates.shape)
print("dtype:", gates.dtype)

if gates.shape == (60, 22):
    print("PASS: exact (60,22)")
else:
    print("FAIL: unexpected gate shape")

if np.isfinite(gates).all():
    print("PASS: all gate values finite")
else:
    print("FAIL: NaN/Inf in gates")

print(
    f"Mean={gates.mean():.6f} | "
    f"Std={gates.std():.6f} | "
    f"Min={gates.min():.6f} | "
    f"Max={gates.max():.6f}"
)

# ----------------------------------------------------------------------------------------------------------------------
# 3. OURS SCORE FILE
# ----------------------------------------------------------------------------------------------------------------------

print("\n3. OURS SCORE FILE")

ours = pd.read_csv(OURS_SCORES)

print("Shape:", ours.shape)
print("Columns:", ours.columns.tolist())

print("Blocks:", sorted(ours["block"].astype(int).tolist()))

if len(ours) == 22:
    print("PASS: 22 blocks")
else:
    print("FAIL: expected 22 blocks")

if set(ours["block"].astype(int)) == set(range(1, 23)):
    print("PASS: complete B1-B22 coverage")
else:
    print("FAIL: incomplete block coverage")

if "ours_score" in ours.columns:
    if np.isfinite(ours["ours_score"].to_numpy(dtype=float)).all():
        print("PASS: Ours scores finite")
    else:
        print("FAIL: Ours scores contain NaN/Inf")
else:
    print("FAIL: ours_score column missing")

ours_order = (
    ours.sort_values(
        "ours_score",
        ascending=False,
        kind="mergesort"
    )["block"]
    .astype(int)
    .tolist()
)

print("Ours removal order:")
print(ours_order)

# ----------------------------------------------------------------------------------------------------------------------
# 4. ALL RANKINGS
# ----------------------------------------------------------------------------------------------------------------------

print("\n4. RANKING CONTENT")

def inspect_ranking(name, path):

    df = pd.read_csv(path)

    print(f"\n{name}")
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())

    # Detect block column
    block_col = None

    for c in df.columns:
        if c.lower() in [
            "block",
            "block_id",
            "removed_block",
            "layer",
            "removed_layer"
        ]:
            block_col = c
            break

    if block_col is None:
        print("FAIL: no block column found")
        return None

    blocks = []

    for value in df[block_col]:

        s = str(value).strip().upper()

        if s.startswith("H"):
            blocks.append(int(s[1:]) + 1)

        elif s.startswith("B"):
            blocks.append(int(s[1:]))

        else:
            blocks.append(int(float(s)))

    print("Block order:", blocks)

    if len(df) == 22:
        print("PASS: 22 rows")
    else:
        print("FAIL: expected 22 rows")

    if set(blocks) == set(range(1, 23)):
        print("PASS: complete B1-B22 permutation")
    else:
        print("FAIL: not a complete B1-B22 permutation")

    return blocks


ddf_order = inspect_ranking(
    "DDF",
    DDF_RANKING
)

ssn_order = inspect_ranking(
    "SSN",
    SSN_RANKING
)

functional_order = inspect_ranking(
    "Functional",
    FUNCTIONAL_RANKING
)

ours_rank_order = inspect_ranking(
    "Ours ranking",
    OURS_RANKING
)

# ----------------------------------------------------------------------------------------------------------------------
# 5. PRUNING PLAN
# ----------------------------------------------------------------------------------------------------------------------

print("\n5. PRUNING PLAN")

plans = pd.read_csv(PRUNING_PLANS)

print("Shape:", plans.shape)
print("Columns:", plans.columns.tolist())

print(plans.to_string(index=False))

expected_levels = [10,20,30,40,50,60,70,80,90]
expected_remove = [2,4,7,9,11,13,15,18,20]
expected_keep = [20,18,15,13,11,9,7,4,2]

if "pruning_percent" in plans.columns:
    print(
        "Levels:",
        plans["pruning_percent"].tolist()
    )

if "remove_count" in plans.columns:
    print(
        "Remove:",
        plans["remove_count"].tolist()
    )

if "keep_count" in plans.columns:
    print(
        "Keep:",
        plans["keep_count"].tolist()
    )

if (
    "pruning_percent" in plans.columns
    and plans["pruning_percent"].tolist() == expected_levels
):
    print("PASS: exact 10-90% pruning levels")
else:
    print("CHECK: pruning levels differ")

if (
    "remove_count" in plans.columns
    and plans["remove_count"].tolist() == expected_remove
):
    print("PASS: exact removal counts")
else:
    print("CHECK: removal counts differ")

if (
    "keep_count" in plans.columns
    and plans["keep_count"].tolist() == expected_keep
):
    print("PASS: exact keep counts")
else:
    print("CHECK: keep counts differ")

# ----------------------------------------------------------------------------------------------------------------------
# 6. SOURCE EEG
# ----------------------------------------------------------------------------------------------------------------------

print("\n6. SOURCE EEG")

expected_subjects = [
    "PN00","PN01","PN06","PN07","PN09","PN10",
    "PN11","PN12","PN13","PN14","PN16","PN17"
]

source_files = sorted(
    [
        f for f in os.listdir(SOURCE_DIR)
        if f.endswith(".npz")
    ]
)

print("NPZ count:", len(source_files))
print("Files:", source_files)

resolved = {}

for subject in expected_subjects:

    expected_name = f"{subject}_epochs.npz"
    path = os.path.join(SOURCE_DIR, expected_name)

    if os.path.isfile(path):
        resolved[subject] = path
        print(f"PASS: {subject} -> {expected_name}")
    else:
        print(f"MISSING: {subject} -> {expected_name}")

print(
    f"Resolved {len(resolved)}/{len(expected_subjects)} subjects"
)

# ----------------------------------------------------------------------------------------------------------------------
# 7. PRUNING PLAN ↔ OURS RANKING
# ----------------------------------------------------------------------------------------------------------------------

print("\n7. OURS RANKING ↔ PRUNING PLAN")

if (
    "removed_blocks" in plans.columns
    and len(ours_order) == 22
):

    consistent = True

    for _, row in plans.iterrows():

        remove_count = int(row["remove_count"])

        planned = [
            int(
                str(x)
                .strip()
                .replace("B", "")
            )
            for x in str(row["removed_blocks"]).split(",")
            if str(x).strip()
        ]

        expected = ours_order[:remove_count]

        if planned != expected:
            print(
                f"CHECK {int(row['pruning_percent'])}%:"
                f" plan={planned}, expected={expected}"
            )
            consistent = False

    if consistent:
        print("PASS: pruning plans follow Ours ranking")
    else:
        print("CHECK: pruning plans do not exactly follow Ours ranking")

# ----------------------------------------------------------------------------------------------------------------------
# 8. OLD 2700 MANIFEST — INSPECT ONLY
# ----------------------------------------------------------------------------------------------------------------------

print("\n8. EXISTING 2700-JOB MANIFEST")

manifest_csv = os.path.join(
    NOVELTY3_DIR,
    "pruning_2700_job_manifest.csv"
)

if os.path.isfile(manifest_csv):

    manifest = pd.read_csv(manifest_csv)

    print("Shape:", manifest.shape)
    print("Columns:", manifest.columns.tolist())

    print(
        "Unique methods:",
        manifest["method"].unique().tolist()
        if "method" in manifest.columns
        else "method column not found"
    )

    if "method" in manifest.columns:
        print(
            "Method counts:\n",
            manifest["method"].value_counts()
        )

    if "pruning_percent" in manifest.columns:
        print(
            "Levels:",
            sorted(
                manifest["pruning_percent"].unique().tolist()
            )
        )

else:
    print("No existing 2700 manifest found.")

# ----------------------------------------------------------------------------------------------------------------------
# FINAL SUMMARY
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "=" * 115)
print("FINAL VERIFICATION SUMMARY")
print("=" * 115)

print(f"Missing required files: {len(missing)}")

if missing:
    print("\nMISSING:")
    for x in missing:
        print(" -", x)
else:
    print("PASS: All required files exist.")

print("\nNO EVALUATION WAS RUN.")
print("NO CLASSIFIER TRAINING WAS RUN.")
print("NO REVE FORWARD WAS RUN.")
print("NO FILES WERE MODIFIED.")

print("\nNEXT STEP:")
print("If the content checks above all pass, we can safely build/run the final resumable evaluation engine.")
```

```python
# ======================================================================================================================
# CELL 13C — FINAL END-TO-END METHODOLOGICAL + IMPLEMENTATION AUDIT
#
# PURPOSE:
#   Verify the complete evaluation pathway before launching the 2700-job experiment.
#
# THIS CELL:
#   - DOES load REVE
#   - DOES load EEG
#   - DOES run a very small number of forwards
#   - DOES train ONE tiny classifier smoke test
#   - DOES NOT launch the 2700-job experiment
#   - DOES NOT modify the archived REVE model
# ======================================================================================================================

import os
import copy
import random
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader
from transformers import AutoModel

print("\n" + "=" * 120)
print("CELL 13C — FINAL END-TO-END METHODOLOGICAL + IMPLEMENTATION AUDIT")
print("=" * 120)

# ----------------------------------------------------------------------------------------------------------------------
# 0. DEVICE
# ----------------------------------------------------------------------------------------------------------------------

assert torch.cuda.is_available(), "CUDA is required."

device = torch.device("cuda")

print("\nDEVICE")
print("GPU:", torch.cuda.get_device_name(0))

# ----------------------------------------------------------------------------------------------------------------------
# 1. CONSTANTS — LOCKED EXPERIMENT DEFINITION
# ----------------------------------------------------------------------------------------------------------------------

EXPECTED_SUBJECTS = [
    "PN00", "PN01", "PN06", "PN07",
    "PN09", "PN10", "PN11", "PN12",
    "PN13", "PN14", "PN16", "PN17"
]

CLASSIFIER_SEEDS = [20, 25, 30, 35, 40]

METHODS = [
    "Random",
    "Functional",
    "DDF",
    "SSN",
    "Ours"
]

PRUNING_LEVELS = [10, 20, 30, 40, 50, 60, 70, 80, 90]

REMOVE_COUNTS = {
    10: 2,
    20: 4,
    30: 7,
    40: 9,
    50: 11,
    60: 13,
    70: 15,
    80: 18,
    90: 20,
}

EEG_CHANNELS_22 = [
    "Fp1", "F3", "C3", "P3", "O1",
    "F7", "T5", "Fc1", "Fc5", "Cp1",
    "Cp5", "F9", "Fz", "Pz", "F4",
    "C4", "P4", "O2", "F8", "T4",
    "T6", "Fc2"
]

BASE_DIR = "/content/drive/MyDrive/Seinna_REVE_LOSO"
NOVELTY3_DIR = os.path.join(BASE_DIR, "novelty3_pruning")

SOURCE_DIR = os.path.join(
    BASE_DIR,
    "subject_epochs_22ch"
)

REVE_STATE = (
    "/content/drive/MyDrive/Seinna_REVE_Novelty3_NEW/"
    "EXPERIMENT_ARCHIVE/REVE_MODEL/reve_base_state_dict.pt"
)

POSITION_STATE = (
    "/content/drive/MyDrive/Seinna_REVE_Novelty3_NEW/"
    "EXPERIMENT_ARCHIVE/REVE_POSITION_MODEL/"
    "reve_positions_state_dict.pt"
)

DDF_PATH = os.path.join(
    NOVELTY3_DIR,
    "rankings",
    "ddf_block_ranking.csv"
)

SSN_PATH = os.path.join(
    NOVELTY3_DIR,
    "rankings",
    "ssn_block_ranking.csv"
)

FUNCTIONAL_PATH = os.path.join(
    NOVELTY3_DIR,
    "rankings",
    "functional_block_ranking.csv"
)

OURS_SCORES = os.path.join(
    NOVELTY3_DIR,
    "gate_training",
    "ours_block_scores.csv"
)

PRUNING_PLANS = os.path.join(
    NOVELTY3_DIR,
    "pruning_plans.csv"
)

print("\nEXPERIMENT CONSTANTS")
print("Subjects:", len(EXPECTED_SUBJECTS))
print("Methods:", METHODS)
print("Pruning levels:", PRUNING_LEVELS)
print("Classifier seeds:", CLASSIFIER_SEEDS)

assert len(EXPECTED_SUBJECTS) == 12
assert len(METHODS) == 5
assert len(PRUNING_LEVELS) == 9
assert len(CLASSIFIER_SEEDS) == 5

# ----------------------------------------------------------------------------------------------------------------------
# 2. SOURCE EEG — RECHECK EXACTLY
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("1. SOURCE EEG RECHECK")
print("-" * 120)

assert os.path.isdir(SOURCE_DIR)

source_files = {}

for subject in EXPECTED_SUBJECTS:

    path = os.path.join(
        SOURCE_DIR,
        f"{subject}_epochs.npz"
    )

    assert os.path.isfile(path), (
        f"Missing source file for {subject}: {path}"
    )

    source_files[subject] = path

    data = np.load(path)

    assert "X" in data.files
    assert "y" in data.files

    X = data["X"]
    y = data["y"]

    assert X.ndim == 3
    assert X.shape[1:] == (1000, 22)
    assert y.ndim == 1
    assert X.shape[0] == y.shape[0]

    assert np.isfinite(X).all()
    assert np.isfinite(y).all()

    assert set(np.unique(y).tolist()).issubset({0, 1})

    print(
        f"PASS: {subject} | "
        f"X={X.shape} | "
        f"y={y.shape}"
    )

    del data, X, y

assert len(source_files) == 12

# ----------------------------------------------------------------------------------------------------------------------
# 3. VERIFY RANKINGS AGAIN
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("2. RANKING RECHECK")
print("-" * 120)

def load_block_order(path, name):

    assert os.path.isfile(path), f"Missing {name}: {path}"

    df = pd.read_csv(path)

    assert len(df) == 22, (
        f"{name}: expected 22 rows, got {len(df)}"
    )

    assert "block" in df.columns, (
        f"{name}: expected block column."
    )

    blocks = df["block"].astype(int).tolist()

    assert len(set(blocks)) == 22
    assert set(blocks) == set(range(1, 23))

    print(f"PASS: {name} ranking")

    return blocks, df


ddf_order, ddf_df = load_block_order(
    DDF_PATH,
    "DDF"
)

ssn_order, ssn_df = load_block_order(
    SSN_PATH,
    "SSN"
)

functional_df = pd.read_csv(FUNCTIONAL_PATH)

assert len(functional_df) == 22
assert "block" in functional_df.columns

functional_order = (
    functional_df
    .sort_values(
        "rank",
        ascending=True
    )["block"]
    .astype(int)
    .tolist()
)

assert set(functional_order) == set(range(1, 23))
assert len(set(functional_order)) == 22

print("PASS: Functional ranking")

# ----------------------------------------------------------------------------------------------------------------------
# 4. OURS — USE CURRENT VERIFIED SCORE FILE
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("3. OURS RECHECK")
print("-" * 120)

ours_df = pd.read_csv(OURS_SCORES)

assert len(ours_df) == 22
assert set(ours_df["block"].astype(int)) == set(range(1, 23))

assert "ours_score" in ours_df.columns

assert np.isfinite(
    ours_df["ours_score"].to_numpy(dtype=float)
).all()

ours_order = (
    ours_df
    .sort_values(
        "ours_score",
        ascending=False,
        kind="mergesort"
    )["block"]
    .astype(int)
    .tolist()
)

assert len(ours_order) == 22
assert len(set(ours_order)) == 22

EXPECTED_OURS_ORDER = [
    8, 10, 9, 2, 11, 14, 15, 6,
    16, 22, 21, 18, 3, 12, 19, 20,
    17, 5, 7, 4, 13, 1
]

assert ours_order == EXPECTED_OURS_ORDER, (
    f"Current Ours ranking changed.\n"
    f"Found:    {ours_order}\n"
    f"Expected: {EXPECTED_OURS_ORDER}"
)

print("PASS: Current Ours score ranking verified")

# ----------------------------------------------------------------------------------------------------------------------
# 5. PRUNING PLAN — AUTHORITATIVE
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("4. PRUNING PLAN RECHECK")
print("-" * 120)

plans = pd.read_csv(PRUNING_PLANS)

assert len(plans) == 9

assert plans["pruning_percent"].tolist() == PRUNING_LEVELS

for _, row in plans.iterrows():

    pct = int(row["pruning_percent"])
    remove_count = int(row["remove_count"])
    keep_count = int(row["keep_count"])

    assert remove_count == REMOVE_COUNTS[pct]
    assert keep_count == 22 - remove_count

    removed = [
        int(x.strip().replace("B", ""))
        for x in str(row["removed_blocks"]).split(",")
        if x.strip()
    ]

    expected = ours_order[:remove_count]

    assert removed == expected, (
        f"{pct}% Ours plan mismatch.\n"
        f"Found:    {removed}\n"
        f"Expected: {expected}"
    )

print("PASS: All nine Ours pruning plans match current Ours ranking")

# ----------------------------------------------------------------------------------------------------------------------
# 6. LOAD OFFICIAL REVE
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("5. LOADING OFFICIAL REVE")
print("-" * 120)

reve_model = AutoModel.from_pretrained(
    "brain-bzh/reve-base",
    trust_remote_code=True,
    torch_dtype=torch.float32
)

print("Loaded class:", type(reve_model).__name__)

# ----------------------------------------------------------------------------------------------------------------------
# 7. LOAD ARCHIVED TRAINED REVE WEIGHTS
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("6. LOADING ARCHIVED REVE WEIGHTS")
print("-" * 120)

reve_checkpoint = torch.load(
    REVE_STATE,
    map_location="cpu",
    weights_only=False
)

if isinstance(reve_checkpoint, dict):

    if "model_state_dict" in reve_checkpoint:
        reve_state = reve_checkpoint["model_state_dict"]

    elif "state_dict" in reve_checkpoint:
        reve_state = reve_checkpoint["state_dict"]

    else:
        # State-dict archive itself
        reve_state = reve_checkpoint

else:
    reve_state = reve_checkpoint

assert isinstance(reve_state, dict)

# Remove possible "reve." prefix if present.
clean_reve_state = {}

for key, value in reve_state.items():

    if key.startswith("reve."):
        clean_key = key[len("reve."):]
    else:
        clean_key = key

    clean_reve_state[clean_key] = value

load_result = reve_model.load_state_dict(
    clean_reve_state,
    strict=False
)

print("Missing keys:", len(load_result.missing_keys))
print("Unexpected keys:", len(load_result.unexpected_keys))

assert len(load_result.missing_keys) == 0, (
    f"Missing REVE weights: {load_result.missing_keys[:20]}"
)

assert len(load_result.unexpected_keys) == 0, (
    f"Unexpected REVE weights: {load_result.unexpected_keys[:20]}"
)

print("PASS: REVE checkpoint loaded with zero missing keys")
print("PASS: REVE checkpoint loaded with zero unexpected keys")

del reve_checkpoint
del reve_state
del clean_reve_state
del load_result

# ----------------------------------------------------------------------------------------------------------------------
# 8. VERIFY REVE STRUCTURE
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("7. REVE ARCHITECTURE AUDIT")
print("-" * 120)

assert hasattr(reve_model, "transformer")
assert hasattr(reve_model.transformer, "layers")

layers = reve_model.transformer.layers

assert len(layers) == 22, (
    f"Expected 22 transformer blocks, got {len(layers)}"
)

print("PASS: exactly 22 REVE transformer blocks")

for i, layer in enumerate(layers):

    assert isinstance(layer, (nn.ModuleList, list, tuple)), (
        f"Block {i+1} has unexpected structure: {type(layer)}"
    )

    assert len(layer) == 2, (
        f"Block {i+1} does not contain attention + FF pair."
    )

print("PASS: every block contains attention + FF")

# ----------------------------------------------------------------------------------------------------------------------
# 9. FREEZE REVE
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("8. FREEZING REVE")
print("-" * 120)

for parameter in reve_model.parameters():
    parameter.requires_grad = False

trainable_reve = [
    name
    for name, parameter in reve_model.named_parameters()
    if parameter.requires_grad
]

assert len(trainable_reve) == 0

print("PASS: all REVE parameters frozen")

# ----------------------------------------------------------------------------------------------------------------------
# 10. LOAD POSITION MODEL
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("9. POSITION MODEL")
print("-" * 120)

position_model = AutoModel.from_pretrained(
    "brain-bzh/reve-positions",
    trust_remote_code=True,
    torch_dtype=torch.float32
)

position_checkpoint = torch.load(
    POSITION_STATE,
    map_location="cpu",
    weights_only=False
)

if isinstance(position_checkpoint, dict):

    if "state_dict" in position_checkpoint:
        position_state = position_checkpoint["state_dict"]

    elif "model_state_dict" in position_checkpoint:
        position_state = position_checkpoint["model_state_dict"]

    else:
        position_state = position_checkpoint

else:
    position_state = position_checkpoint

assert isinstance(position_state, dict)

clean_position_state = {}

for key, value in position_state.items():

    if key.startswith("position_model."):
        clean_key = key[len("position_model."):]
    elif key.startswith("reve_positions."):
        clean_key = key[len("reve_positions."):]
    else:
        clean_key = key

    clean_position_state[clean_key] = value

position_result = position_model.load_state_dict(
    clean_position_state,
    strict=False
)

print(
    "Position missing:",
    len(position_result.missing_keys)
)

print(
    "Position unexpected:",
    len(position_result.unexpected_keys)
)

assert len(position_result.missing_keys) == 0
assert len(position_result.unexpected_keys) == 0

position_model.eval()

for p in position_model.parameters():
    p.requires_grad = False

with torch.no_grad():
    positions = position_model(EEG_CHANNELS_22)

print("Position type:", type(positions))

if isinstance(positions, torch.Tensor):
    print("Position shape:", tuple(positions.shape))

    assert torch.isfinite(positions).all()

else:
    print("Position output is non-tensor; retaining official output.")
    assert positions is not None

print("PASS: position model verified")

# ----------------------------------------------------------------------------------------------------------------------
# 11. POSITION BATCH PREPARATION
# ----------------------------------------------------------------------------------------------------------------------

def prepare_positions(pos, batch_size, device):

    if isinstance(pos, torch.Tensor):

        pos = pos.to(device=device, dtype=torch.float32)

        if pos.ndim == 2:
            pos = pos.unsqueeze(0)

        if pos.shape[0] == 1 and batch_size > 1:
            pos = pos.expand(batch_size, *pos.shape[1:])

        assert pos.shape[0] == batch_size

        return pos

    return pos


# ----------------------------------------------------------------------------------------------------------------------
# 12. STRUCTURED PRUNING FUNCTION
# ----------------------------------------------------------------------------------------------------------------------

def extract_features(
    model,
    eeg,
    positions,
    remove_blocks=None
):

    if remove_blocks is None:
        remove_blocks = []

    remove_blocks = sorted(
        set(int(x) for x in remove_blocks)
    )

    assert all(
        1 <= b <= 22
        for b in remove_blocks
    )

    original_layers = model.transformer.layers

    try:

        modified_layers = nn.ModuleList()

        for block_number, layer in enumerate(
            original_layers,
            start=1
        ):

            if block_number in remove_blocks:

                modified_layers.append(
                    nn.ModuleList([
                        nn.Identity(),
                        nn.Identity()
                    ])
                )

            else:

                modified_layers.append(layer)

        model.transformer.layers = modified_layers

        batch_size = eeg.shape[0]

        pos = prepare_positions(
            positions,
            batch_size,
            eeg.device
        )

        with torch.no_grad():

            output = model(
                eeg,
                pos
            )

        assert isinstance(output, torch.Tensor)

        assert output.ndim == 4, (
            f"Expected REVE output 4D, got {output.shape}"
        )

        assert output.shape[1] == 22
        assert output.shape[2] == 5
        assert output.shape[3] == 512

        assert torch.isfinite(output).all()

        # (B,22,5,512)
        # mean over the five temporal patches
        pooled = output.mean(dim=2)

        # (B,22,512)
        channel_mean = pooled.mean(dim=1)

        channel_std = pooled.std(
            dim=1,
            unbiased=False
        )

        # (B,1024)
        features = torch.cat(
            [
                channel_mean,
                channel_std
            ],
            dim=1
        )

        assert features.shape == (
            batch_size,
            1024
        )

        assert torch.isfinite(features).all()

        return features

    finally:

        model.transformer.layers = original_layers


# ----------------------------------------------------------------------------------------------------------------------
# 13. LOAD ONE SUBJECT FOR SMOKE TEST
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("10. END-TO-END REVE + PRUNING SMOKE TEST")
print("-" * 120)

TEST_SUBJECT = "PN00"

test_data = np.load(
    source_files[TEST_SUBJECT]
)

X_test_subject = test_data["X"].astype(
    np.float32,
    copy=False
)

y_test_subject = test_data["y"].astype(
    np.int64,
    copy=False
)

# Use only TWO epochs for representation test.
X_small = torch.from_numpy(
    np.transpose(
        X_test_subject[:2],
        (0, 2, 1)
    )
)

assert X_small.shape == (
    2, 22, 1000
)

X_small = X_small.to(device)

print(
    f"PASS: {TEST_SUBJECT} smoke-test EEG =",
    tuple(X_small.shape)
)

# ----------------------------------------------------------------------------------------------------------------------
# 14. MOVE MODELS TO GPU
# ----------------------------------------------------------------------------------------------------------------------

reve_model = reve_model.to(device)
position_model = position_model.to(device)

reve_model.eval()
position_model.eval()

with torch.no_grad():
    test_positions = position_model(
        EEG_CHANNELS_22
    )

# ----------------------------------------------------------------------------------------------------------------------
# 15. BASELINE FORWARD
# ----------------------------------------------------------------------------------------------------------------------

features_base = extract_features(
    reve_model,
    X_small,
    test_positions,
    remove_blocks=[]
)

assert features_base.shape == (2, 1024)

print(
    "PASS: 0% pruning representation:",
    tuple(features_base.shape)
)

# ----------------------------------------------------------------------------------------------------------------------
# 16. SINGLE-BLOCK PRUNING FORWARD
# ----------------------------------------------------------------------------------------------------------------------

features_b8 = extract_features(
    reve_model,
    X_small,
    test_positions,
    remove_blocks=[8]
)

assert features_b8.shape == (2, 1024)

diff_b8 = torch.mean(
    torch.abs(
        features_base - features_b8
    )
).item()

assert np.isfinite(diff_b8)
assert diff_b8 > 0

print(
    f"PASS: B8 pruning changes representation "
    f"(mean |Δ| = {diff_b8:.8f})"
)

# ----------------------------------------------------------------------------------------------------------------------
# 17. CURRENT OURS 10% PRUNING
# ----------------------------------------------------------------------------------------------------------------------

ours_10 = ours_order[:REMOVE_COUNTS[10]]

assert ours_10 == [8, 10]

features_ours10 = extract_features(
    reve_model,
    X_small,
    test_positions,
    remove_blocks=ours_10
)

assert features_ours10.shape == (2, 1024)

diff_ours10 = torch.mean(
    torch.abs(
        features_base - features_ours10
    )
).item()

assert np.isfinite(diff_ours10)
assert diff_ours10 > 0

print(
    f"PASS: Ours 10% pruning changes representation "
    f"(mean |Δ| = {diff_ours10:.8f})"
)

# ----------------------------------------------------------------------------------------------------------------------
# 18. VERIFY ARCHITECTURE RESTORATION
# ----------------------------------------------------------------------------------------------------------------------

restored_layers = reve_model.transformer.layers

assert len(restored_layers) == 22

for i, layer in enumerate(restored_layers):

    assert len(layer) == 2

    assert not isinstance(
        layer[0],
        nn.Identity
    )

    assert not isinstance(
        layer[1],
        nn.Identity
    )

print("PASS: original REVE architecture restored after pruning")

# ----------------------------------------------------------------------------------------------------------------------
# 19. VERIFY NO REVE GRADIENTS
# ----------------------------------------------------------------------------------------------------------------------

trainable_reve = [
    name
    for name, parameter in reve_model.named_parameters()
    if parameter.requires_grad
]

assert len(trainable_reve) == 0

print("PASS: REVE remains completely frozen")

# ----------------------------------------------------------------------------------------------------------------------
# 20. VERIFY LOSO TEST IS EXCLUDED FROM TRAINING
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("11. LOSO LEAKAGE AUDIT")
print("-" * 120)

train_subjects = [
    s for s in EXPECTED_SUBJECTS
    if s != TEST_SUBJECT
]

assert TEST_SUBJECT not in train_subjects
assert len(train_subjects) == 11
assert set(train_subjects).isdisjoint({TEST_SUBJECT})

print("Held-out test subject:", TEST_SUBJECT)
print("Inner train subjects:", train_subjects)

print("PASS: held-out subject excluded from inner training set")

# Verify there is no accidental overlap by subject identity.
assert set(EXPECTED_SUBJECTS) == (
    set(train_subjects) | {TEST_SUBJECT}
)

assert len(
    set(train_subjects) | {TEST_SUBJECT}
) == 12

print("PASS: 12-subject LOSO partition is exact")

# ----------------------------------------------------------------------------------------------------------------------
# 21. CREATE INNER TRAIN / VALIDATION SPLIT
#
# IMPORTANT:
#   We do NOT use PN00 for validation.
#   Validation is sampled only from the other 11 subjects.
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("12. INNER VALIDATION AUDIT")
print("-" * 120)

# To keep this smoke test small, use at most 4 epochs/subject.
INNER_MAX_EPOCHS_PER_SUBJECT = 4

rng = np.random.default_rng(2026)

train_features_list = []
train_labels_list = []
train_ids = []

for subject in train_subjects:

    data = np.load(
        source_files[subject]
    )

    X_subject = data["X"].astype(
        np.float32,
        copy=False
    )

    y_subject = data["y"].astype(
        np.int64,
        copy=False
    )

    n_take = min(
        INNER_MAX_EPOCHS_PER_SUBJECT,
        len(X_subject)
    )

    selected = np.arange(
        len(X_subject)
    )

    rng.shuffle(selected)

    selected = selected[:n_take]

    X_subject_small = torch.from_numpy(
        np.transpose(
            X_subject[selected],
            (0, 2, 1)
        )
    ).to(device)

    features = extract_features(
        reve_model,
        X_subject_small,
        test_positions,
        remove_blocks=ours_10
    )

    labels = torch.from_numpy(
        y_subject[selected]
    ).long().to(device)

    assert features.shape == (
        n_take,
        1024
    )

    assert labels.shape == (
        n_take,
    )

    train_features_list.append(
        features.detach().cpu()
    )

    train_labels_list.append(
        labels.detach().cpu()
    )

    train_ids.extend(
        [subject] * n_take
    )

    del data
    del X_subject
    del y_subject
    del X_subject_small
    del features
    del labels

inner_X = torch.cat(
    train_features_list,
    dim=0
)

inner_y = torch.cat(
    train_labels_list,
    dim=0
)

assert inner_X.shape[0] == len(train_ids)
assert inner_X.shape[1] == 1024

assert TEST_SUBJECT not in train_ids

print(
    "Inner dataset shape:",
    tuple(inner_X.shape)
)

print(
    "Inner subjects:",
    sorted(set(train_ids))
)

print(
    "Held-out test subject present in inner data:",
    TEST_SUBJECT in train_ids
)

assert TEST_SUBJECT not in train_ids

print(
    "PASS: test subject is completely absent from inner dataset"
)

# ----------------------------------------------------------------------------------------------------------------------
# 22. BUILD LOCKED CLASSIFIER
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("13. CLASSIFIER SMOKE TEST")
print("-" * 120)

def build_classifier():

    model = nn.Sequential(
        nn.Linear(1024, 512),
        nn.ReLU(),
        nn.Dropout(0.2),
        nn.Linear(512, 128),
        nn.ReLU(),
        nn.Dropout(0.2),
        nn.Linear(128, 2)
    )

    return model


classifier = build_classifier()

parameter_count = sum(
    p.numel()
    for p in classifier.parameters()
)

assert parameter_count == 590722

print(
    "PASS: classifier parameters =",
    parameter_count
)

classifier = classifier.to(device)

# ----------------------------------------------------------------------------------------------------------------------
# 23. TRAIN ONLY CLASSIFIER — VERY SMALL SMOKE TEST
# ----------------------------------------------------------------------------------------------------------------------

smoke_seed = 20

random.seed(smoke_seed)
np.random.seed(smoke_seed)
torch.manual_seed(smoke_seed)
torch.cuda.manual_seed_all(smoke_seed)

classifier.train()

optimizer = torch.optim.AdamW(
    classifier.parameters(),
    lr=1e-3,
    weight_decay=1e-4
)

criterion = nn.CrossEntropyLoss()

loader = DataLoader(
    TensorDataset(
        inner_X,
        inner_y
    ),
    batch_size=8,
    shuffle=True,
    generator=torch.Generator().manual_seed(smoke_seed)
)

initial_loss = None
final_loss = None

for epoch in range(2):

    running_loss = 0.0
    n_seen = 0

    for batch_X, batch_y in loader:

        batch_X = batch_X.to(device)
        batch_y = batch_y.to(device)

        optimizer.zero_grad(
            set_to_none=True
        )

        logits = classifier(
            batch_X
        )

        loss = criterion(
            logits,
            batch_y
        )

        assert torch.isfinite(loss)

        loss.backward()

        optimizer.step()

        running_loss += (
            loss.item() * batch_X.shape[0]
        )

        n_seen += batch_X.shape[0]

    epoch_loss = running_loss / n_seen

    if initial_loss is None:
        initial_loss = epoch_loss

    final_loss = epoch_loss

    print(
        f"Smoke epoch {epoch + 1}/2 "
        f"| loss={epoch_loss:.6f}"
    )

assert np.isfinite(initial_loss)
assert np.isfinite(final_loss)

print("PASS: classifier training completed")

# ----------------------------------------------------------------------------------------------------------------------
# 24. VERIFY REVE STILL HAS NO GRADIENTS
# ----------------------------------------------------------------------------------------------------------------------

reve_has_grad = False

for name, parameter in reve_model.named_parameters():

    if parameter.grad is not None:

        if torch.any(
            parameter.grad != 0
        ):
            reve_has_grad = True
            print(
                "Unexpected REVE gradient:",
                name
            )
            break

assert not reve_has_grad

print("PASS: no REVE gradients produced")

# ----------------------------------------------------------------------------------------------------------------------
# 25. FINAL TEST FOR REPRESENTATION ON HELD-OUT SUBJECT
#
# This is only a forward smoke test — no training on PN00.
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "-" * 120)
print("14. HELD-OUT TEST FORWARD AUDIT")
print("-" * 120)

test_X = torch.from_numpy(
    np.transpose(
        X_test_subject[:4],
        (0, 2, 1)
    )
).to(device)

test_features = extract_features(
    reve_model,
    test_X,
    test_positions,
    remove_blocks=ours_10
)

assert test_features.shape == (
    min(4, len(X_test_subject)),
    1024
)

assert torch.isfinite(
    test_features
).all()

print(
    "PASS: held-out test representation:",
    tuple(test_features.shape)
)

# Classifier prediction ONLY.
classifier.eval()

with torch.no_grad():

    test_logits = classifier(
        test_features
    )

assert test_logits.shape == (
    test_features.shape[0],
    2
)

assert torch.isfinite(
    test_logits
).all()

print(
    "PASS: held-out test prediction:",
    tuple(test_logits.shape)
)

# ----------------------------------------------------------------------------------------------------------------------
# 26. FINAL CLEANUP
# ----------------------------------------------------------------------------------------------------------------------

del test_data
del X_test_subject
del y_test_subject
del X_small
del features_base
del features_b8
del features_ours10
del test_X
del test_features
del test_logits
del inner_X
del inner_y
del classifier
del optimizer
del criterion
del loader

torch.cuda.empty_cache()

# ----------------------------------------------------------------------------------------------------------------------
# FINAL RESULT
# ----------------------------------------------------------------------------------------------------------------------

print("\n" + "=" * 120)
print("CELL 13C — AUDIT COMPLETE")
print("=" * 120)

print("PASS: Exact 12-subject source verified")
print("PASS: All ranking inputs verified")
print("PASS: Current Ours ranking verified")
print("PASS: Ours pruning plans verified")
print("PASS: Official REVE checkpoint verified")
print("PASS: 22 REVE blocks verified")
print("PASS: REVE completely frozen")
print("PASS: Position model verified")
print("PASS: Official REVE output = (B,22,5,512)")
print("PASS: pooled representation = (B,1024)")
print("PASS: structured block pruning verified")
print("PASS: architecture restoration verified")
print("PASS: LOSO held-out subject excluded from inner training")
print("PASS: inner validation pathway audited")
print("PASS: classifier architecture verified")
print("PASS: classifier-only training smoke test completed")
print("PASS: no REVE gradients")
print("PASS: held-out test forward completed")
print()
print("NO 2700-JOB EXPERIMENT WAS STARTED.")
print("=" * 120)
```

```python
# ======================================================================================================================
# CELL 13D — FINAL 12-POINT PRE-EXPERIMENT AUDIT
#
# PURPOSE:
#   Verify every remaining methodological + engineering condition before the
#   full 2700-job experiment.
#
# CHECKS:
#   1.  Method-specific pruning plans
#   2.  Random ranking reproducibility
#   3.  2700 manifest <-> rankings
#   4.  Inner validation construction
#   5.  Class balance in inner validation
#   6.  Classifier hyperparameters
#   7.  Metric edge cases
#   8.  Checkpoint integrity / resume
#   9.  Feature-cache isolation
#  10.  REVE mutation / restoration
#  11.  GPU memory behavior
#  12.  One COMPLETE representative job
#
# IMPORTANT:
#   - Does NOT launch 2700 jobs.
#   - Does NOT overwrite experiment files.
#   - Uses temporary audit files only.
# ======================================================================================================================

import os
import gc
import json
import copy
import random
import shutil
import hashlib
import tempfile
import numpy as np
import pandas as pd

import torch
import torch.nn as nn

from torch.utils.data import TensorDataset, DataLoader
from transformers import AutoModel

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

print("\n" + "=" * 120)
print("CELL 13D — FINAL 12-POINT PRE-EXPERIMENT AUDIT")
print("=" * 120)


# ======================================================================================================================
# 0. LOCKED CONFIGURATION
# ======================================================================================================================

EXPECTED_SUBJECTS = [
    "PN00", "PN01", "PN06", "PN07",
    "PN09", "PN10", "PN11", "PN12",
    "PN13", "PN14", "PN16", "PN17"
]

METHODS = [
    "Random",
    "Functional",
    "DDF",
    "SSN",
    "Ours"
]

PRUNING_LEVELS = [
    10, 20, 30, 40, 50,
    60, 70, 80, 90
]

REMOVE_COUNTS = {
    10: 2,
    20: 4,
    30: 7,
    40: 9,
    50: 11,
    60: 13,
    70: 15,
    80: 18,
    90: 20
}

CLASSIFIER_SEEDS = [20, 25, 30, 35, 40]

# LOCKED BASELINE SETTINGS
CLASSIFIER_BATCH_SIZE = 32
CLASSIFIER_MAX_EPOCHS = 100
CLASSIFIER_LR = 1e-3
CLASSIFIER_WEIGHT_DECAY = 1e-4
CLASSIFIER_PATIENCE = 15

DROPOUT = 0.2

# Fixed representative job
AUDIT_TEST_SUBJECT = "PN00"
AUDIT_METHOD = "Ours"
AUDIT_PRUNING_PERCENT = 10
AUDIT_CLASSIFIER_SEED = 20

# Temporary directory only
AUDIT_TMP = "/content/reve_13d_audit_tmp"

if os.path.isdir(AUDIT_TMP):
    shutil.rmtree(AUDIT_TMP)

os.makedirs(AUDIT_TMP, exist_ok=True)

BASE_DIR = "/content/drive/MyDrive/Seinna_REVE_LOSO"
NOVELTY3_DIR = os.path.join(BASE_DIR, "novelty3_pruning")

SOURCE_DIR = os.path.join(
    BASE_DIR,
    "subject_epochs_22ch"
)

REVE_STATE = (
    "/content/drive/MyDrive/Seinna_REVE_Novelty3_NEW/"
    "EXPERIMENT_ARCHIVE/REVE_MODEL/reve_base_state_dict.pt"
)

POSITION_STATE = (
    "/content/drive/MyDrive/Seinna_REVE_Novelty3_NEW/"
    "EXPERIMENT_ARCHIVE/REVE_POSITION_MODEL/"
    "reve_positions_state_dict.pt"
)

DDF_PATH = os.path.join(
    NOVELTY3_DIR,
    "rankings",
    "ddf_block_ranking.csv"
)

SSN_PATH = os.path.join(
    NOVELTY3_DIR,
    "rankings",
    "ssn_block_ranking.csv"
)

FUNCTIONAL_PATH = os.path.join(
    NOVELTY3_DIR,
    "rankings",
    "functional_block_ranking.csv"
)

OURS_SCORES_PATH = os.path.join(
    NOVELTY3_DIR,
    "gate_training",
    "ours_block_scores.csv"
)

OURS_PLANS_PATH = os.path.join(
    NOVELTY3_DIR,
    "pruning_plans.csv"
)

MANIFEST_PATH = os.path.join(
    NOVELTY3_DIR,
    "pruning_2700_job_manifest.csv"
)

EEG_CHANNELS_22 = [
    "Fp1", "F3", "C3", "P3", "O1",
    "F7", "T5", "Fc1", "Fc5", "Cp1",
    "Cp5", "F9", "Fz", "Pz", "F4",
    "C4", "P4", "O2", "F8", "T4",
    "T6", "Fc2"
]

assert torch.cuda.is_available()

device = torch.device("cuda")

print("\nGPU:", torch.cuda.get_device_name(0))


# ======================================================================================================================
# 1. LOAD ALL RANKINGS
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 1 — METHOD-SPECIFIC PRUNING PLANS")
print("-" * 120)


def verify_permutation(df, block_col, name):

    assert len(df) == 22, (
        f"{name}: expected 22 rows, got {len(df)}"
    )

    blocks = (
        df[block_col]
        .astype(int)
        .tolist()
    )

    assert len(blocks) == 22
    assert len(set(blocks)) == 22
    assert set(blocks) == set(range(1, 23))

    return blocks


ddf_df = pd.read_csv(DDF_PATH)
ssn_df = pd.read_csv(SSN_PATH)
functional_df = pd.read_csv(FUNCTIONAL_PATH)
ours_scores_df = pd.read_csv(OURS_SCORES_PATH)

ddf_order = verify_permutation(
    ddf_df,
    "block",
    "DDF"
)

ssn_order = verify_permutation(
    ssn_df,
    "block",
    "SSN"
)

functional_order = (
    functional_df
    .sort_values(
        "rank",
        ascending=True
    )["block"]
    .astype(int)
    .tolist()
)

assert len(functional_order) == 22
assert len(set(functional_order)) == 22
assert set(functional_order) == set(range(1, 23))

ours_order = (
    ours_scores_df
    .sort_values(
        "ours_score",
        ascending=False,
        kind="mergesort"
    )["block"]
    .astype(int)
    .tolist()
)

assert len(ours_order) == 22
assert len(set(ours_order)) == 22
assert set(ours_order) == set(range(1, 23))

print("DDF top 5:", ddf_order[:5])
print("SSN top 5:", ssn_order[:5])
print("Functional top 5:", functional_order[:5])
print("Ours top 5:", ours_order[:5])

# ----------------------------------------------------------------------------------------------------------------------
# Generate deterministic Random ranking exactly once.
# ----------------------------------------------------------------------------------------------------------------------

RANDOM_RANKING_SEED = 20260917

rng_random = np.random.default_rng(
    RANDOM_RANKING_SEED
)

random_order = (
    rng_random
    .permutation(
        np.arange(1, 23)
    )
    .tolist()
)

assert len(random_order) == 22
assert len(set(random_order)) == 22
assert set(random_order) == set(range(1, 23))

RANKINGS = {
    "Random": random_order,
    "Functional": functional_order,
    "DDF": ddf_order,
    "SSN": ssn_order,
    "Ours": ours_order
}

for method, ranking in RANKINGS.items():

    assert len(ranking) == 22
    assert len(set(ranking)) == 22
    assert set(ranking) == set(range(1, 23))

print("PASS: all five methods have valid 22-block rankings")


# ======================================================================================================================
# 2. RANDOM REPRODUCIBILITY
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 2 — RANDOM RANKING REPRODUCIBILITY")
print("-" * 120)

def make_random_ranking(seed):

    rng = np.random.default_rng(seed)

    return (
        rng
        .permutation(
            np.arange(1, 23)
        )
        .tolist()
    )


random_a = make_random_ranking(
    RANDOM_RANKING_SEED
)

random_b = make_random_ranking(
    RANDOM_RANKING_SEED
)

assert random_a == random_b

print(
    "Random seed:",
    RANDOM_RANKING_SEED
)

print(
    "Random ranking:",
    random_a
)

print(
    "PASS: same seed produces exactly the same ranking"
)


# ======================================================================================================================
# 3. BUILD EXPECTED METHOD-SPECIFIC PRUNING CONFIGURATIONS
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 3 — METHOD-SPECIFIC PRUNING CONFIGURATIONS")
print("-" * 120)

method_plans = {}

for method in METHODS:

    ranking = RANKINGS[method]

    method_plans[method] = {}

    for pct in PRUNING_LEVELS:

        remove_count = REMOVE_COUNTS[pct]

        removed = ranking[:remove_count]
        kept = [
            b for b in range(1, 23)
            if b not in removed
        ]

        assert len(removed) == remove_count
        assert len(kept) == 22 - remove_count
        assert set(removed).isdisjoint(set(kept))
        assert set(removed) | set(kept) == set(range(1, 23))

        method_plans[method][pct] = {
            "remove_count": remove_count,
            "keep_count": len(kept),
            "removed_blocks": removed,
            "kept_blocks": kept
        }

        print(
            f"{method:10s} | "
            f"{pct:2d}% | "
            f"remove={removed}"
        )

print("PASS: all 45 method × pruning configurations verified")


# ======================================================================================================================
# 4. VERIFY EXISTING 2700 MANIFEST AGAINST THE RANKINGS
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 4 — 2700 MANIFEST <-> RANKINGS")
print("-" * 120)

manifest = pd.read_csv(MANIFEST_PATH)

assert len(manifest) == 2700

required_manifest_columns = [
    "job_id",
    "method",
    "pruning_percent",
    "remove_count",
    "keep_count",
    "subject",
    "classifier_seed",
    "random_ranking_seed",
    "ranking_type",
    "ranking_source",
    "removed_blocks",
    "kept_blocks"
]

for column in required_manifest_columns:

    assert column in manifest.columns, (
        f"Missing manifest column: {column}"
    )

assert set(manifest["method"]) == set(METHODS)
assert set(
    manifest["pruning_percent"].astype(int)
) == set(PRUNING_LEVELS)

assert set(
    manifest["subject"]
) == set(EXPECTED_SUBJECTS)

assert set(
    manifest["classifier_seed"].astype(int)
) == set(CLASSIFIER_SEEDS)

# Every method × level × subject × classifier seed
# must occur exactly once.
group_counts = (
    manifest
    .groupby(
        [
            "method",
            "pruning_percent",
            "subject",
            "classifier_seed"
        ]
    )
    .size()
)

assert len(group_counts) == 2700
assert group_counts.eq(1).all()

print("PASS: exactly 2700 unique method × level × subject × seed jobs")

# ----------------------------------------------------------------------------------------------------------------------
# Verify every row's removed_blocks against its method-specific ranking.
# ----------------------------------------------------------------------------------------------------------------------

def parse_blocks(value):

    if pd.isna(value):
        return []

    text_value = str(value).strip()

    if text_value == "":
        return []

    return [
        int(
            item.strip()
            .replace("B", "")
            .replace("b", "")
        )
        for item in text_value.split(",")
        if item.strip()
    ]


for idx, row in manifest.iterrows():

    method = str(row["method"])
    pct = int(row["pruning_percent"])

    expected_plan = method_plans[
        method
    ][pct]

    actual_removed = parse_blocks(
        row["removed_blocks"]
    )

    actual_kept = parse_blocks(
        row["kept_blocks"]
    )

    assert actual_removed == expected_plan["removed_blocks"], (
        f"Manifest row {idx}: {method} {pct}% "
        f"removed blocks mismatch.\n"
        f"Actual:   {actual_removed}\n"
        f"Expected: {expected_plan['removed_blocks']}"
    )

    assert actual_kept == expected_plan["kept_blocks"], (
        f"Manifest row {idx}: {method} {pct}% "
        f"kept blocks mismatch."
    )

    assert int(row["remove_count"]) == (
        expected_plan["remove_count"]
    )

    assert int(row["keep_count"]) == (
        expected_plan["keep_count"]
    )

print("PASS: every one of 2700 manifest rows matches its method-specific ranking")

# ----------------------------------------------------------------------------------------------------------------------
# Random seed must be recorded consistently.
# ----------------------------------------------------------------------------------------------------------------------

random_manifest_seeds = (
    manifest[
        manifest["method"] == "Random"
    ]["random_ranking_seed"]
    .astype(int)
    .unique()
    .tolist()
)

assert random_manifest_seeds == [
    RANDOM_RANKING_SEED
], (
    f"Unexpected Random ranking seeds: {random_manifest_seeds}"
)

print("PASS: Random ranking seed is fixed and recorded consistently")


# ======================================================================================================================
# 5. SOURCE DATA + INNER VALIDATION CONSTRUCTION
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 5 — INNER VALIDATION CONSTRUCTION")
print("-" * 120)

source_files = {}

for subject in EXPECTED_SUBJECTS:

    path = os.path.join(
        SOURCE_DIR,
        f"{subject}_epochs.npz"
    )

    assert os.path.isfile(path)

    source_files[subject] = path

# Held-out subject is never part of inner data.
inner_subjects = [
    s for s in EXPECTED_SUBJECTS
    if s != AUDIT_TEST_SUBJECT
]

assert len(inner_subjects) == 11
assert AUDIT_TEST_SUBJECT not in inner_subjects

print(
    "Held-out test subject:",
    AUDIT_TEST_SUBJECT
)

print(
    "Inner subjects:",
    inner_subjects
)

# ----------------------------------------------------------------------------------------------------------------------
# Deterministic subject-disjoint inner validation split.
#
# We use subjects, not epochs, to avoid subject leakage between inner train
# and validation. Candidate validation subjects are selected deterministically
# from the 11 available subjects, but only after checking that their combined
# labels contain both classes.
# ----------------------------------------------------------------------------------------------------------------------

INNER_SPLIT_SEED = 20260918

candidate_subjects = inner_subjects.copy()

rng_inner = np.random.default_rng(
    INNER_SPLIT_SEED
)

rng_inner.shuffle(
    candidate_subjects
)

# Search for the smallest deterministic validation-subject set containing
# both classes, while leaving at least one other subject for training.
validation_subjects = None

for n_val_subjects in range(
    1,
    len(candidate_subjects)
):

    candidate_val = candidate_subjects[
        :n_val_subjects
    ]

    candidate_train = [
        s for s in inner_subjects
        if s not in candidate_val
    ]

    val_labels = []

    train_labels = []

    for subject in candidate_val:

        data = np.load(
            source_files[subject]
        )

        val_labels.extend(
            data["y"].astype(
                np.int64
            ).tolist()
        )

    for subject in candidate_train:

        data = np.load(
            source_files[subject]
        )

        train_labels.extend(
            data["y"].astype(
                np.int64
            ).tolist()
        )

    val_unique = set(val_labels)
    train_unique = set(train_labels)

    if (
        val_unique == {0, 1}
        and train_unique == {0, 1}
    ):

        validation_subjects = candidate_val
        training_subjects = candidate_train

        break

    del val_labels
    del train_labels

assert validation_subjects is not None
assert len(training_subjects) >= 1

assert set(validation_subjects).isdisjoint(
    set(training_subjects)
)

assert set(validation_subjects).isdisjoint(
    {AUDIT_TEST_SUBJECT}
)

assert set(training_subjects).isdisjoint(
    {AUDIT_TEST_SUBJECT}
)

print(
    "Inner validation subjects:",
    validation_subjects
)

print(
    "Inner training subjects:",
    training_subjects
)

print(
    "PASS: validation subjects are disjoint from training"
)

print(
    "PASS: held-out LOSO test subject is absent from both"
)


# ======================================================================================================================
# 6. CLASS BALANCE
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 6 — CLASS BALANCE IN INNER SPLIT")
print("-" * 120)

def count_subject_labels(subject_list):

    counts = {
        0: 0,
        1: 0
    }

    for subject in subject_list:

        data = np.load(
            source_files[subject]
        )

        labels = data["y"].astype(
            np.int64
        )

        unique, values = np.unique(
            labels,
            return_counts=True
        )

        for cls, count in zip(
            unique,
            values
        ):

            counts[int(cls)] += int(count)

    return counts


train_counts = count_subject_labels(
    training_subjects
)

val_counts = count_subject_labels(
    validation_subjects
)

assert train_counts[0] > 0
assert train_counts[1] > 0

assert val_counts[0] > 0
assert val_counts[1] > 0

print(
    "Inner training:",
    train_counts
)

print(
    "Inner validation:",
    val_counts
)

print(
    "PASS: both inner training and validation contain both classes"
)


# ======================================================================================================================
# 7. CLASSIFIER HYPERPARAMETERS
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 7 — LOCKED CLASSIFIER HYPERPARAMETERS")
print("-" * 120)

def build_classifier():

    model = nn.Sequential(
        nn.Linear(1024, 512),
        nn.ReLU(),
        nn.Dropout(DROPOUT),
        nn.Linear(512, 128),
        nn.ReLU(),
        nn.Dropout(DROPOUT),
        nn.Linear(128, 2)
    )

    return model


classifier_audit = build_classifier()

parameter_count = sum(
    p.numel()
    for p in classifier_audit.parameters()
)

assert parameter_count == 590722

assert CLASSIFIER_BATCH_SIZE == 32
assert CLASSIFIER_MAX_EPOCHS == 100
assert CLASSIFIER_LR == 1e-3
assert CLASSIFIER_WEIGHT_DECAY == 1e-4
assert CLASSIFIER_PATIENCE == 15
assert DROPOUT == 0.2

print(
    "Architecture:",
    classifier_audit
)

print(
    "Parameters:",
    parameter_count
)

print(
    "Batch size:",
    CLASSIFIER_BATCH_SIZE
)

print(
    "Max epochs:",
    CLASSIFIER_MAX_EPOCHS
)

print(
    "Learning rate:",
    CLASSIFIER_LR
)

print(
    "Weight decay:",
    CLASSIFIER_WEIGHT_DECAY
)

print(
    "Patience:",
    CLASSIFIER_PATIENCE
)

print("PASS: classifier architecture and locked hyperparameters verified")


# ======================================================================================================================
# 8. METRIC EDGE-CASE AUDIT
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 8 — METRIC EDGE CASES")
print("-" * 120)

def safe_binary_metrics(
    y_true,
    probabilities
):

    y_true = np.asarray(
        y_true,
        dtype=np.int64
    )

    probabilities = np.asarray(
        probabilities,
        dtype=np.float64
    )

    assert len(y_true) == len(probabilities)
    assert np.isfinite(probabilities).all()

    predictions = (
        probabilities >= 0.5
    ).astype(np.int64)

    result = {}

    result["accuracy"] = accuracy_score(
        y_true,
        predictions
    )

    result["balanced_accuracy"] = (
        balanced_accuracy_score(
            y_true,
            predictions
        )
    )

    result["f1"] = f1_score(
        y_true,
        predictions,
        zero_division=0
    )

    unique_classes = np.unique(
        y_true
    )

    if len(unique_classes) == 2:

        result["auroc"] = roc_auc_score(
            y_true,
            probabilities
        )

        result["auprc"] = average_precision_score(
            y_true,
            probabilities
        )

    else:

        result["auroc"] = np.nan
        result["auprc"] = np.nan

    return result


# Normal binary case
y_normal = np.array(
    [0, 0, 1, 1]
)

p_normal = np.array(
    [0.1, 0.2, 0.8, 0.9]
)

metrics_normal = safe_binary_metrics(
    y_normal,
    p_normal
)

assert np.isfinite(
    metrics_normal["auroc"]
)

assert np.isfinite(
    metrics_normal["auprc"]
)

# Single-class case
y_single = np.array(
    [0, 0, 0, 0]
)

p_single = np.array(
    [0.1, 0.2, 0.3, 0.4]
)

metrics_single = safe_binary_metrics(
    y_single,
    p_single
)

assert np.isfinite(
    metrics_single["accuracy"]
)

assert np.isfinite(
    metrics_single["balanced_accuracy"]
)

assert np.isfinite(
    metrics_single["f1"]
)

assert np.isnan(
    metrics_single["auroc"]
)

assert np.isnan(
    metrics_single["auprc"]
)

print(
    "PASS: normal binary metrics"
)

print(
    "PASS: single-class AUROC/AUPRC handled explicitly as NaN"
)


# ======================================================================================================================
# 9. CHECKPOINT INTEGRITY / RESUME AUDIT
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 9 — CHECKPOINT INTEGRITY / RESUME")
print("-" * 120)

checkpoint_test_dir = os.path.join(
    AUDIT_TMP,
    "checkpoint_test"
)

os.makedirs(
    checkpoint_test_dir,
    exist_ok=True
)

checkpoint_payload = {
    "job_id": 999999,
    "method": AUDIT_METHOD,
    "pruning_percent": AUDIT_PRUNING_PERCENT,
    "subject": AUDIT_TEST_SUBJECT,
    "classifier_seed": AUDIT_CLASSIFIER_SEED,
    "status": "complete",
    "test_metric": 0.123456
}

checkpoint_path = os.path.join(
    checkpoint_test_dir,
    "job_999999.pt"
)

tmp_checkpoint_path = (
    checkpoint_path + ".tmp"
)

torch.save(
    checkpoint_payload,
    tmp_checkpoint_path
)

assert os.path.isfile(
    tmp_checkpoint_path
)

os.replace(
    tmp_checkpoint_path,
    checkpoint_path
)

assert os.path.isfile(
    checkpoint_path
)

loaded_checkpoint = torch.load(
    checkpoint_path,
    map_location="cpu",
    weights_only=False
)

assert loaded_checkpoint == checkpoint_payload

# Verify incomplete checkpoint is not equivalent to completed.
incomplete_payload = dict(
    checkpoint_payload
)

incomplete_payload["status"] = "running"

incomplete_path = os.path.join(
    checkpoint_test_dir,
    "job_999998.pt"
)

torch.save(
    incomplete_payload,
    incomplete_path
)

loaded_incomplete = torch.load(
    incomplete_path,
    map_location="cpu",
    weights_only=False
)

assert loaded_incomplete["status"] != "complete"

print(
    "PASS: atomic checkpoint write/reload"
)

print(
    "PASS: incomplete checkpoint distinguishable from completed job"
)


# ======================================================================================================================
# 10. FEATURE-CACHE ISOLATION AUDIT
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 10 — FEATURE CACHE ISOLATION")
print("-" * 120)

def feature_cache_key(
    subject,
    method,
    pruning_percent
):

    raw = (
        f"{subject}|"
        f"{method}|"
        f"{pruning_percent}"
    )

    digest = hashlib.sha256(
        raw.encode("utf-8")
    ).hexdigest()[:16]

    return (
        f"{subject}__"
        f"{method}__"
        f"{pruning_percent:02d}__"
        f"{digest}"
    )


cache_keys = {}

for subject in EXPECTED_SUBJECTS:

    for method in METHODS:

        for pct in PRUNING_LEVELS:

            key = feature_cache_key(
                subject,
                method,
                pct
            )

            assert key not in cache_keys

            cache_keys[key] = (
                subject,
                method,
                pct
            )

expected_cache_entries = (
    12 * 5 * 9
)

assert len(cache_keys) == expected_cache_entries

# Explicitly verify changing method cannot collide.
test_keys = [
    feature_cache_key(
        AUDIT_TEST_SUBJECT,
        method,
        10
    )
    for method in METHODS
]

assert len(set(test_keys)) == 5

print(
    "Expected unique feature representations:",
    expected_cache_entries
)

print(
    "PASS: subject × method × pruning-level cache keys are unique"
)


# ======================================================================================================================
# 11. LOAD OFFICIAL REVE + POSITION MODEL
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 11 — REVE MUTATION + GPU MEMORY")
print("-" * 120)

reve_model = AutoModel.from_pretrained(
    "brain-bzh/reve-base",
    trust_remote_code=True,
    torch_dtype=torch.float32
)

reve_checkpoint = torch.load(
    REVE_STATE,
    map_location="cpu",
    weights_only=False
)

if isinstance(reve_checkpoint, dict):

    if "model_state_dict" in reve_checkpoint:
        reve_state = reve_checkpoint["model_state_dict"]

    elif "state_dict" in reve_checkpoint:
        reve_state = reve_checkpoint["state_dict"]

    else:
        reve_state = reve_checkpoint

else:
    reve_state = reve_checkpoint

clean_reve_state = {}

for key, value in reve_state.items():

    if key.startswith("reve."):
        clean_key = key[len("reve."):]
    else:
        clean_key = key

    clean_reve_state[clean_key] = value

result = reve_model.load_state_dict(
    clean_reve_state,
    strict=False
)

assert len(result.missing_keys) == 0
assert len(result.unexpected_keys) == 0

for p in reve_model.parameters():
    p.requires_grad = False

assert all(
    not p.requires_grad
    for p in reve_model.parameters()
)

position_model = AutoModel.from_pretrained(
    "brain-bzh/reve-positions",
    trust_remote_code=True,
    torch_dtype=torch.float32
)

position_checkpoint = torch.load(
    POSITION_STATE,
    map_location="cpu",
    weights_only=False
)

if isinstance(position_checkpoint, dict):

    if "state_dict" in position_checkpoint:
        position_state = position_checkpoint["state_dict"]

    elif "model_state_dict" in position_checkpoint:
        position_state = position_checkpoint["model_state_dict"]

    else:
        position_state = position_checkpoint

else:
    position_state = position_checkpoint

clean_position_state = {}

for key, value in position_state.items():

    if key.startswith("position_model."):
        clean_key = key[len("position_model."):]
    elif key.startswith("reve_positions."):
        clean_key = key[len("reve_positions."):]
    else:
        clean_key = key

    clean_position_state[clean_key] = value

position_result = position_model.load_state_dict(
    clean_position_state,
    strict=False
)

assert len(position_result.missing_keys) == 0
assert len(position_result.unexpected_keys) == 0

position_model.eval()

for p in position_model.parameters():
    p.requires_grad = False

reve_model.eval()

reve_model = reve_model.to(device)
position_model = position_model.to(device)

with torch.no_grad():

    positions = position_model(
        EEG_CHANNELS_22
    )

assert isinstance(
    positions,
    torch.Tensor
)

assert positions.shape == (
    22,
    3
)

assert torch.isfinite(
    positions
).all()

print(
    "PASS: official REVE loaded"
)

print(
    "PASS: official position model loaded"
)

print(
    "PASS: all REVE parameters frozen"
)


# ----------------------------------------------------------------------------------------------------------------------
# Official feature extractor
# ----------------------------------------------------------------------------------------------------------------------

def prepare_positions(
    pos,
    batch_size,
    device
):

    if isinstance(pos, torch.Tensor):

        pos = pos.to(
            device=device,
            dtype=torch.float32
        )

        if pos.ndim == 2:
            pos = pos.unsqueeze(0)

        if pos.shape[0] == 1 and batch_size > 1:
            pos = pos.expand(
                batch_size,
                *pos.shape[1:]
            )

        assert pos.shape[0] == batch_size

    return pos


def extract_features(
    model,
    eeg,
    positions,
    remove_blocks=None
):

    if remove_blocks is None:
        remove_blocks = []

    remove_blocks = sorted(
        set(
            int(b)
            for b in remove_blocks
        )
    )

    assert all(
        1 <= b <= 22
        for b in remove_blocks
    )

    original_layers = (
        model.transformer.layers
    )

    try:

        modified_layers = nn.ModuleList()

        for block_number, layer in enumerate(
            original_layers,
            start=1
        ):

            if block_number in remove_blocks:

                modified_layers.append(
                    nn.ModuleList([
                        nn.Identity(),
                        nn.Identity()
                    ])
                )

            else:

                modified_layers.append(
                    layer
                )

        model.transformer.layers = (
            modified_layers
        )

        pos = prepare_positions(
            positions,
            eeg.shape[0],
            eeg.device
        )

        with torch.no_grad():

            output = model(
                eeg,
                pos
            )

        assert isinstance(
            output,
            torch.Tensor
        )

        assert output.shape[1:] == (
            22,
            5,
            512
        )

        assert torch.isfinite(
            output
        ).all()

        pooled = output.mean(
            dim=2
        )

        channel_mean = pooled.mean(
            dim=1
        )

        channel_std = pooled.std(
            dim=1,
            unbiased=False
        )

        features = torch.cat(
            [
                channel_mean,
                channel_std
            ],
            dim=1
        )

        assert features.shape == (
            eeg.shape[0],
            1024
        )

        assert torch.isfinite(
            features
        ).all()

        return features

    finally:

        model.transformer.layers = (
            original_layers
        )


# ----------------------------------------------------------------------------------------------------------------------
# Verify restoration repeatedly.
# ----------------------------------------------------------------------------------------------------------------------

original_layers_reference = (
    reve_model.transformer.layers
)

data_pn00 = np.load(
    source_files[AUDIT_TEST_SUBJECT]
)

X_pn00 = data_pn00["X"].astype(
    np.float32,
    copy=False
)

y_pn00 = data_pn00["y"].astype(
    np.int64,
    copy=False
)

X_test_small = torch.from_numpy(
    np.transpose(
        X_pn00[:4],
        (0, 2, 1)
    )
).to(device)

features_0 = extract_features(
    reve_model,
    X_test_small,
    positions,
    remove_blocks=[]
)

features_b8 = extract_features(
    reve_model,
    X_test_small,
    positions,
    remove_blocks=[8]
)

features_ours = extract_features(
    reve_model,
    X_test_small,
    positions,
    remove_blocks=method_plans[
        "Ours"
    ][10]["removed_blocks"]
)

assert (
    reve_model.transformer.layers
    is original_layers_reference
)

assert len(
    reve_model.transformer.layers
) == 22

for layer in reve_model.transformer.layers:

    assert len(layer) == 2

    assert not isinstance(
        layer[0],
        nn.Identity
    )

    assert not isinstance(
        layer[1],
        nn.Identity
    )

diff_b8 = torch.mean(
    torch.abs(
        features_0 - features_b8
    )
).item()

diff_ours = torch.mean(
    torch.abs(
        features_0 - features_ours
    )
).item()

assert diff_b8 > 0
assert diff_ours > 0

print(
    f"PASS: B8 changes representation | mean |Δ|={diff_b8:.8f}"
)

print(
    f"PASS: Ours 10% changes representation | mean |Δ|={diff_ours:.8f}"
)

print(
    "PASS: REVE architecture restored after repeated pruning"
)


# ----------------------------------------------------------------------------------------------------------------------
# GPU memory stability
# ----------------------------------------------------------------------------------------------------------------------

torch.cuda.empty_cache()
gc.collect()

memory_before = (
    torch.cuda.memory_allocated()
)

for i in range(5):

    _ = extract_features(
        reve_model,
        X_test_small,
        positions,
        remove_blocks=(
            method_plans[
                "Ours"
            ][10]["removed_blocks"]
        )
    )

    del _

    torch.cuda.empty_cache()
    gc.collect()

memory_after = (
    torch.cuda.memory_allocated()
)

memory_reserved = (
    torch.cuda.memory_reserved()
)

print(
    "GPU allocated before:",
    memory_before
)

print(
    "GPU allocated after:",
    memory_after
)

print(
    "GPU reserved after:",
    memory_reserved
)

# A small allocator difference is acceptable.
# A large persistent accumulation is not.
memory_growth = (
    memory_after - memory_before
)

assert memory_growth < (
    512 * 1024 * 1024
), (
    f"Potential GPU memory accumulation: "
    f"{memory_growth / 1024**2:.2f} MB"
)

print(
    "PASS: no large persistent GPU memory accumulation detected"
)


# ======================================================================================================================
# 12. COMPLETE REPRESENTATIVE LOSO JOB
# ======================================================================================================================

print("\n" + "=" * 120)
print("CHECK 12 — COMPLETE REPRESENTATIVE LOSO JOB")
print("=" * 120)

print(
    "Test subject:",
    AUDIT_TEST_SUBJECT
)

print(
    "Method:",
    AUDIT_METHOD
)

print(
    "Pruning:",
    AUDIT_PRUNING_PERCENT,
    "%"
)

print(
    "Classifier seed:",
    AUDIT_CLASSIFIER_SEED
)

audit_removed_blocks = method_plans[
    AUDIT_METHOD
][AUDIT_PRUNING_PERCENT]["removed_blocks"]

print(
    "Removed blocks:",
    audit_removed_blocks
)

# ----------------------------------------------------------------------------------------------------------------------
# Build inner TRAIN representations.
#
# IMPORTANT:
#   - held-out PN00 is NEVER loaded here.
#   - training and validation are subject-disjoint.
#   - same pruning configuration is used for train and validation.
# ----------------------------------------------------------------------------------------------------------------------

inner_train_X_list = []
inner_train_y_list = []

inner_val_X_list = []
inner_val_y_list = []

for subject in training_subjects:

    data = np.load(
        source_files[subject]
    )

    X_subject = data["X"].astype(
        np.float32,
        copy=False
    )

    y_subject = data["y"].astype(
        np.int64,
        copy=False
    )

    X_tensor = torch.from_numpy(
        np.transpose(
            X_subject,
            (0, 2, 1)
        )
    ).to(device)

    # Process in bounded batches to avoid T4 OOM.
    subject_features = []

    for start in range(
        0,
        len(X_tensor),
        CLASSIFIER_BATCH_SIZE
    ):

        end = min(
            start + CLASSIFIER_BATCH_SIZE,
            len(X_tensor)
        )

        batch = X_tensor[
            start:end
        ]

        batch_features = extract_features(
            reve_model,
            batch,
            positions,
            remove_blocks=audit_removed_blocks
        )

        subject_features.append(
            batch_features.detach().cpu()
        )

    subject_features = torch.cat(
        subject_features,
        dim=0
    )

    labels_tensor = torch.from_numpy(
        y_subject
    ).long()

    assert subject_features.shape == (
        len(X_subject),
        1024
    )

    assert labels_tensor.shape == (
        len(X_subject),
    )

    if subject in validation_subjects:

        inner_val_X_list.append(
            subject_features
        )

        inner_val_y_list.append(
            labels_tensor
        )

    else:

        inner_train_X_list.append(
            subject_features
        )

        inner_train_y_list.append(
            labels_tensor
        )

    del data
    del X_subject
    del y_subject
    del X_tensor
    del subject_features
    del labels_tensor

    torch.cuda.empty_cache()
    gc.collect()


inner_train_X = torch.cat(
    inner_train_X_list,
    dim=0
)

inner_train_y = torch.cat(
    inner_train_y_list,
    dim=0
)

inner_val_X = torch.cat(
    inner_val_X_list,
    dim=0
)

inner_val_y = torch.cat(
    inner_val_y_list,
    dim=0
)

assert inner_train_X.shape[1] == 1024
assert inner_val_X.shape[1] == 1024

assert set(
    inner_train_y.numpy().tolist()
) == {0, 1}

assert set(
    inner_val_y.numpy().tolist()
) == {0, 1}

print(
    "Inner train:",
    tuple(inner_train_X.shape)
)

print(
    "Inner validation:",
    tuple(inner_val_X.shape)
)

print(
    "PASS: complete inner feature extraction"
)


# ----------------------------------------------------------------------------------------------------------------------
# Verify held-out subject was not used.
# ----------------------------------------------------------------------------------------------------------------------

assert AUDIT_TEST_SUBJECT not in training_subjects
assert AUDIT_TEST_SUBJECT not in validation_subjects

print(
    "PASS: PN00 remains completely excluded from classifier training/validation"
)


# ======================================================================================================================
# CLASSIFIER TRAINING — FULL LOCKED SETTINGS
# ======================================================================================================================

random.seed(
    AUDIT_CLASSIFIER_SEED
)

np.random.seed(
    AUDIT_CLASSIFIER_SEED
)

torch.manual_seed(
    AUDIT_CLASSIFIER_SEED
)

torch.cuda.manual_seed_all(
    AUDIT_CLASSIFIER_SEED
)

classifier = build_classifier().to(
    device
)

optimizer = torch.optim.AdamW(
    classifier.parameters(),
    lr=CLASSIFIER_LR,
    weight_decay=CLASSIFIER_WEIGHT_DECAY
)

criterion = nn.CrossEntropyLoss()

train_generator = (
    torch.Generator()
    .manual_seed(
        AUDIT_CLASSIFIER_SEED
    )
)

train_loader = DataLoader(
    TensorDataset(
        inner_train_X,
        inner_train_y
    ),
    batch_size=CLASSIFIER_BATCH_SIZE,
    shuffle=True,
    generator=train_generator
)

best_val_loss = float("inf")
best_state = None
best_epoch = 0
epochs_without_improvement = 0

training_history = []

for epoch in range(
    CLASSIFIER_MAX_EPOCHS
):

    classifier.train()

    running_loss = 0.0
    n_seen = 0

    for batch_X, batch_y in train_loader:

        batch_X = batch_X.to(
            device,
            non_blocking=True
        )

        batch_y = batch_y.to(
            device,
            non_blocking=True
        )

        optimizer.zero_grad(
            set_to_none=True
        )

        logits = classifier(
            batch_X
        )

        loss = criterion(
            logits,
            batch_y
        )

        assert torch.isfinite(loss)

        loss.backward()

        optimizer.step()

        running_loss += (
            loss.item()
            * batch_X.shape[0]
        )

        n_seen += (
            batch_X.shape[0]
        )

    train_loss = (
        running_loss / n_seen
    )

    # Inner validation ONLY.
    classifier.eval()

    with torch.no_grad():

        val_logits_list = []

        for start in range(
            0,
            len(inner_val_X),
            CLASSIFIER_BATCH_SIZE
        ):

            end = min(
                start + CLASSIFIER_BATCH_SIZE,
                len(inner_val_X)
            )

            batch_X = inner_val_X[
                start:end
            ].to(device)

            batch_y = inner_val_y[
                start:end
            ].to(device)

            logits = classifier(
                batch_X
            )

            val_logits_list.append(
                logits.detach().cpu()
            )

        val_logits = torch.cat(
            val_logits_list,
            dim=0
        )

        val_loss = criterion(
            val_logits,
            inner_val_y
        ).item()

    assert np.isfinite(train_loss)
    assert np.isfinite(val_loss)

    training_history.append({
        "epoch": epoch + 1,
        "train_loss": train_loss,
        "val_loss": val_loss
    })

    if val_loss < best_val_loss:

        best_val_loss = val_loss

        best_state = copy.deepcopy(
            classifier.state_dict()
        )

        best_epoch = epoch + 1

        epochs_without_improvement = 0

    else:

        epochs_without_improvement += 1

    if (
        epoch == 0
        or (epoch + 1) % 10 == 0
        or epochs_without_improvement == CLASSIFIER_PATIENCE
    ):

        print(
            f"Epoch {epoch + 1:3d} | "
            f"train_loss={train_loss:.6f} | "
            f"val_loss={val_loss:.6f} | "
            f"best_epoch={best_epoch}"
        )

    if (
        epochs_without_improvement
        >= CLASSIFIER_PATIENCE
    ):

        print(
            "Early stopping triggered."
        )

        break


assert best_state is not None
assert 1 <= best_epoch <= CLASSIFIER_MAX_EPOCHS

classifier.load_state_dict(
    best_state
)

print(
    "PASS: full representative classifier training completed"
)

print(
    "Best epoch:",
    best_epoch
)

print(
    "Best inner validation loss:",
    best_val_loss
)


# ======================================================================================================================
# FINAL HELD-OUT PN00 EVALUATION
# ======================================================================================================================

print("\n" + "-" * 120)
print("HELD-OUT TEST EVALUATION")
print("-" * 120)

X_test_tensor = torch.from_numpy(
    np.transpose(
        X_pn00,
        (0, 2, 1)
    )
).to(device)

test_feature_batches = []

classifier.eval()

with torch.no_grad():

    for start in range(
        0,
        len(X_test_tensor),
        CLASSIFIER_BATCH_SIZE
    ):

        end = min(
            start + CLASSIFIER_BATCH_SIZE,
            len(X_test_tensor)
        )

        batch = X_test_tensor[
            start:end
        ]

        features = extract_features(
            reve_model,
            batch,
            positions,
            remove_blocks=audit_removed_blocks
        )

        logits = classifier(
            features
        )

        probabilities = torch.softmax(
            logits,
            dim=1
        )[:, 1]

        test_feature_batches.append(
            probabilities.detach().cpu()
        )

test_probabilities = torch.cat(
    test_feature_batches,
    dim=0
).numpy()

assert len(
    test_probabilities
) == len(y_pn00)

assert np.isfinite(
    test_probabilities
).all()

test_metrics = safe_binary_metrics(
    y_pn00,
    test_probabilities
)

print(
    "Test samples:",
    len(y_pn00)
)

print(
    "Test class counts:",
    {
        int(k): int(v)
        for k, v in zip(
            *np.unique(
                y_pn00,
                return_counts=True
            )
        )
    }
)

print(
    "Test metrics:"
)

for metric_name, metric_value in test_metrics.items():

    print(
        f"  {metric_name}: {metric_value}"
    )

assert np.isfinite(
    test_metrics["accuracy"]
)

assert np.isfinite(
    test_metrics["balanced_accuracy"]
)

assert np.isfinite(
    test_metrics["f1"]
)

# PN00 is known to contain both classes, therefore these should be defined.
assert np.isfinite(
    test_metrics["auroc"]
)

assert np.isfinite(
    test_metrics["auprc"]
)

print(
    "PASS: held-out test metrics computed successfully"
)


# ======================================================================================================================
# VERIFY TEST SUBJECT NEVER ENTERED TRAINING DATA
# ======================================================================================================================

assert AUDIT_TEST_SUBJECT not in (
    training_subjects
)

assert AUDIT_TEST_SUBJECT not in (
    validation_subjects
)

# Explicitly verify feature sample counts are composed only from the
# non-test subjects.
expected_train_samples = sum(
    len(
        np.load(
            source_files[s]
        )["y"]
    )
    for s in training_subjects
)

expected_val_samples = sum(
    len(
        np.load(
            source_files[s]
        )["y"]
    )
    for s in validation_subjects
)

assert len(inner_train_X) == (
    expected_train_samples
)

assert len(inner_val_X) == (
    expected_val_samples
)

assert len(test_probabilities) == (
    len(y_pn00)
)

print(
    "PASS: train/validation/test sample accounting is consistent"
)


# ======================================================================================================================
# FINAL REVE FREEZE CHECK
# ======================================================================================================================

reve_trainable = [
    name
    for name, p in reve_model.named_parameters()
    if p.requires_grad
]

assert len(reve_trainable) == 0

reve_gradient_found = False

for name, p in reve_model.named_parameters():

    if p.grad is not None:

        if torch.any(
            p.grad != 0
        ):

            reve_gradient_found = True

            print(
                "Unexpected REVE gradient:",
                name
            )

            break

assert not reve_gradient_found

print(
    "PASS: REVE remained frozen throughout representative classifier training"
)


# ======================================================================================================================
# FINAL MEMORY CLEANUP
# ======================================================================================================================

del classifier
del optimizer
del criterion
del train_loader

del inner_train_X
del inner_train_y
del inner_val_X
del inner_val_y

del X_test_tensor
del test_feature_batches
del test_probabilities

del features_0
del features_b8
del features_ours

del X_test_small
del X_pn00
del y_pn00
del data_pn00

gc.collect()
torch.cuda.empty_cache()


# ======================================================================================================================
# FINAL SUMMARY
# ======================================================================================================================

print("\n" + "=" * 120)
print("CELL 13D — FINAL AUDIT COMPLETE")
print("=" * 120)

print("PASS  1. Method-specific pruning plans")
print("PASS  2. Random ranking reproducibility")
print("PASS  3. 2700 manifest <-> rankings")
print("PASS  4. Inner validation construction")
print("PASS  5. Class balance in inner split")
print("PASS  6. Locked classifier hyperparameters")
print("PASS  7. Metric edge cases")
print("PASS  8. Checkpoint integrity / resume")
print("PASS  9. Feature-cache isolation")
print("PASS 10. REVE mutation / restoration")
print("PASS 11. GPU memory behavior")
print("PASS 12. Complete representative LOSO job")

print("\nRepresentative configuration:")
print(
    f"  Test subject      = {AUDIT_TEST_SUBJECT}"
)
print(
    f"  Method            = {AUDIT_METHOD}"
)
print(
    f"  Pruning           = {AUDIT_PRUNING_PERCENT}%"
)
print(
    f"  Removed blocks    = {audit_removed_blocks}"
)
print(
    f"  Classifier seed   = {AUDIT_CLASSIFIER_SEED}"
)
print(
    f"  Best epoch        = {best_epoch}"
)

print("\nNO 2700-JOB EXPERIMENT WAS STARTED.")
print("NO EXPERIMENT RESULTS WERE OVERWRITTEN.")
print("TEMPORARY AUDIT FILES WERE CREATED ONLY UNDER:")
print(AUDIT_TMP)

print("=" * 120)
```

```python
# ======================================================================================================================
# CELL 13D — CONTINUATION AUDIT
# ======================================================================================================================
# PURPOSE:
#   Check the remaining methodological/implementation requirements.
#   Failed checks are recorded and the audit CONTINUES.
#
#   NO REVE loading
#   NO classifier training
#   NO experiment execution
#   NO files modified
# ======================================================================================================================

import os
import json
import hashlib
import numpy as np
import pandas as pd
import torch
import torch.nn as nn

print("\n" + "=" * 120)
print("CELL 13D — CONTINUATION AUDIT")
print("=" * 120)

# ======================================================================================================================
# CONFIGURATION
# ======================================================================================================================

SUBJECTS = [
    "PN00", "PN01", "PN06", "PN07",
    "PN09", "PN10", "PN11", "PN12",
    "PN13", "PN14", "PN16", "PN17"
]

METHODS = [
    "Random",
    "Functional",
    "DDF",
    "SSN",
    "Ours"
]

LEVELS = [10, 20, 30, 40, 50, 60, 70, 80, 90]

REMOVE_COUNTS = {
    10: 2,
    20: 4,
    30: 7,
    40: 9,
    50: 11,
    60: 13,
    70: 15,
    80: 18,
    90: 20
}

CLASSIFIER_SEEDS = [20, 25, 30, 35, 40]

BATCH_SIZE = 32
MAX_EPOCHS = 100
LEARNING_RATE = 1e-3
WEIGHT_DECAY = 1e-4
PATIENCE = 15
DROPOUT = 0.2

BASE_DIR = "/content/drive/MyDrive/Seinna_REVE_LOSO"
NOVELTY3_DIR = os.path.join(BASE_DIR, "novelty3_pruning")
SOURCE_DIR = os.path.join(BASE_DIR, "subject_epochs_22ch")

MANIFEST_PATH = os.path.join(
    NOVELTY3_DIR,
    "pruning_2700_job_manifest.csv"
)

DDF_PATH = os.path.join(
    NOVELTY3_DIR,
    "rankings",
    "ddf_block_ranking.csv"
)

SSN_PATH = os.path.join(
    NOVELTY3_DIR,
    "rankings",
    "ssn_block_ranking.csv"
)

FUNCTIONAL_PATH = os.path.join(
    NOVELTY3_DIR,
    "rankings",
    "functional_block_ranking.csv"
)

OURS_SCORES_PATH = os.path.join(
    NOVELTY3_DIR,
    "gate_training",
    "ours_block_scores.csv"
)

OURS_PLANS_PATH = os.path.join(
    NOVELTY3_DIR,
    "pruning_plans.csv"
)

GATE_MATRIX_PATH = os.path.join(
    NOVELTY3_DIR,
    "gate_training",
    "gate_matrix_60x22.npy"
)

GATE_STATS_PATH = os.path.join(
    NOVELTY3_DIR,
    "gate_training",
    "gate_aggregate_statistics.csv"
)

# ======================================================================================================================
# FAILURE REGISTRY
# ======================================================================================================================

FAILURES = []

def fail(check, message):
    FAILURES.append((check, str(message)))
    print(f"FAIL [{check}] {message}")

def passed(check, message):
    print(f"PASS [{check}] {message}")

# ======================================================================================================================
# HELPER
# ======================================================================================================================

def parse_blocks(value):
    if pd.isna(value):
        return []

    text = str(value).strip()

    if not text:
        return []

    return [
        int(
            item.strip()
            .replace("B", "")
            .replace("b", "")
        )
        for item in text.split(",")
        if item.strip()
    ]

# ======================================================================================================================
# 1. LOAD MANIFEST
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 1 — MANIFEST")
print("-" * 120)

manifest = None

try:
    assert os.path.isfile(MANIFEST_PATH), (
        f"Missing manifest: {MANIFEST_PATH}"
    )

    manifest = pd.read_csv(MANIFEST_PATH)

    assert len(manifest) == 2700, (
        f"Expected 2700 rows, got {len(manifest)}"
    )

    passed(
        "Manifest",
        "2700 rows loaded"
    )

except Exception as e:
    fail("Manifest", e)

# ======================================================================================================================
# 2. LOAD METHOD RANKINGS
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 2 — METHOD RANKINGS")
print("-" * 120)

rankings = {}

ranking_specs = {
    "DDF": (DDF_PATH, "ddf_removability"),
    "SSN": (SSN_PATH, "ssn_removability"),
    "Functional": (FUNCTIONAL_PATH, "functional_score"),
    "Ours": (OURS_SCORES_PATH, "ours_score")
}

for method, (path, score_column) in ranking_specs.items():

    try:
        df = pd.read_csv(path)

        assert len(df) == 22, (
            f"{method}: expected 22 rows, got {len(df)}"
        )

        assert "block" in df.columns

        assert score_column in df.columns, (
            f"{method}: missing {score_column}"
        )

        assert np.isfinite(
            df[score_column].astype(float)
        ).all()

        order = (
            df.sort_values(
                score_column,
                ascending=False,
                kind="mergesort"
            )["block"]
            .astype(int)
            .tolist()
        )

        assert len(order) == 22
        assert len(set(order)) == 22
        assert set(order) == set(range(1, 23))

        rankings[method] = order

        passed(
            f"{method} ranking",
            str(order)
        )

    except Exception as e:
        fail(f"{method} ranking", e)

# ======================================================================================================================
# 3. RANDOM RANKING REPRODUCIBILITY
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 3 — RANDOM RANKING REPRODUCIBILITY")
print("-" * 120)

if manifest is not None:

    try:

        random_rows = manifest[
            manifest["method"].astype(str) == "Random"
        ].copy()

        assert len(random_rows) == 540

        random_orders = {}

        for seed in CLASSIFIER_SEEDS:

            rows = random_rows[
                random_rows[
                    "random_ranking_seed"
                ].astype(int) == seed
            ].copy()

            assert len(rows) == 108, (
                f"seed {seed}: expected 108 rows, "
                f"got {len(rows)}"
            )

            seed_order = None

            for level in LEVELS:

                level_rows = rows[
                    rows[
                        "pruning_percent"
                    ].astype(int) == level
                ]

                assert len(level_rows) == 12

                unique_strings = (
                    level_rows[
                        "removed_blocks"
                    ]
                    .astype(str)
                    .unique()
                    .tolist()
                )

                assert len(unique_strings) == 1, (
                    f"seed {seed}, level {level}% has "
                    f"{len(unique_strings)} configurations"
                )

                current = parse_blocks(
                    unique_strings[0]
                )

                if seed_order is None:
                    seed_order = current
                else:
                    assert current == (
                        seed_order[:len(current)]
                    ), (
                        f"seed {seed}: non-cumulative "
                        f"ranking at {level}%"
                    )

            random_orders[seed] = seed_order

            print(
                f"Random seed {seed}: "
                f"{seed_order}"
            )

        unique_orders = {
            tuple(order)
            for order in random_orders.values()
        }

        if len(unique_orders) == 1:

            passed(
                "Random reproducibility",
                "identical ranking across classifier seeds"
            )

        else:

            fail(
                "Random reproducibility",
                (
                    f"{len(unique_orders)} different "
                    f"Random rankings are present"
                )
            )

    except Exception as e:
        fail("Random reproducibility", e)

# ======================================================================================================================
# 4. METHOD-SPECIFIC MANIFEST <-> RANKING
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 4 — METHOD-SPECIFIC MANIFEST <-> RANKING")
print("-" * 120)

if manifest is not None:

    mismatch_count = 0

    for idx, row in manifest.iterrows():

        method = str(row["method"])
        level = int(row["pruning_percent"])

        try:

            remove_count = REMOVE_COUNTS[level]

            if method == "Random":

                seed = int(
                    row["random_ranking_seed"]
                )

                if seed not in random_orders:
                    mismatch_count += 1
                    continue

                expected_order = (
                    random_orders[seed]
                )

            else:

                if method not in rankings:
                    mismatch_count += 1
                    continue

                expected_order = rankings[
                    method
                ]

            expected_removed = (
                expected_order[
                    :remove_count
                ]
            )

            actual_removed = parse_blocks(
                row["removed_blocks"]
            )

            expected_kept = [
                block
                for block in range(1, 23)
                if block not in expected_removed
            ]

            actual_kept = parse_blocks(
                row["kept_blocks"]
            )

            if actual_removed != expected_removed:
                mismatch_count += 1

                if mismatch_count <= 10:
                    print(
                        f"Mismatch row {idx}: "
                        f"{method} {level}%"
                    )
                    print(
                        "  Actual:",
                        actual_removed
                    )
                    print(
                        "  Expected:",
                        expected_removed
                    )

            if actual_kept != expected_kept:
                mismatch_count += 1

                if mismatch_count <= 10:
                    print(
                        f"Kept mismatch row {idx}: "
                        f"{method} {level}%"
                    )

        except Exception:
            mismatch_count += 1

    if mismatch_count == 0:

        passed(
            "Method-specific manifest",
            "all 2700 rows match their method ranking"
        )

    else:

        fail(
            "Method-specific manifest",
            f"{mismatch_count} mismatches"
        )

# ======================================================================================================================
# 5. INNER VALIDATION CLASS BALANCE
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 5 — INNER VALIDATION CLASS BALANCE")
print("-" * 120)

source_files = {}

for subject in SUBJECTS:

    path = os.path.join(
        SOURCE_DIR,
        f"{subject}_epochs.npz"
    )

    if os.path.isfile(path):
        source_files[subject] = path
    else:
        fail(
            "Source EEG",
            f"missing {subject}"
        )

if len(source_files) == 12:

    passed(
        "Source EEG",
        "12/12 source files available"
    )

# IMPORTANT:
# This check tests a deterministic subject-level inner split.
# It does NOT alter the final engine.

INNER_SPLIT_SEED = 20260918

inner_split_failures = 0

for test_subject in SUBJECTS:

    remaining = [
        s
        for s in SUBJECTS
        if s != test_subject
    ]

    rng = np.random.default_rng(
        INNER_SPLIT_SEED
    )

    shuffled = remaining.copy()
    rng.shuffle(shuffled)

    # Find the smallest subject-level validation set
    # that contains both classes while the training set
    # also contains both classes.

    found = False

    for n_val in range(
        1,
        len(shuffled)
    ):

        val_subjects = shuffled[:n_val]

        train_subjects = [
            s
            for s in remaining
            if s not in val_subjects
        ]

        train_labels = []
        val_labels = []

        for subject in train_subjects:

            y = np.load(
                source_files[subject]
            )["y"]

            train_labels.extend(
                y.tolist()
            )

        for subject in val_subjects:

            y = np.load(
                source_files[subject]
            )["y"]

            val_labels.extend(
                y.tolist()
            )

        if (
            set(train_labels) == {0, 1}
            and
            set(val_labels) == {0, 1}
        ):

            found = True
            break

    if not found:

        inner_split_failures += 1

        print(
            f"FAIL: {test_subject} "
            f"no valid two-class inner split"
        )

    else:

        print(
            f"PASS: {test_subject} | "
            f"train subjects={len(train_subjects)} | "
            f"validation subjects={len(val_subjects)} | "
            f"train classes={sorted(set(train_labels))} | "
            f"val classes={sorted(set(val_labels))}"
        )

if inner_split_failures == 0:

    passed(
        "Inner validation class balance",
        "all 12 LOSO folds have a valid two-class subject-level split"
    )

else:

    fail(
        "Inner validation class balance",
        f"{inner_split_failures} LOSO folds failed"
    )

# ======================================================================================================================
# 6. CLASSIFIER HYPERPARAMETERS
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 6 — CLASSIFIER HYPERPARAMETERS")
print("-" * 120)

try:

    classifier = nn.Sequential(
        nn.Linear(1024, 512),
        nn.ReLU(),
        nn.Dropout(0.2),
        nn.Linear(512, 128),
        nn.ReLU(),
        nn.Dropout(0.2),
        nn.Linear(128, 2)
    )

    parameter_count = sum(
        p.numel()
        for p in classifier.parameters()
    )

    assert parameter_count == 590722

    passed(
        "Classifier architecture",
        "1024 -> 512 -> 128 -> 2 | 590722 parameters"
    )

    print(
        "Batch size:",
        BATCH_SIZE
    )

    print(
        "Max epochs:",
        MAX_EPOCHS
    )

    print(
        "Learning rate:",
        LEARNING_RATE
    )

    print(
        "Weight decay:",
        WEIGHT_DECAY
    )

    print(
        "Patience:",
        PATIENCE
    )

    print(
        "Dropout:",
        DROPOUT
    )

except Exception as e:
    fail("Classifier architecture", e)

# ======================================================================================================================
# 7. METRIC EDGE CASES
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 7 — METRIC EDGE CASES")
print("-" * 120)

try:

    from sklearn.metrics import (
        accuracy_score,
        balanced_accuracy_score,
        f1_score,
        roc_auc_score,
        average_precision_score
    )

    y_two = np.array(
        [0, 0, 1, 1]
    )

    p_two = np.array(
        [0.1, 0.2, 0.8, 0.9]
    )

    auroc = roc_auc_score(
        y_two,
        p_two
    )

    auprc = average_precision_score(
        y_two,
        p_two
    )

    assert np.isfinite(auroc)
    assert np.isfinite(auprc)

    y_one = np.array(
        [0, 0, 0, 0]
    )

    p_one = np.array(
        [0.1, 0.2, 0.3, 0.4]
    )

    if len(np.unique(y_one)) < 2:

        auroc_one = np.nan
        auprc_one = np.nan

    else:

        auroc_one = roc_auc_score(
            y_one,
            p_one
        )

        auprc_one = average_precision_score(
            y_one,
            p_one
        )

    assert np.isnan(auroc_one)
    assert np.isnan(auprc_one)

    passed(
        "Metric edge cases",
        "single-class AUROC/AUPRC handled explicitly as NaN"
    )

except Exception as e:
    fail("Metric edge cases", e)

# ======================================================================================================================
# 8. CHECKPOINT / RESUME MANIFEST UNIQUENESS
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 8 — CHECKPOINT / RESUME JOB UNIQUENESS")
print("-" * 120)

if manifest is not None:

    try:

        key_columns = [
            "method",
            "pruning_percent",
            "subject",
            "classifier_seed"
        ]

        group_sizes = (
            manifest
            .groupby(key_columns)
            .size()
        )

        assert len(group_sizes) == 2700

        assert (
            group_sizes == 1
        ).all()

        job_ids = (
            manifest["job_id"]
            .astype(int)
        )

        assert len(job_ids) == 2700
        assert len(set(job_ids)) == 2700

        passed(
            "Checkpoint/resume manifest",
            "2700 unique jobs with no duplicate method-level-subject-seed keys"
        )

    except Exception as e:
        fail("Checkpoint/resume manifest", e)

# ======================================================================================================================
# 9. FEATURE CACHE KEY ISOLATION
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 9 — FEATURE CACHE ISOLATION")
print("-" * 120)

try:

    keys = {}

    for subject in SUBJECTS:

        for method in METHODS:

            for level in LEVELS:

                raw = (
                    f"{subject}|"
                    f"{method}|"
                    f"{level}"
                )

                digest = hashlib.sha256(
                    raw.encode("utf-8")
                ).hexdigest()[:16]

                key = (
                    f"{subject}__"
                    f"{method}__"
                    f"{level:02d}__"
                    f"{digest}"
                )

                keys[key] = (
                    subject,
                    method,
                    level
                )

    expected = (
        12 * 5 * 9
    )

    assert len(keys) == expected

    passed(
        "Feature cache isolation",
        f"{expected} unique subject × method × pruning-level keys"
    )

except Exception as e:
    fail("Feature cache isolation", e)

# ======================================================================================================================
# 10. GPU MEMORY ENVIRONMENT
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 10 — GPU MEMORY ENVIRONMENT")
print("-" * 120)

try:

    if torch.cuda.is_available():

        device = torch.cuda.current_device()

        total_gb = (
            torch.cuda.get_device_properties(
                device
            ).total_memory
            / (1024 ** 3)
        )

        allocated_gb = (
            torch.cuda.memory_allocated(device)
            / (1024 ** 3)
        )

        reserved_gb = (
            torch.cuda.memory_reserved(device)
            / (1024 ** 3)
        )

        print(
            "GPU:",
            torch.cuda.get_device_name(device)
        )

        print(
            f"Total GPU memory: {total_gb:.2f} GB"
        )

        print(
            f"Currently allocated: {allocated_gb:.2f} GB"
        )

        print(
            f"Currently reserved: {reserved_gb:.2f} GB"
        )

        passed(
            "GPU environment",
            "CUDA available"
        )

    else:

        fail(
            "GPU environment",
            "CUDA is not available"
        )

except Exception as e:
    fail("GPU environment", e)

# ======================================================================================================================
# 11. EXISTING EVALUATION FILES
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 11 — EXISTING EVALUATION OUTPUTS")
print("-" * 120)

EVAL_DIR = os.path.join(
    NOVELTY3_DIR,
    "evaluation_results"
)

if os.path.isdir(EVAL_DIR):

    files_found = []

    for root, dirs, filenames in os.walk(
        EVAL_DIR
    ):

        for filename in filenames:

            files_found.append(
                os.path.join(
                    root,
                    filename
                )
            )

    print(
        "Evaluation directory exists."
    )

    print(
        "Existing files:",
        len(files_found)
    )

    for path in files_found[:30]:
        print(
            " ",
            path
        )

else:

    print(
        "Evaluation directory does not exist yet."
    )

# ======================================================================================================================
# 12. FINAL REPORT
# ======================================================================================================================

print("\n" + "=" * 120)
print("CELL 13D — FINAL AUDIT REPORT")
print("=" * 120)

if len(FAILURES) == 0:

    print(
        "PASS: No failures detected in this continuation audit."
    )

else:

    print(
        f"TOTAL FAILURES: {len(FAILURES)}"
    )

    print()

    for number, (check, message) in enumerate(
        FAILURES,
        start=1
    ):

        print(
            f"{number}. [{check}]"
        )

        print(
            f"   {message}"
        )

print("\nNO REVE EXPERIMENT WAS RUN.")
print("NO CLASSIFIER TRAINING WAS RUN.")
print("NO FILES WERE MODIFIED.")
print("NO 2700-JOB EVALUATION WAS STARTED.")

print("=" * 120)
```

```python
# ======================================================================================================================
# CELL 13E — REPAIR METHOD RANKINGS + REBUILD 2700-JOB MANIFEST
# ======================================================================================================================
# PURPOSE:
#   1. Detect the actual DDF/SSN ranking score columns.
#   2. Preserve the verified Functional and Ours rankings.
#   3. Create ONE deterministic Random ranking.
#   4. Rebuild all 2700 manifest rows using method-specific pruning.
#   5. Verify every method × level × subject × classifier-seed combination.
#
# NO REVE
# NO CLASSIFIER
# NO TRAINING
# NO FORWARD PASS
#
# ======================================================================================================================

import os
import numpy as np
import pandas as pd

print("\n" + "=" * 120)
print("CELL 13E — REPAIR + REBUILD 2700-JOB MANIFEST")
print("=" * 120)

# ======================================================================================================================
# CONSTANTS
# ======================================================================================================================

SUBJECTS = [
    "PN00", "PN01", "PN06", "PN07",
    "PN09", "PN10", "PN11", "PN12",
    "PN13", "PN14", "PN16", "PN17"
]

METHODS = [
    "Random",
    "Functional",
    "DDF",
    "SSN",
    "Ours"
]

LEVELS = [10, 20, 30, 40, 50, 60, 70, 80, 90]

REMOVE_COUNTS = {
    10: 2,
    20: 4,
    30: 7,
    40: 9,
    50: 11,
    60: 13,
    70: 15,
    80: 18,
    90: 20
}

CLASSIFIER_SEEDS = [20, 25, 30, 35, 40]

# Use the already-existing seed-20 Random ranking as the canonical
# deterministic Random ranking.
RANDOM_RANKING_SEED = 20

BASE_DIR = "/content/drive/MyDrive/Seinna_REVE_LOSO"
NOVELTY3_DIR = os.path.join(
    BASE_DIR,
    "novelty3_pruning"
)

RANKING_DIR = os.path.join(
    NOVELTY3_DIR,
    "rankings"
)

DDF_PATH = os.path.join(
    RANKING_DIR,
    "ddf_block_ranking.csv"
)

SSN_PATH = os.path.join(
    RANKING_DIR,
    "ssn_block_ranking.csv"
)

FUNCTIONAL_PATH = os.path.join(
    RANKING_DIR,
    "functional_block_ranking.csv"
)

OURS_SCORES_PATH = os.path.join(
    NOVELTY3_DIR,
    "gate_training",
    "ours_block_scores.csv"
)

MANIFEST_PATH = os.path.join(
    NOVELTY3_DIR,
    "pruning_2700_job_manifest.csv"
)

# ======================================================================================================================
# HELPER
# ======================================================================================================================

def get_ranking_column(df, method):
    """
    Find the actual ranking/removability column without assuming
    the filename's exact schema.
    """

    candidates = {
        "DDF": [
            "removability",
            "ddf_removability",
            "ddf_norm",
            "ddf"
        ],
        "SSN": [
            "removability",
            "ssn_removability",
            "ssn_norm",
            "ssn"
        ]
    }

    for column in candidates[method]:

        if column in df.columns:

            values = pd.to_numeric(
                df[column],
                errors="coerce"
            )

            if values.notna().all():

                return column

    return None


def verify_permutation(order, method):
    """
    Verify an exact B1-B22 permutation.
    """

    assert len(order) == 22, (
        f"{method}: expected 22 blocks, got {len(order)}"
    )

    assert len(set(order)) == 22, (
        f"{method}: duplicate block detected"
    )

    assert set(order) == set(range(1, 23)), (
        f"{method}: blocks are not exactly B1-B22"
    )


# ======================================================================================================================
# CHECK DDF
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 1 — DDF RANKING")
print("-" * 120)

ranking_orders = {}

try:

    ddf = pd.read_csv(DDF_PATH)

    print(
        "DDF columns:",
        ddf.columns.tolist()
    )

    ddf_column = get_ranking_column(
        ddf,
        "DDF"
    )

    assert ddf_column is not None, (
        "Could not identify numeric DDF ranking column"
    )

    print(
        "DDF ranking column:",
        ddf_column
    )

    ddf_order = (
        ddf
        .sort_values(
            ddf_column,
            ascending=False,
            kind="mergesort"
        )["block"]
        .astype(int)
        .tolist()
    )

    verify_permutation(
        ddf_order,
        "DDF"
    )

    ranking_orders["DDF"] = ddf_order

    print(
        "DDF order:",
        ddf_order
    )

    print(
        "PASS: DDF ranking verified"
    )

except Exception as e:

    print(
        "FAIL: DDF ranking:",
        e
    )

# ======================================================================================================================
# CHECK SSN
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 2 — SSN RANKING")
print("-" * 120)

try:

    ssn = pd.read_csv(SSN_PATH)

    print(
        "SSN columns:",
        ssn.columns.tolist()
    )

    ssn_column = get_ranking_column(
        ssn,
        "SSN"
    )

    assert ssn_column is not None, (
        "Could not identify numeric SSN ranking column"
    )

    print(
        "SSN ranking column:",
        ssn_column
    )

    ssn_order = (
        ssn
        .sort_values(
            ssn_column,
            ascending=False,
            kind="mergesort"
        )["block"]
        .astype(int)
        .tolist()
    )

    verify_permutation(
        ssn_order,
        "SSN"
    )

    ranking_orders["SSN"] = ssn_order

    print(
        "SSN order:",
        ssn_order
    )

    print(
        "PASS: SSN ranking verified"
    )

except Exception as e:

    print(
        "FAIL: SSN ranking:",
        e
    )

# ======================================================================================================================
# CHECK FUNCTIONAL
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 3 — FUNCTIONAL RANKING")
print("-" * 120)

try:

    functional = pd.read_csv(
        FUNCTIONAL_PATH
    )

    assert "block" in functional.columns

    assert "functional_score" in functional.columns

    functional_order = (
        functional
        .sort_values(
            "functional_score",
            ascending=False,
            kind="mergesort"
        )["block"]
        .astype(int)
        .tolist()
    )

    verify_permutation(
        functional_order,
        "Functional"
    )

    ranking_orders["Functional"] = (
        functional_order
    )

    print(
        "Functional order:",
        functional_order
    )

    print(
        "PASS: Functional ranking verified"
    )

except Exception as e:

    print(
        "FAIL: Functional ranking:",
        e
    )

# ======================================================================================================================
# CHECK OURS
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 4 — OURS RANKING")
print("-" * 120)

try:

    ours = pd.read_csv(
        OURS_SCORES_PATH
    )

    assert "block" in ours.columns

    assert "ours_score" in ours.columns

    ours_order = (
        ours
        .sort_values(
            "ours_score",
            ascending=False,
            kind="mergesort"
        )["block"]
        .astype(int)
        .tolist()
    )

    verify_permutation(
        ours_order,
        "Ours"
    )

    ranking_orders["Ours"] = ours_order

    print(
        "Ours order:",
        ours_order
    )

    print(
        "PASS: Ours ranking verified"
    )

except Exception as e:

    print(
        "FAIL: Ours ranking:",
        e
    )

# ======================================================================================================================
# CREATE ONE FIXED RANDOM RANKING
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 5 — FIXED RANDOM RANKING")
print("-" * 120)

random_rng = np.random.default_rng(
    RANDOM_RANKING_SEED
)

random_order = random_rng.permutation(
    np.arange(1, 23)
).tolist()

verify_permutation(
    random_order,
    "Random"
)

ranking_orders["Random"] = random_order

print(
    "Random ranking seed:",
    RANDOM_RANKING_SEED
)

print(
    "Random order:",
    random_order
)

expected_existing_seed20 = [
    3, 21, 14, 2, 16, 5, 1, 11,
    10, 15, 8, 20, 22, 6, 13, 17,
    9, 4, 7, 19, 18, 12
]

if random_order == expected_existing_seed20:

    print(
        "PASS: fixed Random ranking exactly matches "
        "the existing seed-20 ranking"
    )

else:

    print(
        "WARNING: generated Random ranking differs "
        "from the previous seed-20 manifest"
    )

# ======================================================================================================================
# DISPLAY ALL FIVE RANKINGS
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 6 — ALL FIVE RANKINGS")
print("-" * 120)

for method in METHODS:

    if method in ranking_orders:

        print(
            f"{method:12s}:",
            ranking_orders[method]
        )

# ======================================================================================================================
# BUILD METHOD-SPECIFIC PRUNING CONFIGURATIONS
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 7 — METHOD-SPECIFIC PRUNING CONFIGURATIONS")
print("-" * 120)

method_level_plans = {}

for method in METHODS:

    assert method in ranking_orders, (
        f"Missing ranking for {method}"
    )

    order = ranking_orders[method]

    method_level_plans[method] = {}

    print(
        f"\n{method}"
    )

    for level in LEVELS:

        remove_count = REMOVE_COUNTS[level]

        removed = (
            order[:remove_count]
        )

        kept = [
            block
            for block in range(1, 23)
            if block not in removed
        ]

        assert len(removed) == remove_count

        assert len(kept) == (
            22 - remove_count
        )

        assert len(
            set(removed)
        ) == remove_count

        assert (
            set(removed) | set(kept)
        ) == set(range(1, 23))

        assert (
            set(removed) & set(kept)
        ) == set()

        method_level_plans[
            method
        ][level] = {
            "removed": removed,
            "kept": kept
        }

        print(
            f"{level:3d}% | remove={removed}"
        )

print(
    "\nPASS: all 45 method × pruning configurations verified"
)

# ======================================================================================================================
# REBUILD 2700-JOB MANIFEST
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 8 — REBUILD 2700-JOB MANIFEST")
print("-" * 120)

rows = []

job_id = 1

for method in METHODS:

    for level in LEVELS:

        plan = method_level_plans[
            method
        ][level]

        for subject in SUBJECTS:

            for classifier_seed in CLASSIFIER_SEEDS:

                if method == "Random":

                    random_seed_value = (
                        RANDOM_RANKING_SEED
                    )

                else:

                    random_seed_value = np.nan

                rows.append({
                    "job_id": job_id,
                    "method": method,
                    "pruning_percent": level,
                    "remove_count": REMOVE_COUNTS[level],
                    "keep_count": 22 - REMOVE_COUNTS[level],
                    "subject": subject,
                    "classifier_seed": classifier_seed,
                    "random_ranking_seed": random_seed_value,
                    "ranking_type": (
                        "fixed_random"
                        if method == "Random"
                        else "deterministic"
                    ),
                    "ranking_source": (
                        "seed20_fixed"
                        if method == "Random"
                        else f"{method}_ranking"
                    ),
                    "removed_blocks": ",".join(
                        f"B{b}"
                        for b in plan["removed"]
                    ),
                    "kept_blocks": ",".join(
                        f"B{b}"
                        for b in plan["kept"]
                    )
                })

                job_id += 1

manifest_new = pd.DataFrame(
    rows
)

# ======================================================================================================================
# MANIFEST STRUCTURE CHECK
# ======================================================================================================================

assert len(manifest_new) == 2700

assert (
    manifest_new["job_id"].nunique()
    == 2700
)

assert (
    manifest_new[
        [
            "method",
            "pruning_percent",
            "subject",
            "classifier_seed"
        ]
    ]
    .drop_duplicates()
    .shape[0]
    == 2700
)

print(
    "PASS: exactly 2700 unique "
    "method × level × subject × seed jobs"
)

# ======================================================================================================================
# VERIFY METHOD COUNTS
# ======================================================================================================================

method_counts = (
    manifest_new["method"]
    .value_counts()
    .reindex(METHODS)
)

print("\nMethod counts:")
print(method_counts)

assert (
    method_counts == 540
).all()

print(
    "PASS: each method has exactly 540 jobs"
)

# ======================================================================================================================
# VERIFY LEVEL COUNTS
# ======================================================================================================================

level_counts = (
    manifest_new
    .groupby(
        ["method", "pruning_percent"]
    )
    .size()
)

for method in METHODS:

    for level in LEVELS:

        assert (
            level_counts[
                (method, level)
            ]
            == 60
        )

print(
    "PASS: each method × pruning level "
    "has exactly 60 jobs"
)

# ======================================================================================================================
# VERIFY RANDOM IS IDENTICAL FOR EVERY CLASSIFIER SEED
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 9 — RANDOM REPRODUCIBILITY")
print("-" * 120)

random_manifest = manifest_new[
    manifest_new["method"] == "Random"
].copy()

for level in LEVELS:

    strings = (
        random_manifest[
            random_manifest[
                "pruning_percent"
            ] == level
        ]["removed_blocks"]
        .unique()
        .tolist()
    )

    assert len(strings) == 1, (
        f"Random {level}% has "
        f"{len(strings)} different configurations"
    )

    expected = ",".join(
        f"B{b}"
        for b in random_order[
            :REMOVE_COUNTS[level]
        ]
    )

    assert strings[0] == expected

print(
    "PASS: Random uses ONE fixed ranking "
    "for all classifier seeds and all subjects"
)

# ======================================================================================================================
# VERIFY ALL METHOD-SPECIFIC ROWS
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 10 — COMPLETE METHOD-SPECIFIC MANIFEST AUDIT")
print("-" * 120)

mismatches = []

for idx, row in manifest_new.iterrows():

    method = row["method"]

    level = int(
        row["pruning_percent"]
    )

    expected_removed = (
        method_level_plans[
            method
        ][level]["removed"]
    )

    expected_kept = (
        method_level_plans[
            method
        ][level]["kept"]
    )

    actual_removed = [
        int(
            x.strip().replace("B", "")
        )
        for x in str(
            row["removed_blocks"]
        ).split(",")
        if x.strip()
    ]

    actual_kept = [
        int(
            x.strip().replace("B", "")
        )
        for x in str(
            row["kept_blocks"]
        ).split(",")
        if x.strip()
    ]

    if actual_removed != expected_removed:

        mismatches.append(
            (
                idx,
                method,
                level,
                "removed",
                actual_removed,
                expected_removed
            )
        )

    if actual_kept != expected_kept:

        mismatches.append(
            (
                idx,
                method,
                level,
                "kept",
                actual_kept,
                expected_kept
            )
        )

if len(mismatches) == 0:

    print(
        "PASS: all 2700 manifest rows match "
        "their method-specific ranking"
    )

else:

    print(
        f"FAIL: {len(mismatches)} manifest mismatches"
    )

    for item in mismatches[:10]:

        print(item)

    raise AssertionError(
        "Method-specific manifest verification failed"
    )

# ======================================================================================================================
# VERIFY NESTED PRUNING
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 11 — NESTED PRUNING")
print("-" * 120)

for method in METHODS:

    previous = []

    for level in LEVELS:

        current = (
            method_level_plans[
                method
            ][level]["removed"]
        )

        if previous:

            assert current[:len(previous)] == previous

        previous = current

print(
    "PASS: pruning sets are cumulative "
    "within every method"
)

# ======================================================================================================================
# VERIFY RANDOM SEED FIELD
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 12 — RANDOM SEED FIELD")
print("-" * 120)

random_seeds = (
    random_manifest[
        "random_ranking_seed"
    ]
    .dropna()
    .astype(int)
    .unique()
    .tolist()
)

assert random_seeds == [
    RANDOM_RANKING_SEED
]

print(
    "PASS: Random ranking seed is fixed to",
    RANDOM_RANKING_SEED
)

# ======================================================================================================================
# SAVE ONLY AFTER ALL CHECKS PASS
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 13 — SAVE REPAIRED MANIFEST")
print("-" * 120)

# Backup the currently existing manifest before replacement.
backup_path = (
    MANIFEST_PATH.replace(
        ".csv",
        "_OLD_before_13E.csv"
    )
)

if os.path.isfile(MANIFEST_PATH):

    if not os.path.isfile(backup_path):

        old_manifest = pd.read_csv(
            MANIFEST_PATH
        )

        old_manifest.to_csv(
            backup_path,
            index=False
        )

        print(
            "Backup created:",
            backup_path
        )

    else:

        print(
            "Backup already exists:",
            backup_path
        )

manifest_new.to_csv(
    MANIFEST_PATH,
    index=False
)

print(
    "Saved repaired manifest:",
    MANIFEST_PATH
)

# ======================================================================================================================
# RELOAD AND FINAL VERIFY
# ======================================================================================================================

print("\n" + "-" * 120)
print("CHECK 14 — RELOAD FINAL MANIFEST")
print("-" * 120)

manifest_check = pd.read_csv(
    MANIFEST_PATH
)

assert len(manifest_check) == 2700

assert (
    manifest_check[
        [
            "method",
            "pruning_percent",
            "subject",
            "classifier_seed"
        ]
    ]
    .drop_duplicates()
    .shape[0]
    == 2700
)

assert (
    manifest_check["method"]
    .value_counts()
    .reindex(METHODS)
    == 540
).all()

print(
    "PASS: repaired manifest reloaded successfully"
)

print("\n" + "=" * 120)
print("CELL 13E — COMPLETE")
print("=" * 120)

print(
    "PASS: DDF ranking"
)

print(
    "PASS: SSN ranking"
)

print(
    "PASS: Functional ranking"
)

print(
    "PASS: Ours ranking"
)

print(
    "PASS: ONE fixed Random ranking"
)

print(
    "PASS: 45 method × pruning configurations"
)

print(
    "PASS: 2700 unique jobs"
)

print(
    "PASS: method-specific pruning"
)

print(
    "PASS: Random reproducibility"
)

print(
    "PASS: nested pruning"
)

print(
    "PASS: manifest reload"
)

print()
print(
    "NO REVE WAS LOADED."
)

print(
    "NO CLASSIFIER WAS TRAINED."
)

print(
    "NO EVALUATION WAS RUN."
)

print("=" * 120)
```

```python
# ==============================================================================================================
# CELL 13F-REPAIR — CORRECT RANDOM RANKING GENERATOR
# ==============================================================================

import numpy as np

RANDOM_RANKING_SEED = 20

def generate_fixed_random_ranking(seed=20):
    rng = np.random.default_rng(seed)
    blocks = np.arange(1, 23)
    rng.shuffle(blocks)
    return blocks.tolist()


expected_random_order = [
    3, 21, 14, 2, 16, 5, 1, 11, 10, 15, 8,
    20, 22, 6, 13, 17, 9, 4, 7, 19, 18, 12
]

random_order = generate_fixed_random_ranking(
    RANDOM_RANKING_SEED
)

assert random_order == expected_random_order, (
    f"Random ranking mismatch.\n"
    f"Generated: {random_order}\n"
    f"Expected:  {expected_random_order}"
)

print("=" * 110)
print("CELL 13F-REPAIR — RANDOM RANKING")
print("=" * 110)
print("PASS: correct RNG = np.random.default_rng(20)")
print("PASS: Random ranking exactly matches Cell 13E")
print()
print("Random ranking:")
print(random_order)
print("=" * 110)
```

```python
# ==============================================================================================================
# CELL 13F-DIAGNOSTIC — INSPECT ACTUAL OURS RANKING FILE
# ==============================================================================================================

import os
import pandas as pd
import numpy as np

RANKING_PATH = (
    "/content/drive/MyDrive/"
    "Seinna_REVE_LOSO/"
    "novelty3_pruning/"
    "rankings/"
    "ours_block_ranking.csv"
)

print("=" * 110)
print("CELL 13F-DIAGNOSTIC — ACTUAL OURS RANKING")
print("=" * 110)

assert os.path.exists(RANKING_PATH), (
    f"Missing file:\n{RANKING_PATH}"
)

df = pd.read_csv(RANKING_PATH)

print()
print("File:")
print(RANKING_PATH)

print()
print("Shape:")
print(df.shape)

print()
print("Columns:")
print(df.columns.tolist())

print()
print("First 22 rows:")
print(df.to_string(index=False))

print()
print("-" * 110)
print("BLOCK COLUMN CHECK")
print("-" * 110)

assert "block" in df.columns

blocks = df["block"].astype(int).tolist()

print("Blocks:")
print(blocks)

print()
print("Sorted blocks:")
print(sorted(blocks))

assert sorted(blocks) == list(range(1, 23))

print("PASS: B1-B22 complete")


# ==============================================================================================================
# FIND POSSIBLE RANK COLUMNS
# ==============================================================================================================

print()
print("-" * 110)
print("POSSIBLE RANKING COLUMNS")
print("-" * 110)

candidate_columns = [
    "ours_rank",
    "rank",
    "ours_removability",
    "ours_importance",
    "ours_score",
]

for column in candidate_columns:

    if column in df.columns:

        print()
        print(f"FOUND: {column}")

        values = df[column]

        print(
            "dtype:",
            values.dtype
        )

        print(
            "min:",
            values.min()
        )

        print(
            "max:",
            values.max()
        )

        print(
            "unique:",
            values.nunique()
        )

        if pd.api.types.is_numeric_dtype(values):

            print()
            print(
                "Ascending block order:"
            )

            asc = (
                df.sort_values(
                    column,
                    ascending=True
                )["block"]
                .astype(int)
                .tolist()
            )

            print(asc)

            print()
            print(
                "Descending block order:"
            )

            desc = (
                df.sort_values(
                    column,
                    ascending=False
                )["block"]
                .astype(int)
                .tolist()
            )

            print(desc)


# ==============================================================================================================
# CURRENT VERIFIED Ours ORDER FROM Ours SCORE FILE
# ==============================================================================================================

print()
print("-" * 110)
print("REFERENCE OURS ORDER")
print("-" * 110)

SCORE_PATH = (
    "/content/drive/MyDrive/"
    "Seinna_REVE_LOSO/"
    "novelty3_pruning/"
    "gate_training/"
    "ours_block_scores.csv"
)

score_df = pd.read_csv(
    SCORE_PATH
)

assert "ours_rank" in score_df.columns
assert "block" in score_df.columns

reference_order = (
    score_df
    .sort_values(
        "ours_rank",
        ascending=True
    )["block"]
    .astype(int)
    .tolist()
)

print(
    "Ours score reference order:"
)

print(
    reference_order
)


# ==============================================================================================================
# CHECK WHETHER THE RANKING FILE MATCHES THE REFERENCE
# ==============================================================================================================

print()
print("-" * 110)
print("REFERENCE COMPARISON")
print("-" * 110)

matches_found = False

for column in candidate_columns:

    if column not in df.columns:
        continue

    if not pd.api.types.is_numeric_dtype(
        df[column]
    ):
        continue

    ascending_order = (
        df.sort_values(
            column,
            ascending=True
        )["block"]
        .astype(int)
        .tolist()
    )

    descending_order = (
        df.sort_values(
            column,
            ascending=False
        )["block"]
        .astype(int)
        .tolist()
    )

    if ascending_order == reference_order:

        print(
            f"PASS: {column} ascending matches Ours reference"
        )

        matches_found = True

    if descending_order == reference_order:

        print(
            f"PASS: {column} descending matches Ours reference"
        )

        matches_found = True


print()

if matches_found:

    print(
        "PASS: Ours ranking file contains a column that reproduces the verified Ours order."
    )

else:

    print(
        "WARNING: No ranking column reproduces the verified Ours order."
    )

    print(
        "DO NOT modify the ranking file."
    )

    print(
        "The diagnostic output above will identify the discrepancy."
    )

print()
print("=" * 110)
print("DIAGNOSTIC COMPLETE — NO FILES MODIFIED")
print("=" * 110)
```

```python
# ==============================================================================================================
# CELL 13G — SINGLE REPRESENTATIVE END-TO-END EVALUATION
# ==============================================================================================================
#
# ONE JOB ONLY.
#
# Pipeline:
#   EEG
#    ↓
#   Official frozen REVE
#    ↓
#   Structured block pruning
#    ↓
#   (B,22,5,512)
#    ↓
#   patch mean
#    ↓
#   channel mean + channel std
#    ↓
#   (B,1024)
#    ↓
#   inner train / validation
#    ↓
#   classifier
#    ↓
#   held-out LOSO test
#    ↓
#   metrics + checkpoint
#
# IMPORTANT:
#   - Outer test subject is NEVER used for early stopping.
#   - REVE remains frozen.
#   - Only one job is executed.
# ==============================================================================================================

import os
import json
import random
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader
from transformers import AutoModel


print("=" * 110)
print("CELL 13G — SINGLE REPRESENTATIVE END-TO-END EVALUATION")
print("=" * 110)


# ==============================================================================================================
# 1. CONFIGURATION
# ==============================================================================================================

BASE_DIR = "/content/drive/MyDrive/Seinna_REVE_LOSO"

NOVELTY3_DIR = os.path.join(
    BASE_DIR,
    "novelty3_pruning"
)

MANIFEST_PATH = os.path.join(
    NOVELTY3_DIR,
    "pruning_2700_job_manifest.csv"
)

SOURCE_DIR = os.path.join(
    BASE_DIR,
    "subject_epochs_22ch"
)

REVE_STATE_PATH = (
    "/content/drive/MyDrive/"
    "Seinna_REVE_Novelty3_NEW/"
    "EXPERIMENT_ARCHIVE/"
    "REVE_MODEL/"
    "reve_base_state_dict.pt"
)

POSITION_STATE_PATH = (
    "/content/drive/MyDrive/"
    "Seinna_REVE_Novelty3_NEW/"
    "EXPERIMENT_ARCHIVE/"
    "REVE_POSITION_MODEL/"
    "reve_positions_state_dict.pt"
)

CHECKPOINT_DIR = os.path.join(
    NOVELTY3_DIR,
    "evaluation_results",
    "job_checkpoints"
)

METRIC_DIR = os.path.join(
    NOVELTY3_DIR,
    "evaluation_results",
    "metrics"
)

os.makedirs(
    CHECKPOINT_DIR,
    exist_ok=True
)

os.makedirs(
    METRIC_DIR,
    exist_ok=True
)

DEVICE = torch.device(
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

CLASSIFIER_BATCH_SIZE = 32
MAX_EPOCHS = 100
LEARNING_RATE = 1e-3
WEIGHT_DECAY = 1e-4
PATIENCE = 15
DROPOUT = 0.2

# Use the first manifest job only.
manifest = pd.read_csv(
    MANIFEST_PATH
)

assert len(manifest) == 2700

job = manifest.iloc[0].copy()

JOB_ID = int(
    job["job_id"]
)

METHOD = str(
    job["method"]
)

PRUNING_PERCENT = int(
    job["pruning_percent"]
)

REMOVE_COUNT = int(
    job["remove_count"]
)

TEST_SUBJECT = str(
    job["subject"]
)

CLASSIFIER_SEED = int(
    job["classifier_seed"]
)

print()
print("Selected representative job:")
print(
    f"Job ID           : {JOB_ID}"
)
print(
    f"Method            : {METHOD}"
)
print(
    f"Pruning           : {PRUNING_PERCENT}%"
)
print(
    f"Blocks removed    : {REMOVE_COUNT}"
)
print(
    f"Outer test subject: {TEST_SUBJECT}"
)
print(
    f"Classifier seed   : {CLASSIFIER_SEED}"
)
print(
    f"Device            : {DEVICE}"
)


# ==============================================================================================================
# 2. REPRODUCIBILITY
# ==============================================================================================================

def set_seed(seed):

    random.seed(seed)

    np.random.seed(seed)

    torch.manual_seed(seed)

    if torch.cuda.is_available():

        torch.cuda.manual_seed_all(seed)


set_seed(
    CLASSIFIER_SEED
)

print()
print(
    "PASS: random seeds initialized"
)


# ==============================================================================================================
# 3. VERIFY SELECTED MANIFEST JOB
# ==============================================================================================================

assert METHOD in [
    "Random",
    "Functional",
    "DDF",
    "SSN",
    "Ours"
]

assert PRUNING_PERCENT in [
    10, 20, 30, 40, 50,
    60, 70, 80, 90
]

assert REMOVE_COUNT > 0

assert TEST_SUBJECT in [
    "PN00", "PN01", "PN06", "PN07",
    "PN09", "PN10", "PN11", "PN12",
    "PN13", "PN14", "PN16", "PN17"
]

assert CLASSIFIER_SEED in [
    20, 25, 30, 35, 40
]

print(
    "PASS: selected manifest job is valid"
)


# ==============================================================================================================
# 4. LOAD SOURCE EEG
# ==============================================================================================================

source_path = os.path.join(
    SOURCE_DIR,
    f"{TEST_SUBJECT}_epochs.npz"
)

assert os.path.exists(
    source_path
)

with np.load(
    source_path
) as data:

    X_test = np.asarray(
        data["X"],
        dtype=np.float32
    )

    y_test = np.asarray(
        data["y"],
        dtype=np.int64
    )

assert X_test.ndim == 3

assert X_test.shape[1:] == (
    1000,
    22
)

assert y_test.ndim == 1

assert len(X_test) == len(y_test)

assert np.isfinite(
    X_test
).all()

assert set(
    np.unique(y_test)
) == {0, 1}

print()
print(
    f"PASS: {TEST_SUBJECT} EEG loaded"
)

print(
    f"EEG shape: {X_test.shape}"
)

print(
    f"Labels: {y_test.shape}"
)


# ==============================================================================================================
# 5. LOAD OFFICIAL REVE
# ==============================================================================================================

print()
print("-" * 110)
print("LOADING OFFICIAL REVE")
print("-" * 110)

reve_model = AutoModel.from_pretrained(
    "brain-bzh/reve-base",
    trust_remote_code=True,
    torch_dtype=torch.float32
)

reve_checkpoint = torch.load(
    REVE_STATE_PATH,
    map_location="cpu",
    weights_only=False
)

reve_state = (
    reve_checkpoint.get(
        "state_dict",
        reve_checkpoint
    )
)

if "model_state_dict" in reve_checkpoint:

    reve_state = reve_checkpoint[
        "model_state_dict"
    ]

# Handle possible "reve." prefix.
clean_reve_state = {}

for key, value in reve_state.items():

    if key.startswith("reve."):

        clean_reve_state[
            key[len("reve."):]
        ] = value

    else:

        clean_reve_state[
            key
        ] = value


load_result = reve_model.load_state_dict(
    clean_reve_state,
    strict=False
)

assert len(
    load_result.missing_keys
) == 0, (
    f"Missing REVE keys: "
    f"{load_result.missing_keys}"
)

reve_model = reve_model.to(
    DEVICE
)

reve_model.eval()

for parameter in reve_model.parameters():

    parameter.requires_grad = False

print(
    "PASS: official REVE loaded"
)

print(
    "PASS: REVE parameters frozen"
)


# ==============================================================================================================
# 6. LOAD POSITION MODEL
# ==============================================================================================================

print()
print("-" * 110)
print("LOADING REVE POSITION MODEL")
print("-" * 110)

position_model = AutoModel.from_pretrained(
    "brain-bzh/reve-positions",
    trust_remote_code=True,
    torch_dtype=torch.float32
)

position_checkpoint = torch.load(
    POSITION_STATE_PATH,
    map_location="cpu",
    weights_only=False
)

position_state = (
    position_checkpoint.get(
        "state_dict",
        position_checkpoint
    )
)

if "model_state_dict" in position_checkpoint:

    position_state = position_checkpoint[
        "model_state_dict"
    ]

clean_position_state = {}

for key, value in position_state.items():

    if key.startswith("reve."):

        clean_position_state[
            key[len("reve."):]
        ] = value

    else:

        clean_position_state[
            key
        ] = value


position_load = position_model.load_state_dict(
    clean_position_state,
    strict=False
)

assert len(
    position_load.missing_keys
) == 0, (
    f"Missing position keys: "
    f"{position_load.missing_keys}"
)

position_model = position_model.to(
    DEVICE
)

position_model.eval()

for parameter in position_model.parameters():

    parameter.requires_grad = False

print(
    "PASS: position model loaded"
)


# ==============================================================================================================
# 7. VERIFY 22-CHANNEL MONTAGE
# ==============================================================================================================

EEG_CHANNELS_22 = [
    "Fp1", "F3", "C3", "P3", "O1",
    "F7", "T5", "Fc1", "Fc5", "Cp1",
    "Cp5", "F9", "Fz", "Pz", "F4",
    "C4", "P4", "O2", "F8", "T4",
    "T6", "Fc2"
]

assert len(
    EEG_CHANNELS_22
) == 22

positions = position_model(
    EEG_CHANNELS_22
)

if isinstance(
    positions,
    (tuple, list)
):

    positions = positions[0]

positions = torch.as_tensor(
    positions,
    dtype=torch.float32,
    device=DEVICE
)

if positions.ndim == 2:

    positions = positions.unsqueeze(0)

assert positions.ndim == 3

assert positions.shape[1] == 22

print()
print(
    f"PASS: position tensor shape = {tuple(positions.shape)}"
)


# ==============================================================================================================
# 8. VERIFY SELECTED BLOCKS
# ==============================================================================================================

removed_blocks = [
    int(value.strip().replace("B", ""))
    for value in str(
        job["removed_blocks"]
    ).split(",")
    if value.strip()
]

assert len(
    removed_blocks
) == REMOVE_COUNT

assert len(
    set(removed_blocks)
) == REMOVE_COUNT

assert all(
    1 <= block <= 22
    for block in removed_blocks
)

print()
print(
    "Removed blocks:"
)

print(
    removed_blocks
)

print(
    f"PASS: exactly {REMOVE_COUNT} blocks selected"
)


# ==============================================================================================================
# 9. FEATURE EXTRACTION FUNCTION
# ==============================================================================================================

def extract_features(
    eeg_numpy,
    removed_blocks
):

    eeg = torch.from_numpy(
        eeg_numpy
    ).float().to(
        DEVICE
    )

    # Source format:
    #   (B,1000,22)
    #
    # REVE expects:
    #   (B,22,1000)

    eeg = eeg.transpose(
        1,
        2
    )

    original_layers = (
        reve_model.transformer.layers
    )

    try:

        layers = list(
            original_layers
        )

        for block_number in removed_blocks:

            index = block_number - 1

            assert 0 <= index < 22

            layers[index] = nn.ModuleList([
                nn.Identity(),
                nn.Identity()
            ])

        reve_model.transformer.layers = (
            nn.ModuleList(
                layers
            )
        )

        with torch.no_grad():

            output = reve_model(
                eeg,
                positions
            )

        assert torch.is_tensor(
            output
        )

        assert output.ndim == 4

        assert output.shape[1] == 22

        assert output.shape[2] == 5

        assert output.shape[3] == 512

        # (B,22,5,512)
        #
        # Mean across five patches.
        x = output.mean(
            dim=2
        )

        # (B,22,512)
        channel_mean = x.mean(
            dim=1
        )

        channel_std = x.std(
            dim=1,
            unbiased=False
        )

        features = torch.cat(
            [
                channel_mean,
                channel_std
            ],
            dim=1
        )

        assert features.shape[1] == 1024

        assert torch.isfinite(
            features
        ).all()

        return features.cpu().numpy().astype(
            np.float32
        )

    finally:

        reve_model.transformer.layers = (
            original_layers
        )


# ==============================================================================================================
# 10. TEST FEATURE EXTRACTION
# ==============================================================================================================

print()
print("-" * 110)
print("REVE FORWARD PASS")
print("-" * 110)

# Use a small batch first.
smoke_X = X_test[:2]

smoke_features = extract_features(
    smoke_X,
    removed_blocks
)

assert smoke_features.shape == (
    2,
    1024
)

assert np.isfinite(
    smoke_features
).all()

print(
    f"PASS: pruned REVE output features = {smoke_features.shape}"
)

print(
    "PASS: features finite"
)

print(
    "PASS: REVE architecture restored after pruning"
)


# ==============================================================================================================
# 11. LOAD INNER TRAIN / VALIDATION SUBJECTS
# ==============================================================================================================

remaining_subjects = sorted([
    subject
    for subject in [
        "PN00", "PN01", "PN06", "PN07",
        "PN09", "PN10", "PN11", "PN12",
        "PN13", "PN14", "PN16", "PN17"
    ]
    if subject != TEST_SUBJECT
])

assert len(
    remaining_subjects
) == 11

VALIDATION_SUBJECT = remaining_subjects[0]

TRAIN_SUBJECTS = remaining_subjects[1:]

assert len(
    TRAIN_SUBJECTS
) == 10

assert TEST_SUBJECT not in TRAIN_SUBJECTS

assert TEST_SUBJECT != VALIDATION_SUBJECT

print()
print(
    "Inner split:"
)

print(
    "Train:",
    TRAIN_SUBJECTS
)

print(
    "Validation:",
    VALIDATION_SUBJECT
)

print(
    "Outer test:",
    TEST_SUBJECT
)

print(
    "PASS: no outer-test leakage"
)


# ==============================================================================================================
# 12. LOAD TRAIN / VALIDATION DATA
# ==============================================================================================================

def load_subject(
    subject
):

    path = os.path.join(
        SOURCE_DIR,
        f"{subject}_epochs.npz"
    )

    assert os.path.exists(
        path
    )

    with np.load(path) as data:

        X = np.asarray(
            data["X"],
            dtype=np.float32
        )

        y = np.asarray(
            data["y"],
            dtype=np.int64
        )

    assert X.ndim == 3
    assert X.shape[1:] == (
        1000,
        22
    )

    assert len(X) == len(y)

    assert np.isfinite(
        X
    ).all()

    return X, y


train_X_parts = []
train_y_parts = []

for subject in TRAIN_SUBJECTS:

    X_part, y_part = load_subject(
        subject
    )

    train_X_parts.append(
        X_part
    )

    train_y_parts.append(
        y_part
    )

train_X = np.concatenate(
    train_X_parts,
    axis=0
)

train_y = np.concatenate(
    train_y_parts,
    axis=0
)

val_X, val_y = load_subject(
    VALIDATION_SUBJECT
)

assert set(
    np.unique(train_y)
) == {0, 1}

assert set(
    np.unique(val_y)
) == {0, 1}

print()
print(
    f"PASS: training epochs = {len(train_X)}"
)

print(
    f"PASS: validation epochs = {len(val_X)}"
)

print(
    "PASS: train contains both classes"
)

print(
    "PASS: validation contains both classes"
)


# ==============================================================================================================
# 13. FEATURE EXTRACTION FOR TRAIN / VALIDATION / TEST
# ==============================================================================================================

print()
print("-" * 110)
print("EXTRACTING REPRESENTATIONS")
print("-" * 110)

def extract_all(
    X,
    batch_size=32
):

    chunks = []

    for start in range(
        0,
        len(X),
        batch_size
    ):

        end = min(
            start + batch_size,
            len(X)
        )

        chunk = extract_features(
            X[start:end],
            removed_blocks
        )

        chunks.append(
            chunk
        )

    features = np.concatenate(
        chunks,
        axis=0
    )

    assert features.shape == (
        len(X),
        1024
    )

    assert np.isfinite(
        features
    ).all()

    return features


train_features = extract_all(
    train_X
)

val_features = extract_all(
    val_X
)

test_features = extract_all(
    X_test
)

assert train_features.shape == (
    len(train_y),
    1024
)

assert val_features.shape == (
    len(val_y),
    1024
)

assert test_features.shape == (
    len(y_test),
    1024
)

print(
    "PASS: train features:",
    train_features.shape
)

print(
    "PASS: validation features:",
    val_features.shape
)

print(
    "PASS: test features:",
    test_features.shape
)


# ==============================================================================================================
# 14. RELEASE EEG ARRAYS FROM GPU MEMORY / RAM
# ==============================================================================================================

del train_X_parts
del train_y_parts
del train_X
del val_X

if torch.cuda.is_available():

    torch.cuda.empty_cache()


# ==============================================================================================================
# 15. CLASSIFIER
# ==============================================================================================================

class SienaClassifier(
    nn.Module
):

    def __init__(self):

        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(
                1024,
                512
            ),
            nn.ReLU(),
            nn.Dropout(
                0.2
            ),
            nn.Linear(
                512,
                128
            ),
            nn.ReLU(),
            nn.Dropout(
                0.2
            ),
            nn.Linear(
                128,
                2
            )
        )

    def forward(
        self,
        x
    ):

        return self.network(
            x
        )


classifier = SienaClassifier().to(
    DEVICE
)

parameter_count = sum(
    parameter.numel()
    for parameter in classifier.parameters()
)

assert parameter_count == 590722

print()
print(
    f"PASS: classifier parameters = {parameter_count}"
)


# ==============================================================================================================
# 16. DATALOADERS
# ==============================================================================================================

train_dataset = TensorDataset(
    torch.from_numpy(
        train_features
    ),
    torch.from_numpy(
        train_y
    )
)

val_dataset = TensorDataset(
    torch.from_numpy(
        val_features
    ),
    torch.from_numpy(
        val_y
    )
)

test_tensor = torch.from_numpy(
    test_features
).float().to(
    DEVICE
)

train_loader = DataLoader(
    train_dataset,
    batch_size=CLASSIFIER_BATCH_SIZE,
    shuffle=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=CLASSIFIER_BATCH_SIZE,
    shuffle=False
)


# ==============================================================================================================
# 17. TRAINING
# ==============================================================================================================

criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.Adam(
    classifier.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

best_val_loss = float(
    "inf"
)

best_state = None

epochs_without_improvement = 0

print()
print("-" * 110)
print("CLASSIFIER TRAINING")
print("-" * 110)

for epoch in range(
    1,
    MAX_EPOCHS + 1
):

    classifier.train()

    running_loss = 0.0
    sample_count = 0

    for batch_features, batch_labels in train_loader:

        batch_features = batch_features.to(
            DEVICE
        ).float()

        batch_labels = batch_labels.to(
            DEVICE
        ).long()

        optimizer.zero_grad(
            set_to_none=True
        )

        logits = classifier(
            batch_features
        )

        loss = criterion(
            logits,
            batch_labels
        )

        assert torch.isfinite(
            loss
        )

        loss.backward()

        optimizer.step()

        running_loss += (
            loss.item()
            * len(batch_labels)
        )

        sample_count += len(
            batch_labels
        )

    train_loss = (
        running_loss
        / sample_count
    )

    classifier.eval()

    validation_loss = 0.0
    validation_count = 0

    with torch.no_grad():

        for batch_features, batch_labels in val_loader:

            batch_features = batch_features.to(
                DEVICE
            ).float()

            batch_labels = batch_labels.to(
                DEVICE
            ).long()

            logits = classifier(
                batch_features
            )

            loss = criterion(
                logits,
                batch_labels
            )

            validation_loss += (
                loss.item()
                * len(batch_labels)
            )

            validation_count += len(
                batch_labels
            )

    val_loss = (
        validation_loss
        / validation_count
    )

    if val_loss < best_val_loss:

        best_val_loss = val_loss

        best_state = {
            key: value.detach().cpu().clone()
            for key, value
            in classifier.state_dict().items()
        }

        epochs_without_improvement = 0

    else:

        epochs_without_improvement += 1

    if (
        epoch == 1
        or epoch % 10 == 0
        or epochs_without_improvement == 0
    ):

        print(
            f"Epoch {epoch:03d} | "
            f"train_loss={train_loss:.6f} | "
            f"val_loss={val_loss:.6f}"
        )

    if epochs_without_improvement >= PATIENCE:

        print(
            f"Early stopping at epoch {epoch}"
        )

        break


assert best_state is not None

classifier.load_state_dict(
    best_state
)

print()
print(
    f"PASS: classifier trained"
)

print(
    f"Best validation loss = {best_val_loss:.6f}"
)


# ==============================================================================================================
# 18. HELD-OUT TEST PREDICTION
# ==============================================================================================================

classifier.eval()

with torch.no_grad():

    test_logits = classifier(
        test_tensor
    )

    test_probabilities = torch.softmax(
        test_logits,
        dim=1
    )[:, 1].cpu().numpy()

    test_predictions = (
        test_logits
        .argmax(
            dim=1
        )
        .cpu()
        .numpy()
    )

assert len(
    test_probabilities
) == len(y_test)

assert np.isfinite(
    test_probabilities
).all()

print()
print(
    "PASS: held-out test prediction completed"
)


# ==============================================================================================================
# 19. METRICS
# ==============================================================================================================

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

accuracy = accuracy_score(
    y_test,
    test_predictions
)

balanced_accuracy = balanced_accuracy_score(
    y_test,
    test_predictions
)

f1 = f1_score(
    y_test,
    test_predictions,
    zero_division=0
)

if len(
    np.unique(y_test)
) == 2:

    auroc = roc_auc_score(
        y_test,
        test_probabilities
    )

    auprc = average_precision_score(
        y_test,
        test_probabilities
    )

else:

    auroc = np.nan
    auprc = np.nan


assert np.isfinite(
    [
        accuracy,
        balanced_accuracy,
        f1
    ]
).all()

print()
print("-" * 110)
print("HELD-OUT TEST METRICS")
print("-" * 110)

print(
    f"Accuracy           : {accuracy:.6f}"
)

print(
    f"Balanced accuracy  : {balanced_accuracy:.6f}"
)

print(
    f"F1                 : {f1:.6f}"
)

print(
    f"AUROC              : {auroc:.6f}"
)

print(
    f"AUPRC              : {auprc:.6f}"
)


# ==============================================================================================================
# 20. SAVE REPRESENTATIVE CHECKPOINT
# ==============================================================================================================

representative_checkpoint_path = os.path.join(
    CHECKPOINT_DIR,
    f"job_{JOB_ID:04d}_REPRESENTATIVE.pt"
)

representative_metric_path = os.path.join(
    METRIC_DIR,
    f"job_{JOB_ID:04d}_REPRESENTATIVE.json"
)

checkpoint_payload = {
    "job_id": JOB_ID,
    "method": METHOD,
    "pruning_percent": PRUNING_PERCENT,
    "remove_count": REMOVE_COUNT,
    "removed_blocks": removed_blocks,
    "subject": TEST_SUBJECT,
    "classifier_seed": CLASSIFIER_SEED,
    "best_val_loss": float(
        best_val_loss
    ),
    "classifier_state_dict": {
        key: value.cpu()
        for key, value
        in classifier.state_dict().items()
    },
}

torch.save(
    checkpoint_payload,
    representative_checkpoint_path
)

metrics_payload = {
    "job_id": JOB_ID,
    "method": METHOD,
    "pruning_percent": PRUNING_PERCENT,
    "remove_count": REMOVE_COUNT,
    "removed_blocks": removed_blocks,
    "subject": TEST_SUBJECT,
    "classifier_seed": CLASSIFIER_SEED,
    "accuracy": float(accuracy),
    "balanced_accuracy": float(
        balanced_accuracy
    ),
    "f1": float(f1),
    "auroc": (
        float(auroc)
        if np.isfinite(auroc)
        else None
    ),
    "auprc": (
        float(auprc)
        if np.isfinite(auprc)
        else None
    ),
    "best_val_loss": float(
        best_val_loss
    ),
}

with open(
    representative_metric_path,
    "w"
) as handle:

    json.dump(
        metrics_payload,
        handle,
        indent=2
    )


assert os.path.exists(
    representative_checkpoint_path
)

assert os.path.exists(
    representative_metric_path
)

print()
print(
    "PASS: representative checkpoint saved"
)

print(
    representative_checkpoint_path
)

print(
    "PASS: representative metrics saved"
)

print(
    representative_metric_path
)


# ==============================================================================================================
# 21. FINAL ARCHITECTURE RESTORATION CHECK
# ==============================================================================================================

assert len(
    reve_model.transformer.layers
) == 22

print()
print("=" * 110)
print("CELL 13G — REPRESENTATIVE JOB COMPLETE")
print("=" * 110)

print(
    "PASS: official REVE"
)

print(
    "PASS: structured block pruning"
)

print(
    "PASS: 1024-D representation"
)

print(
    "PASS: inner train/validation"
)

print(
    "PASS: held-out LOSO test"
)

print(
    "PASS: classifier training"
)

print(
    "PASS: metrics"
)

print(
    "PASS: checkpoint"
)

print(
    "PASS: REVE architecture restored"
)

print()
print(
    "ONLY ONE JOB WAS EXECUTED."
)

print(
    "THE FULL 2760-JOB EXPERIMENT HAS NOT STARTED."
)

print("=" * 110)
```

# 21. AGGREGATION FUNCTION
# ============================================================

def make_aggregate_table(
    metric
):

    grouped = (
        final_results
        .groupby(
            [
                "method",
                "pruning_percent"
            ]
        )[metric]
        .agg(
            [
                "mean",
                "std",
                "count"
            ]
        )
        .reset_index()
    )


    assert (
        grouped["count"] == 60
    ).all(), (
        f"Invalid count for metric: {metric}"
    )


    # Convert proportions to percentages.

    grouped["mean"] = (
        grouped["mean"]
        * 100
    )

    grouped["std"] = (
        grouped["std"]
        * 100
    )


    # Mean ± SD.

    grouped["mean_sd"] = (
        grouped["mean"]
        .map(
            lambda x: f"{x:.2f}"
        )
        +
        " ± "
        +
        grouped["std"]
        .map(
            lambda x: f"{x:.2f}"
        )
    )


    table = (
        grouped
        .pivot(
            index="method",
            columns="pruning_percent",
            values="mean_sd"
        )
        .reindex(
            index=METHODS,
            columns=PRUNING_LEVELS
        )
    )


    table.columns = [
        f"{x}%"
        for x in table.columns
    ]


    # No missing table cells.

    assert not table.isna().any().any(), (
        f"Missing aggregation cells "
        f"for metric: {metric}"
    )


    return (
        table,
        grouped
    )

for metric in METRICS:

```python
# ============================================================
# CELL 13I-DIAGNOSTIC
# ACTUAL STRUCTURE — NO PREDEFINED COUNTS
# ============================================================

import os
import json
import pandas as pd
import numpy as np


print("=" * 110)
print("CELL 13I — ACTUAL RESULT STRUCTURE DIAGNOSTIC")
print("=" * 110)


# ============================================================
# 1. PATH
# ============================================================

BASE_DIR = "/content/drive/MyDrive/Seinna_REVE_LOSO"

METRICS_DIR = os.path.join(
    BASE_DIR,
    "novelty3_pruning",
    "evaluation_results",
    "metrics"
)

assert os.path.isdir(
    METRICS_DIR
)

print("\nMetrics directory:")
print(METRICS_DIR)


# ============================================================
# 2. LOAD ALL JSON FILES
# ============================================================

json_files = sorted(
    f
    for f in os.listdir(METRICS_DIR)
    if f.endswith(".json")
)

print(
    "\nJSON files found:",
    len(json_files)
)


records = []

for filename in json_files:

    path = os.path.join(
        METRICS_DIR,
        filename
    )

    with open(
        path,
        "r"
    ) as f:

        record = json.load(f)

    record["_filename"] = filename

    records.append(
        record
    )


df = pd.DataFrame(
    records
)

print(
    "Records loaded:",
    len(df)
)


# ============================================================
# 3. REMOVE REPRESENTATIVE RECORD ONLY
# ============================================================

representative_mask = pd.Series(
    False,
    index=df.index
)

if "imported_from_representative" in df.columns:

    representative_mask = (
        df["imported_from_representative"]
        .fillna(False)
        .astype(bool)
    )

representative_records = df.loc[
    representative_mask
].copy()

production_df = df.loc[
    ~representative_mask
].copy()


print(
    "\nRepresentative records:",
    len(representative_records)
)

print(
    "Production records:",
    len(production_df)
)


if len(representative_records) > 0:

    print(
        "\nRepresentative record(s):"
    )

    print(
        representative_records[
            [
                "_filename",
                "job_id",
                "baseline",
                "method",
                "pruning_percent",
                "subject",
                "classifier_seed"
            ]
        ].to_string(
            index=False
        )
    )


# ============================================================
# 4. BASIC STRUCTURE
# ============================================================

production_df["baseline_bool"] = (
    production_df["baseline"]
    .astype(bool)
)

production_df["pruning_percent_num"] = pd.to_numeric(
    production_df["pruning_percent"],
    errors="coerce"
)

production_df["classifier_seed_num"] = pd.to_numeric(
    production_df["classifier_seed"],
    errors="coerce"
)

production_df["subject_str"] = (
    production_df["subject"]
    .astype(str)
    .str.strip()
)

production_df["method_str"] = (
    production_df["method"]
    .astype(str)
    .str.strip()
)


# ============================================================
# 5. ACTUAL BASELINE / PRUNING COUNTS
# ============================================================

actual_baseline = production_df.loc[
    production_df["baseline_bool"]
].copy()

actual_pruning = production_df.loc[
    ~production_df["baseline_bool"]
].copy()


print("\n")
print("=" * 110)
print("ACTUAL BASELINE / PRUNING STRUCTURE")
print("=" * 110)

print(
    "\nActual baseline records:",
    len(actual_baseline)
)

print(
    "Actual pruning records:",
    len(actual_pruning)
)

print(
    "Total production records:",
    len(production_df)
)


# ============================================================
# 6. ACTUAL BASELINE DETAILS
# ============================================================

print("\n")
print("=" * 110)
print("BASELINE RECORDS — ACTUAL")
print("=" * 110)

baseline_columns = [
    "_filename",
    "job_id",
    "method_str",
    "pruning_percent_num",
    "remove_count",
    "keep_count",
    "subject_str",
    "classifier_seed_num",
    "accuracy",
    "balanced_accuracy",
    "f1",
    "auroc"
]

print(
    actual_baseline[
        baseline_columns
    ].sort_values(
        [
            "subject_str",
            "classifier_seed_num",
            "job_id"
        ]
    ).to_string(
        index=False
    )
)


# ============================================================
# 7. ACTUAL BASELINE SUBJECT × SEED COUNTS
# ============================================================

baseline_subject_seed = (
    actual_baseline
    .groupby(
        [
            "subject_str",
            "classifier_seed_num"
        ]
    )
    .size()
    .reset_index(
        name="count"
    )
)


print("\n")
print("=" * 110)
print("BASELINE SUBJECT × SEED COUNTS")
print("=" * 110)

print(
    baseline_subject_seed.to_string(
        index=False
    )
)


# ============================================================
# 8. FIND BASELINE DUPLICATES
# ============================================================

baseline_duplicates = (
    actual_baseline[
        [
            "subject_str",
            "classifier_seed_num"
        ]
    ]
    .value_counts()
    .reset_index(
        name="count"
    )
)

baseline_duplicates = baseline_duplicates.loc[
    baseline_duplicates["count"] > 1
]


print("\n")
print("=" * 110)
print("BASELINE DUPLICATES")
print("=" * 110)

if len(baseline_duplicates) == 0:

    print(
        "NONE"
    )

else:

    print(
        baseline_duplicates.to_string(
            index=False
        )
    )

    duplicate_keys = baseline_duplicates[
        [
            "subject_str",
            "classifier_seed_num"
        ]
    ]

    duplicate_records = actual_baseline.merge(
        duplicate_keys,
        on=[
            "subject_str",
            "classifier_seed_num"
        ],
        how="inner"
    )

    print(
        "\nDuplicate baseline records:"
    )

    print(
        duplicate_records[
            baseline_columns
        ].sort_values(
            [
                "subject_str",
                "classifier_seed_num",
                "job_id"
            ]
        ).to_string(
            index=False
        )
    )


# ============================================================
# 9. ACTUAL PRUNING METHODS
# ============================================================

print("\n")
print("=" * 110)
print("ACTUAL METHODS")
print("=" * 110)

print(
    sorted(
        actual_pruning[
            "method_str"
        ].unique()
    )
)


# ============================================================
# 10. ACTUAL PRUNING LEVELS
# ============================================================

print("\n")
print("=" * 110)
print("ACTUAL PRUNING LEVELS")
print("=" * 110)

print(
    sorted(
        actual_pruning[
            "pruning_percent_num"
        ].unique()
    )
)


# ============================================================
# 11. ACTUAL SUBJECTS
# ============================================================

print("\n")
print("=" * 110)
print("ACTUAL SUBJECTS")
print("=" * 110)

print(
    sorted(
        actual_pruning[
            "subject_str"
        ].unique()
    )
)

print(
    "\nNumber of unique subjects:",
    actual_pruning[
        "subject_str"
    ].nunique()
)


# ============================================================
# 12. ACTUAL CLASSIFIER SEEDS
# ============================================================

print("\n")
print("=" * 110)
print("ACTUAL CLASSIFIER SEEDS")
print("=" * 110)

print(
    sorted(
        actual_pruning[
            "classifier_seed_num"
        ].unique()
    )
)

print(
    "\nNumber of unique seeds:",
    actual_pruning[
        "classifier_seed_num"
    ].nunique()
)


# ============================================================
# 13. METHOD × PRUNING COUNTS
# ============================================================

method_pruning_counts = (
    actual_pruning
    .groupby(
        [
            "method_str",
            "pruning_percent_num"
        ]
    )
    .size()
    .reset_index(
        name="count"
    )
    .sort_values(
        [
            "method_str",
            "pruning_percent_num"
        ]
    )
)


print("\n")
print("=" * 110)
print("ACTUAL METHOD × PRUNING COUNTS")
print("=" * 110)

print(
    method_pruning_counts.to_string(
        index=False
    )
)


# ============================================================
# 14. SUBJECT × SEED UNIQUENESS PER CELL
# ============================================================

cell_structure = (
    actual_pruning
    .groupby(
        [
            "method_str",
            "pruning_percent_num"
        ]
    )
    .agg(
        records=(
            "job_id",
            "size"
        ),
        subjects=(
            "subject_str",
            "nunique"
        ),
        seeds=(
            "classifier_seed_num",
            "nunique"
        )
    )
    .reset_index()
    .sort_values(
        [
            "method_str",
            "pruning_percent_num"
        ]
    )
)


print("\n")
print("=" * 110)
print("CELL STRUCTURE")
print("=" * 110)

print(
    cell_structure.to_string(
        index=False
    )
)


# ============================================================
# 15. JOB-ID DUPLICATES
# ============================================================

job_id_counts = (
    production_df[
        "job_id"
    ]
    .value_counts()
)

duplicate_job_ids = job_id_counts.loc[
    job_id_counts > 1
]


print("\n")
print("=" * 110)
print("DUPLICATE JOB IDs")
print("=" * 110)

if len(duplicate_job_ids) == 0:

    print(
        "NONE"
    )

else:

    print(
        duplicate_job_ids.to_string()
    )


# ============================================================
# 16. MISSING / EXTRA SUBJECT-SEED PAIRS
#    BASED ONLY ON ACTUAL DATA
# ============================================================

actual_subjects = sorted(
    actual_pruning[
        "subject_str"
    ].unique()
)

actual_seeds = sorted(
    actual_pruning[
        "classifier_seed_num"
    ].unique()
)

observed_subject_seed = set(
    zip(
        actual_pruning[
            "subject_str"
        ],
        actual_pruning[
            "classifier_seed_num"
        ]
    )
)

all_possible_subject_seed = {
    (
        subject,
        seed
    )
    for subject in actual_subjects
    for seed in actual_seeds
}

missing_subject_seed = sorted(
    all_possible_subject_seed
    -
    observed_subject_seed
)

print("\n")
print("=" * 110)
print("ACTUAL SUBJECT × SEED COVERAGE")
print("=" * 110)

print(
    "Possible subject × seed pairs:",
    len(all_possible_subject_seed)
)

print(
    "Observed subject × seed pairs:",
    len(observed_subject_seed)
)

print(
    "Missing subject × seed pairs:",
    len(missing_subject_seed)
)

if missing_subject_seed:

    print(
        missing_subject_seed
    )


# ============================================================
# 17. FINAL DIAGNOSTIC
# ============================================================

print("\n")
print("=" * 110)
print("DIAGNOSTIC COMPLETE")
print("=" * 110)

print(
    "\nNothing was modified."
)

print(
    "No evaluation was rerun."
)

print(
    "No records were deleted."
)

print(
    "No predefined experiment counts were assumed."
)

print(
    "\nUse the output above to determine the actual "
    "baseline/pruning structure before aggregation."
)
```

```python
# ============================================================
# CELL 13I — ACCURACY TABLE ONLY
# ============================================================

import os
import json
import glob
import numpy as np
import pandas as pd

METRICS_DIR = "/content/drive/MyDrive/Seinna_REVE_LOSO/novelty3_pruning/evaluation_results/metrics"

# ------------------------------------------------------------
# 1. Load all JSON records
# ------------------------------------------------------------

json_files = sorted(glob.glob(os.path.join(METRICS_DIR, "*.json")))

records = []

for filepath in json_files:
    with open(filepath, "r") as f:
        record = json.load(f)

    record["_filename"] = os.path.basename(filepath)
    records.append(record)

df = pd.DataFrame(records)

# ------------------------------------------------------------
# 2. Exclude representative record(s) only
# ------------------------------------------------------------

if "imported_from_representative" in df.columns:
    representative_mask = (
        df["imported_from_representative"]
        .fillna(False)
        .astype(bool)
    )
else:
    representative_mask = (
        df["_filename"]
        .str.contains("REPRESENTATIVE", case=False, na=False)
    )

df_prod = df.loc[~representative_mask].copy()

# ------------------------------------------------------------
# 3. Identify pruning records
# ------------------------------------------------------------

# Use method field directly where available
if "method" in df_prod.columns:
    df_prod["method_str"] = df_prod["method"].astype(str)
elif "method_str" in df_prod.columns:
    df_prod["method_str"] = df_prod["method_str"].astype(str)
else:
    raise KeyError("No method column found in the metric records.")

# Convert pruning percentage numerically
if "pruning_percent" in df_prod.columns:
    df_prod["pruning_percent_num"] = pd.to_numeric(
        df_prod["pruning_percent"],
        errors="coerce"
    )
elif "pruning_percent_num" in df_prod.columns:
    df_prod["pruning_percent_num"] = pd.to_numeric(
        df_prod["pruning_percent_num"],
        errors="coerce"
    )
else:
    raise KeyError("No pruning_percent column found.")

# Keep only the actual pruning methods and positive pruning levels.
# Methods are discovered from the files rather than hard-coded.
pruning_mask = (
    df_prod["method_str"].ne("Baseline")
    & df_prod["pruning_percent_num"].notna()
    & (df_prod["pruning_percent_num"] > 0)
)

pruning_df = df_prod.loc[pruning_mask].copy()

# ------------------------------------------------------------
# 4. Verify Accuracy exists and is numeric
# ------------------------------------------------------------

if "accuracy" not in pruning_df.columns:
    raise KeyError("The metric records do not contain an 'accuracy' field.")

pruning_df["accuracy_num"] = pd.to_numeric(
    pruning_df["accuracy"],
    errors="coerce"
)

if pruning_df["accuracy_num"].isna().any():
    bad = pruning_df.loc[
        pruning_df["accuracy_num"].isna(),
        ["_filename", "method_str", "pruning_percent_num", "accuracy"]
    ]
    print("Records with invalid accuracy:")
    display(bad)
    raise ValueError("Invalid accuracy values detected.")

# Check accuracy is stored as a fraction.
if pruning_df["accuracy_num"].max() <= 1.0:
    pruning_df["accuracy_percent"] = pruning_df["accuracy_num"] * 100.0
else:
    pruning_df["accuracy_percent"] = pruning_df["accuracy_num"]

# ------------------------------------------------------------
# 5. Dynamically determine actual methods and pruning levels
# ------------------------------------------------------------

methods = sorted(
    pruning_df["method_str"].dropna().unique().tolist()
)

pruning_levels = sorted(
    pruning_df["pruning_percent_num"].dropna().unique().tolist()
)

print("Actual methods:", methods)
print("Actual pruning levels:", pruning_levels)

# ------------------------------------------------------------
# 6. Aggregate Accuracy
# ------------------------------------------------------------

accuracy_table = (
    pruning_df
    .groupby(
        ["pruning_percent_num", "method_str"],
        as_index=False
    )["accuracy_percent"]
    .mean()
    .pivot(
        index="pruning_percent_num",
        columns="method_str",
        values="accuracy_percent"
    )
    .reindex(index=pruning_levels)
    .reindex(columns=methods)
)

# ------------------------------------------------------------
# 7. Format as percentage table
# ------------------------------------------------------------

accuracy_table.index = [
    f"{int(x)}%" if float(x).is_integer() else f"{x:g}%"
    for x in accuracy_table.index
]

accuracy_table.index.name = "Pruning"

accuracy_table = accuracy_table.round(2)

print("\n" + "=" * 100)
print("ACCURACY (%)")
print("=" * 100)

display(accuracy_table)
```

```python
# ============================================================
# CELL 13I — FULL ACCURACY VERIFICATION + RECALCULATION
# ============================================================

import os
import json
import glob
import numpy as np
import pandas as pd

METRICS_DIR = (
    "/content/drive/MyDrive/Seinna_REVE_LOSO/"
    "novelty3_pruning/evaluation_results/metrics"
)

print("=" * 100)
print("FULL ACCURACY VERIFICATION")
print("=" * 100)

# ------------------------------------------------------------
# 1. Load every JSON
# ------------------------------------------------------------

json_files = sorted(glob.glob(os.path.join(METRICS_DIR, "*.json")))

print(f"\nJSON files found: {len(json_files)}")

records = []

for filepath in json_files:
    with open(filepath, "r") as f:
        r = json.load(f)

    r["_filename"] = os.path.basename(filepath)
    records.append(r)

df = pd.DataFrame(records)

print(f"Records loaded: {len(df)}")

# ------------------------------------------------------------
# 2. Remove representative record ONLY
# ------------------------------------------------------------

if "imported_from_representative" in df.columns:
    rep_mask = (
        df["imported_from_representative"]
        .fillna(False)
        .astype(bool)
    )
else:
    rep_mask = df["_filename"].str.contains(
        "REPRESENTATIVE",
        case=False,
        na=False
    )

representative_df = df.loc[rep_mask].copy()
production_df = df.loc[~rep_mask].copy()

print(f"Representative records excluded: {len(representative_df)}")
print(f"Production records: {len(production_df)}")

# ------------------------------------------------------------
# 3. Verify required columns
# ------------------------------------------------------------

required_columns = [
    "method",
    "pruning_percent",
    "subject",
    "classifier_seed",
    "accuracy",
    "completed",
    "reve_frozen",
    "outer_test_used_for_early_stopping"
]

missing_columns = [
    c for c in required_columns
    if c not in production_df.columns
]

if missing_columns:
    raise KeyError(
        f"Missing required columns: {missing_columns}"
    )

# ------------------------------------------------------------
# 4. Keep ONLY actual pruning evaluations
# ------------------------------------------------------------

production_df["method"] = (
    production_df["method"]
    .astype(str)
    .str.strip()
)

production_df["pruning_percent_num"] = pd.to_numeric(
    production_df["pruning_percent"],
    errors="coerce"
)

pruning_df = production_df[
    (production_df["method"] != "Baseline") &
    production_df["pruning_percent_num"].notna() &
    (production_df["pruning_percent_num"] > 0)
].copy()

print(f"\nPruning records: {len(pruning_df)}")

# ------------------------------------------------------------
# 5. Verify completion status
# ------------------------------------------------------------

completed_numeric = (
    pruning_df["completed"]
    .fillna(False)
    .astype(bool)
)

print(f"Completed pruning records: {completed_numeric.sum()}")
print(
    f"Incomplete pruning records: "
    f"{(~completed_numeric).sum()}"
)

if (~completed_numeric).any():
    print("\nWARNING — incomplete records:")
    display(
        pruning_df.loc[
            ~completed_numeric,
            [
                "_filename",
                "method",
                "pruning_percent_num",
                "subject",
                "classifier_seed"
            ]
        ]
    )

# ------------------------------------------------------------
# 6. Verify REVE was frozen
# ------------------------------------------------------------

reve_frozen_values = (
    pruning_df["reve_frozen"]
    .dropna()
    .astype(bool)
    .unique()
)

print("\nREVE frozen values:", reve_frozen_values)

if len(reve_frozen_values) != 1 or reve_frozen_values[0] is not True:
    print(
        "WARNING: REVE-frozen status is not uniformly True."
    )

# ------------------------------------------------------------
# 7. Verify outer test subject was NOT used for early stopping
# ------------------------------------------------------------

early_stop_values = (
    pruning_df["outer_test_used_for_early_stopping"]
    .dropna()
    .astype(bool)
    .unique()
)

print(
    "Outer test used for early stopping:",
    early_stop_values
)

# ------------------------------------------------------------
# 8. Accuracy validation
# ------------------------------------------------------------

pruning_df["accuracy_raw"] = pd.to_numeric(
    pruning_df["accuracy"],
    errors="coerce"
)

invalid_accuracy = (
    pruning_df["accuracy_raw"].isna() |
    ~np.isfinite(pruning_df["accuracy_raw"])
)

if invalid_accuracy.any():
    print("\nINVALID ACCURACY RECORDS:")
    display(
        pruning_df.loc[
            invalid_accuracy,
            [
                "_filename",
                "method",
                "pruning_percent_num",
                "subject",
                "classifier_seed",
                "accuracy"
            ]
        ]
    )
    raise ValueError(
        "Invalid accuracy values found."
    )

# Determine whether stored accuracy is fraction or percentage
acc_max = pruning_df["accuracy_raw"].max()

if acc_max <= 1.0:
    pruning_df["accuracy_percent"] = (
        pruning_df["accuracy_raw"] * 100.0
    )
    accuracy_storage = "fraction [0,1]"
else:
    pruning_df["accuracy_percent"] = (
        pruning_df["accuracy_raw"]
    )
    accuracy_storage = "percentage [0,100]"

print(
    "\nAccuracy storage detected:",
    accuracy_storage
)

# ------------------------------------------------------------
# 9. Discover actual methods / levels / subjects / seeds
# ------------------------------------------------------------

actual_methods = sorted(
    pruning_df["method"].unique().tolist()
)

actual_levels = sorted(
    pruning_df["pruning_percent_num"].unique().tolist()
)

actual_subjects = sorted(
    pruning_df["subject"].dropna().astype(str).unique().tolist()
)

actual_seeds = sorted(
    pd.to_numeric(
        pruning_df["classifier_seed"],
        errors="coerce"
    ).dropna().astype(int).unique().tolist()
)

print("\nActual methods:")
print(actual_methods)

print("\nActual pruning levels:")
print(actual_levels)

print("\nActual subjects:")
print(actual_subjects)

print("\nActual classifier seeds:")
print(actual_seeds)

# ------------------------------------------------------------
# 10. Check duplicate evaluation keys
# ------------------------------------------------------------

key_columns = [
    "method",
    "pruning_percent_num",
    "subject",
    "classifier_seed"
]

duplicate_mask = pruning_df.duplicated(
    subset=key_columns,
    keep=False
)

print(
    "\nDuplicate method × pruning × subject × seed records:",
    duplicate_mask.sum()
)

if duplicate_mask.any():
    print("\nDUPLICATE RECORDS:")
    display(
        pruning_df.loc[
            duplicate_mask,
            key_columns + [
                "_filename",
                "accuracy_percent"
            ]
        ].sort_values(key_columns)
    )

# ------------------------------------------------------------
# 11. Count every method × pruning cell
# ------------------------------------------------------------

counts = (
    pruning_df
    .groupby(
        ["method", "pruning_percent_num"]
    )
    .size()
    .unstack(fill_value=0)
)

print("\n" + "=" * 100)
print("NUMBER OF RECORDS PER METHOD × PRUNING")
print("=" * 100)

display(counts)

# ------------------------------------------------------------
# 12. Identify missing subject × seed combinations
#     separately for EVERY method × pruning level
# ------------------------------------------------------------

missing_rows = []

for method in actual_methods:

    method_df = pruning_df[
        pruning_df["method"] == method
    ]

    for level in actual_levels:

        cell = method_df[
            method_df["pruning_percent_num"] == level
        ]

        observed_pairs = set(
            zip(
                cell["subject"].astype(str),
                pd.to_numeric(
                    cell["classifier_seed"],
                    errors="coerce"
                ).astype("Int64")
            )
        )

        expected_pairs = set(
            (subject, seed)
            for subject in actual_subjects
            for seed in actual_seeds
        )

        missing_pairs = sorted(
            expected_pairs - observed_pairs
        )

        for subject, seed in missing_pairs:
            missing_rows.append({
                "method": method,
                "pruning_percent": level,
                "missing_subject": subject,
                "missing_seed": seed
            })

missing_df = pd.DataFrame(missing_rows)

print("\n" + "=" * 100)
print("MISSING EVALUATIONS")
print("=" * 100)

if len(missing_df) == 0:
    print("No missing subject × seed combinations.")
else:
    display(missing_df)

# ------------------------------------------------------------
# 13. Accuracy mean + SD
# ------------------------------------------------------------

accuracy_summary = (
    pruning_df
    .groupby(
        ["pruning_percent_num", "method"]
    )["accuracy_percent"]
    .agg(
        n="count",
        mean="mean",
        sd="std"
    )
    .reset_index()
)

# ------------------------------------------------------------
# 14. Create mean ± SD table
# ------------------------------------------------------------

accuracy_summary["mean_sd"] = (
    accuracy_summary["mean"].map(
        lambda x: f"{x:.2f}"
    )
    + " ± "
    + accuracy_summary["sd"].fillna(0).map(
        lambda x: f"{x:.2f}"
    )
    + " (n="
    + accuracy_summary["n"].astype(str)
    + ")"
)

accuracy_table = (
    accuracy_summary
    .pivot(
        index="pruning_percent_num",
        columns="method",
        values="mean_sd"
    )
    .reindex(
        index=actual_levels,
        columns=actual_methods
    )
)

accuracy_table.index = [
    f"{int(x)}%"
    if float(x).is_integer()
    else f"{x:g}%"
    for x in accuracy_table.index
]

accuracy_table.index.name = "Pruning"

print("\n" + "=" * 100)
print("ACCURACY — MEAN ± SD (%)")
print("=" * 100)

display(accuracy_table)

# ------------------------------------------------------------
# 15. Also display raw numerical means
# ------------------------------------------------------------

mean_table = (
    accuracy_summary
    .pivot(
        index="pruning_percent_num",
        columns="method",
        values="mean"
    )
    .reindex(
        index=actual_levels,
        columns=actual_methods
    )
    .round(4)
)

mean_table.index = [
    f"{int(x)}%"
    if float(x).is_integer()
    else f"{x:g}%"
    for x in mean_table.index
]

mean_table.index.name = "Pruning"

print("\n" + "=" * 100)
print("RAW MEAN ACCURACY (%)")
print("=" * 100)

display(mean_table)

# ------------------------------------------------------------
# 16. Show distribution of accuracy at every pruning level
# ------------------------------------------------------------

level_summary = (
    pruning_df
    .groupby("pruning_percent_num")["accuracy_percent"]
    .agg(
        n="count",
        mean="mean",
        sd="std",
        minimum="min",
        maximum="max"
    )
    .reindex(actual_levels)
)

print("\n" + "=" * 100)
print("OVERALL ACCURACY DISTRIBUTION BY PRUNING LEVEL")
print("=" * 100)

display(level_summary.round(4))

# ------------------------------------------------------------
# 17. Save verified tables
# ------------------------------------------------------------

output_dir = (
    "/content/drive/MyDrive/Seinna_REVE_LOSO/"
    "novelty3_pruning/evaluation_results"
)

os.makedirs(output_dir, exist_ok=True)

accuracy_table.to_csv(
    os.path.join(
        output_dir,
        "accuracy_table_verified_mean_sd.csv"
    )
)

mean_table.to_csv(
    os.path.join(
        output_dir,
        "accuracy_table_verified_mean.csv"
    )
)

accuracy_summary.to_csv(
    os.path.join(
        output_dir,
        "accuracy_summary_verified.csv"
    ),
    index=False
)

if len(missing_df) > 0:
    missing_df.to_csv(
        os.path.join(
            output_dir,
            "missing_pruning_evaluations.csv"
        ),
        index=False
    )

print("\nSaved verified accuracy tables to:")
print(output_dir)

print("\nDONE.")
```
